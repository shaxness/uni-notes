(qui breve approfondimento sulla questione privacy, sicurezza e controllo dei dati nel cloud)

Utilizzare il cloud significa affidare dati ed elaborazioni a un'infrastruttura gestita da qualcun altro. Oltre ai vantaggi tecnici quindi **bisogna considerare dove vengono conservati i dati, chi può accedervi e per quali finalità possono essere usati**.

Si distingue:
- Privacy == riguarda il trattamento dei dati personali e cosa possono fare le aziende con essi. 
- Sicurezza == riguarda la protezione da accessi non autorizzati
- Conformità normativa == riguarda il rispetto delle norme vigenti

Quindi un sistema può potenzialmente essere molto sicuro, ma usare i tuoi dati in modo non conforme alle leggi sulla privacy...

Si parla di **Sovranità dei dati** quando si fa riferimento al controllo e alla gestione dei dati in conformità con le leggi del paese in cui risiedono. Nel cloud quindi scegliere una regione nella quale conservare i dati ha quindi conseguenze che vanno oltre la latenza: da essa dipendono anche le leggi che regolano la privacy e la sicurezza dei dati. 

Ciò è importante soprattutto per servizi pubblici che conservano informazioni delicate sui cittadini, come gli ospedali; è successo che dati di questo tipo siano stati venduti ad esempio alle assicurazioni sanitarie senza il consenso degli interessati, violando le normative sulla privacy.  
A questo proposito sono annesse le iniziative per realizzare infrastrutture cloud nazionali.

Le frasi sui dirigenti di Sun e Google mostrano un evidente atteggiamento di disinteresse verso la privacy degli utenti, considerando la perdità della privacy come una conseguenza quasi inevitabile all'uso dei servizi digitali. Ringraziamo Dio per le garanzie europee (GDPR).

Quando si inseriscono dati in un servizio, **dobbiamo consideerare ciò che il gestore può farne**. Ciò vale anche e soprattutto per i servizi di AI generativa.
