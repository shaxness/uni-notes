Il Cloud Computing è un modello che permette di utilizzare risorse informatiche attraverso la rete, senza dover possedere e gestire direttamente tutta l'infrastruttura fisica necessaria.

La parola cloud deriva dal simbolo usato nei diagrammi per rappresentare Internet, nascondendo i dettagli della rete sottostante. Il termine computing comprende invece le tre attività di Computation, Storage e Coordination (coordinare componenti e attività, es. la comunicazione tra servizi). La capacità informatica non è più legata solo a un server gestito dall'utente, ma diventa una risorsa accessibile tramite la rete.

Le risorse fisiche utilizzate sono ospitate nei data center, e il cloud le rende disponibili agli utenti tramite servizi nascondendogli molti dettagli nella loro gestione (es. un'azienda può richiedere un server virtuale senza dover acquistare, installare e mantenere l'hardware fisico).

Ora un esempio per vedere un caso in cui l'applicazione cloud può essere migliore rispetto alla tradizionale: immaginiamo di realizzare un servizio di riproduzione video per molti utenti.
1. Tradizionale: una possibilità è utilizzare un'unica applicazione con più thread: i diversi core del processore lavorano in parallelo per gestire le richieste degli utenti. Tuttavia tre difficoltà: 
    - Complessità della programmazione: i thread condividono memoria e risorse --> difficoltà nel coordinare per evitare race condition e deadlock
    - Minore isolamento dei guasti: un errore fatale per un thread può compromettere il processo condiviso e quindi il servizio di altri utenti
    - Limiti di scalabilità
2. Cloud: l'idea alternativa è creare un server video single-threaded per ogni utente, eseguito in un **container** (container == ambiente di esecuzione isolato che contiene tutto il necessario per eseguire un'applicazione; più container possono essere eseguiti sulla stessa macchina quindi ovviamente non serve un server fisico per ogni utente). Vantaggi:
   - Ogni istanza svolge un compito più semplice
   - Un guasto dell'applicaziione in un container tende a restare confinato in quell'istanza
   - Per gestire più utenti basta aggiungere nuove istanze 
  
Il vero problema che ha effettivamente portato alla diffusione del cloud computing del resto sta proprio nei concetti di **scalabilità** e **complessità**. I grandi servizi devono essere in grado di gestire milioni di richieste al secondo, memorizzare quantità enormi di dati e **affrontare variazioni molto rapide del carico** (es. passare da 10k a 100k richieste al secondo per un certo periodo, come per Amazon durante il periodo natalizio. L'idea è nata proprio così a partire da Amazon).  
Ciò crea un problema di dimensionamento per cui se si acquistano risorse per il massimo carico possibile, durante poi i periodi normali una parte dell'infrastruttura resta inutilizzata --> **l'utilizzo del cloud permette di pagare solo per le risorse effettivamente utilizzate**.  
Anche la quantità di dati aumenta la complessità (si parla di **exabyte di dati**, 1 EB = 10^18 byte ergo un miliardo di gigabyte). Il cloud risponde anche alla memorizzazione perché lo storage può essere distribuito su molte macchine: se i dati aumentano il provider può aggiungere quindi dispositivi e server di memorizzazione.

**Le idee che hanno preceduto il cloud**:
- **Utility computing**: considera le risorse informatiche come un servizio da consumare secondo necessità come l'acqua, l'elettricità, il gas o la telefonia. L'utente non deve conoscere l'infrastruttura che produce il servizio ma accedere direttamente alle risorse offerte dal fornitore pagando con il modello previsto.
- **Grid computing**: mette insieme risorse di organizzazioni diverse, spesso distribuite geograficamente, per svolgere attività di calcolo (es. università che mettono insieme server per simulazione scientifica). La caratteristica centrale è la condivisione e coordinamento di risorse tra più organizzazioni
- **Autonomic Computing**: punta a costruire sistemi capaci di gestirsi autonomamente, adattandosi alle variazioni del carico e risolvendo problemi senza intervento umano. Si punta addirittura al recupero da guasti con intervento umano ridotto (idea rilevante per cloud perché gestire ogni macchina è impraticabile su infrastrutture molto grandi)
- **Software as a Service** (SaaS): fornire l'applicazione come servizio accessibile attraverso la rete. L'utente usa il software mentre il fornitore gestisce infrastruttura e esecuzione dell'applicazione.

Si mettono ora in evidenza aspetti diversi dell'organizzazione del calcolo:
- **Distributed Computing**: più computer collegati attraverso una rete collaborano per svolgere un'attività. Ogni nodo possiede risorse proprie e comunica tramite messaggi, e i nodi possono essere eterogenei e relativamente indipendenti (si parla di **loosely coupled**). 
- **Cluster Computing**: un cluster è un insieme di macchine che lavorano in modo strettamente cooridnato. In questo caso i nodi sono invece **tightly coupled**, **homogeneous** (caratteristiche fisiche simili) e organizzati per offrire l'astrazione di un unico sistema.
- **Grid Computing**: come detto prima collega risorse su larga scala, appartenenti a organizzazioni diverse distribuite geograficamente
- **Cloud Computing**: mette al centro il nodo in cui le risorse vengono offerte e utilizzate. Le risorse sono disponibili su richiesta, si accede tramite interfacce di rete, possono essere presentate come risorse virtuali.  
Mentre il cluster descrive soprattutto come sono organizzate le macchine, il cloud descrive come le risorse vengono erogate al cliente.

Definizioni di cloud computing:
1. **Armbrust**: definisce il cloud come comprendente sia le applicazioni offerte come servizi sia l'infrastruttura hardware e software che le supporta. I punti principali sono l'apparente disponibilità di risorse illimitate, il pay per use, "riduzione dell'investimento iniziale" (nel senso che non serve comprare l'infrastruttura fisica)
2. **NIST**: definisce il cloud come accesso tramite rete, su richiesta, a un insieme condiviso di risorse configurabili, rilasciabili rapidamente e con poco intervento manuale
3. **Vaquero**: cloud come pool di risorse virtualizzate facilmente accessibili. Si mette in evidenza la riconfiguraziione dinamica in funzione del carico, il pay per use e le garanzie definite tramite SLA Service Level Agreement (accordo tra fornitore e cliente che definisce le caratteristiche del servizio, es. disponibilità, prestazioni, sicurezza, ecc.)
   
In generale quindi Il cloud computing è un modello di erogazione di risorse informatiche accessibili tramite rete, disponibili su richiesta e rapidamente allocabili e rilasciabili, con gestione automatizzata e utilizzo misurabile. Da queste definizioni cogliamo le caratteristiche del cloud:
1. **On-demand self-service**: l'utente può richiedere risorse senza interazione con il fornitore, autonomamente, tramite ad esempio un portale o un'API
2. **Broad network access**: le risorse sono accessibili tramite rete e dispositivi eterogenei (PC, smartphone, tablet, ecc.). Nel cloud privato l'accesso può avvenire attraverso una rete aziendale, non deve essere necessariamente aperto a internet
3. **Elasticity**: capacità di aumentare e ridurre risorse rapidamente in funzione della domanda.  
Importante distinguere scalabilità (capacità di sostenere carico maggiore aumentando le risorse) da elasticità (capacità di adattare le risorse in funzione del carico). Un sistema può infatti essere scalabile ma poco elastico, ad es. può crescere aggiungendo server ma può richiedere settimane per acquistarli e configurarli. L'elasticità inoltre non implica che l'adattamento sia automatico, è l'**autoscaling** il meccanismo che automatizza questo adattamento.
4. **Resource pooling e multi-tenancy**: il provider organizza le risorse in un pool condiviso, dal quale assegna capacità ai clienti. La multi-tenancy permette a più clienti di usare la stessa infrastruttura fisica mantenendo però isolamento logico
5. **Resource visualization**: la **virtualizzazione** separa la risorsa vista dal cliente dalla sua realizzazione fiisca. Ad es. un cliente vede una macchina virtuale con certe CPU e memoria, mentre il provider gestisce il server sul quale viene eseguita
6. **Pay-per-use**: il costo dipende dall'uso delle risorse secondo metriche e contratto previsti. Il vantaggio è poter iniziare senza acquistare tutta l'infrastruttura
7. **Measured service**: il consumo viene misurato e monitorato. Collegato al pay-per-use ma distinto; measured service riguarda la misurazione tecnica mentre il pay-per-use riguarda il modello economico (potrei avere un incentivo per cui pago meno di quanto uso e quello è il pay-per-use, ma il fornitore comunque misura l'uso effettivo delle risorse per monitoraggio e ottimizzazione)

### Deployment Models
Il **Deployment Model** descrive a chi è destinata l'infrastruttura cloud e come è organizzata. Si discutono tre modelli principali:
1. **Public Cloud**. Qui il provider gestisce un'infrastruttura che offre servizi a clienti diversi --> piattaforma multi-tenant (diversi clienti condividono parti dell'infrastruttura fisica).  
"Pubblico" ovviamente non significa che i dati siano pubblicamente accessibili, ma che il servizio è offerto a una platea generale di clienti. L'accesso alle singole risorse è quindi soggetto a controlli e autorizzazioni. Il cliente utilizza le risorse del provider senza preoccuparsi di dove siano fisicamente collocate e di come gestire l'hardware. Esempi: Amazon AWS, Microsoft Azure, Google Cloud Platform.
2. **Private Cloud**: destinato esclusivamente a una singola organizzazione. Al suo interno possono esserci più utilizzatori, come gruppi di ricerca o unità aziendali. Può essere gestito dall'organizzazione stessa o da un provider esterno o da una combinazione dei due; inoltre può trovarsi nei locali dell'organizzazione oppure essere ospitato esternamente.  
Ciò che lo rende privato è **l'uso esclusivo**, non la posizione fisica.  
**I vantaggi** principali sono maggiore controllo e possibilità di personalizzare infrastruttura e politiche di accesso (maggiore sicurezza nel senso **maggiore controllo sull'isolamento e sulle configurazioni. Non è garanzia automatica, dipende dalla corretta gestione).  
Gli svantaggi sono capacità più limitata, difficoltà nell'espansione e costi più elevati (perché non si sfruttano economie di scala). Se l'infrastruttura fisica è piena tocca aggiungere hardware, non basta creare nuove istanze virtuali come nel public cloud.
3. **Hybrid Cloud**: combina infrastrutture cloud distinte, tipicamente uno privato e uno pubblico, attraverso tecnologie che consentono di integrare servizi e trasferire dati o applicazioni. Le infrastrutture restano quindi separate, ma collaborano.  
Ad es. un'azienda può conservare un DB nel cloud privato ed eseguire alcuni servizi applicativi nel pubblico.

Tra i **vantaggi** del cloud ibrido:
- **Bilanciare utilizzo delle risorse e costi**: l'azienda può usare l'infrastruttura privata per il carico abituale e ricorrere al pubblico se serve capacità aggiuntiva (es. picchi di carico stagionali). Il risparmio è una possibilità, bisogna comunque però considerare i costi di integrazione e trasferimento dati
- **Ridurre il vendor lock-in**: il vendor lock-in è una situazione in cui abbandonare un fornitore diventa difficile o costoso. Con un'architettura ibrida si riduce cioè la dipendenza da un fornitore cloud, se progettata per mantenere dati e applicazioni trasferibili tra i due ambienti.
- **Aumentare la flessibilità**: si può scegliere l'ambiente più adatto a ciascuna esigenza. Es. se si vuole maggiore controllo su alcuni dati, li si può conservare nel cloud privato; oppure ancora per il recupero dopo un guasto grave si può predisporre un ambiente alternativo nel cloud pubblico (se avviene un evento grave e si vuole ripristinare il servizio si parla di **disaster recovery**, se il data center privato non va si possono prendere le copie nel pubblico preparate in anticipo); se si vogliono servire utenti in aree geografiche diverse, si può usare il cloud pubblico per distribuire i servizi più vicino agli utenti; se si vogliono gestire picchi di carico si può attivare capacità pubblica aggiuntiva etc...

La modalità per cui se il cloud privato non ha capacità sufficiente si ricorre al pubblico per gestire il carico aggiuntivo si chiama **cloud bursting**. Il servizio per i clienti in eccesso viene reindirizzato al cloud pubblico! Chiaramente la connessione tra i due ambienti deve essere protetta e veloce.

<img src="imgs/cloud_bursting.png" width="600">

Al contrario gli **svantaggi** del cloud ibrido:
- **Complessità operativa**: occorre gestire più ambienti, spesso con configurazioni diverse. Creazione delle risorse, aggiornamenti, monitoraggio e autorizzazioni devono essere coordinati
- **Sicurezza e compliance**: aumentano i componenti e collegamenti da proteggere: questa è la maggiore **superficie di attacco**. Diventa anche più complesso controllare chi può accedere ai dati e dove questi vengono conservati o trasferiti
- **Possibili ritardi**: la comunicazione tra cloud implica latenza. Ad es. se un'applicazione del cloud pubblico consulta continuamente un db del privato, ogni scambio deve attraversare il collegamento tra i due
- **Troubleshooting più difficile**: quando il servizio rallenta bisogna capire se il problema riguarda l'applicazione, il cloud privato, quello pubblico o la rete
- **Costi di integrazione**: connessioni protette, trasferimenti dati, strumenti di gestione e competenze specialistiche hanno un costo significativo. Avere un cloud ibrido può far risparmiare sull'hardware necessario per i picchi (cloud bursting), ma anche aumentare altre parti della spesa.

### Cloud Stack
Un **cloud provider** è un'organizzazione che offre servizi cloud (es. Amazon AWS, Microsoft Azure, Google Cloud). Si deve distinguere tra chi offre l'infrastruttura cloud da chi poi invece la utilizza per fornire un servizio agli utenti (es. un'azienda che offre una piattaforma di streaming può utilizzare risorse di un provider per eseguire le proprie applicazioni e conservare i dati).

**Cloud Stack**: il cloud può offrire servizi a diversi livelli di astrazione (più alto -> più dettagli tecnici vengono gestiti dal provider). I tre modelli fondamentali sono (da più basso a più alto livello di astrazione):
- **Infrastructure as a Service (IaaS)**: il cliente riceve risorse di calcolo, storage e rete e l'attività principale del cliente è configurare l'ambiente ed eseguire il proprio software.
- **Platform as a Service (PaaS)**: il cliente riceve una piattaforma pronta per eseguire applicazioni. L'attività principale del cliente è sviluppare e distribuire le proprie applicazioni senza preoccuparsi di gestire l'infrastruttura sottostante.
- **Software as a Service (SaaS)**: il cliente riceve un'applicazione pronta all'uso. L'attività principale del cliente è utilizzare l'applicazione senza preoccuparsi di gestire né l'infrastruttura né la piattaforma sottostante.

<img src="imgs/cloud_stack.png" width="500">

Alla base ci sono data center, server e dispositivi di storage. Sopra vengono costruiti servizi infrastrutturali, poi le piattaforme e infine le applicazioni. Le funzioni laterali come fatturazione, misurazione e gestione SLA supportano l'erogazione dei servizi a tutti questi livelli. 

Mentre IaaS, PaaS e SaaS rispondono alla domanda **"cosa viene offerto al cliente?"**, i modelli di deployment rispondono alla domanda **"a chi è destinata l'infrastruttura cloud e come è organizzata?"**

#### Infrastructure as a Service (IaaS)
Con IaaS il provider offre risorse informatiche fondamentali:
- Capacità di calcolo
- Storage e memoria
- Reti virtuali e componenti di rete

Il cliente può utilizzarle per installare ed eseguire il proprio software, compresi sistema operativo e applicazioni, nel caso di una VM (es. il cliente ottiene una VM Linux, installa un server web e configura la propria applicazione). 

La divisione delle responsabilità è la seguente:
- il **provider** gestisce l'infrastruttura fisica e il livello di virtualizzazione
- il **cliente** gestisce il sistema operativo, il software installato e le relative configurazioni

In generale il cliente può configurare alcuni aspetti della rete virtuale come regole di accesso, ma non gestisce direttamente la rete fisica del data center. IaaS non significa in generale solo VM, ma comprende anche servizi infrastrutturali di memorizzazione e rete.

Un esempio di IaaS è Amazon EC2 (Elastic Compute Cloud). Nell'esempio della slide si mostra come se ne richiede una VM: la creazione della macchina avviene attraverso un'interfaccia software, senza ovviamente acquistare e installare un server fisico. Tra le scelte principali ci sono Regione (influisce tra le altre cose in termini di distanza degli utenti e sulla collocazione delle risorse), Immagine Software AMI (un Amazon Machine Image è un modello usato per creare la VM: contiene SO ed eventualmente software preinstallato), Configurazione delle risorse (si sceglie una combinazione di CPU, memoria e altre capacità. T-shirt sizes come combinazioni predefinite), Security Group (regole sul traffico di rete consentito verso la VM, es. accesso SSH solo da una certa sorgente), Chiavi di accesso (una coppia di chiavi pubblica/privata per accedere alla VM).  
Una volta creata la VM, il cliente deve comunque configurare e gestire il software da utilizzare. 

#### Platform as a Service (PaaS)
Con PaaS il cliente riceve una piattaforma che permette di sviluppare, distribuire ed eseguire applicazioni.  
La differenza rispetto a IaaS è il livello al quale lavora il cliente: mentre con IaaS parte da risorse infrastrutturali e prepara l'ambiente, con PaaS parte da una piattaforma già pronta per gestire gran parte dell'ambiente di esecuzione (es. uno sviluppatore carica un'applicazione Python e la piattaforma prepara le risorse necessarie per eseguirla).

Il cliente quindi controlla il codice e le configurazioni applicative esposte al servizio, mentre il provider gestisce i livelli sottostanti. 

Il vantaggio è la **riduzione del lavoro di gestione**, ma con il compromesso che l'applicazione deve rispettare i linguaggi e le librerie supportate dalla piattaforma. 

Il **Function as a Service (FaaS)** è un sottoinsieme di PaaS che permette di distribuire singole funzioni, eseguite in risposta ad eventi o richieste.

L'esempio presentato di PaaS è Elastic Beanstalk, sempre di quella merdaccia di Amazon. Il cliente carica il pacchetto dell'applicazione e fornisce le configurazioni necessarie. La piattaforma crea e configura le risorse che la eseguiranno.  

<img src="imgs/elastic_beanstalk.png" width="400">

Route 53 è il servizio di gestione del DNS che associa il nome dell'applicazione al punto di accesso appropriato. C'è poi il Load Balancer per la distribuzione delle richieste, che le inoltra adeguatamente alle istanze dell'applicazione e un Auto Scaling Group che permette l'adattamento in base al numero di istanze (si aumentano o diminuiscono il numero di VM EC2 secondo le politiche configurate). Infine il Security Group definisce le regole di accesso alla rete per le istanze dell'applicazione.  
Il database è gestito separatamente dalle istanze applicative.

Quindi in generale con PaaS l'astrazione riguarda le risorse infrastrutturali e la gestione del sistema operativo: le VM esistono ma la piattaforma automatizza gran parte della loro gestione. 

IaaS vs PaaS: Con IaaS si parte dalle risorse infrastrutturali e la gestione del SO è a carico del cliente. La libertà di configurazione è maggiore e così anche il lavoro per la gestione dell'ambiente. Con PaaS il cliente parte da una piattaforma già pronta e gestita dal provider, quindi il lavoro di gestione è minore ma anche la libertà di configurazione è ridotta.

Inoltre c'è un grosso problema con PaaS: una PaaS può imporre vincoli all'architettura dell'applicazione e alla gestione dei dati, e se il codice dipende molto dalla piattaforma può essere difficile migrare l'applicazione su un'altra --> **vendor lock-in**

#### Software as a Service (SaaS)
Con SaaS il cliente utilizza un'applicazione già realizzata ed eseguita dal provider (es. Gmail, Microsoft Teams, Google Docs, Mega...).  
In questo caso il cliente non gestisce server e reti, SO, ambiente di esecuzione, codice dell'applicazione, tuttavia può gestire le impostazioni consentite, gli utenti e i contenuti inseriti (es. per Gmail si configurano account e preferenze, non si installa e aggiorna il software del server di posta).

Vediamo come cambiano le responsabilità passando dalla gestione locale al SaaS:
- locale (on-premises): l'organizzazione gestisce tutti i livelli
- IaaS: il provider gestisce l'infrastruttura sottostante
- PaaS: il provider gestisce anche l'ambiente d'esecuzione
- SaaS: il provider gestisce anche l'applicazione

Esistono anche modelli più specifici come **Database as a Service (DBaaS)**, **Function as a Service (FaaS)**, **Monitoring as a Service (MaaS)** (per monitorare l'attività dell'applicazione), **Generative AI as a Service (GAIaaS)** (per fornire LLM come servizio) e **X as a Service (XaaS)** per indicare qualsiasi servizio erogato come servizio cloud.

### Modelli di prezzo 
Il cloud può adottare modalità di pagamento diverse:
- **on-demand e pay-per-use**: il costo dipende dalle risorse utilizzate e dalla durata o quantità di utilizzo. Adatto a carichi variabili e ambienti temporanei.
- **fixed o reserved**: il cliente assume impegno economico per ottenere condizioni più convenienti dell'on-demand.
- **dedicated**: le risorse fisiche sono dedicate al cliente. Questo offre maggiore isolamento fisico ma tende a costare di più
- **spot**: il cliente utilizza capacità disponibile a condizioni economiche vantaggiose, accettando però che il fornitore può revocare le risorse in qualsiasi momento. Adatto a carichi di lavoro flessibili e tolleranti a interruzioni (es. elaborazioni batch che possono essere interrotte e riprese senza problemi. Non va quindi bene per applicazioni critiche).

### Capacity Planning
Il **capacity planning** consiste nel determinare quante risorse servono e di quali dimensioni. Può riguardare ogni tier del cloud stack. 

Si definisce **previsioning** l'assegnazione e preparazione effettiva di queste risorse. Nel **previsioning statico** la capacità è stabilita in anticipo e resta fissa.  
Se invece la domanda varia si presentano due possibilità: **over-previsioning** (capacità disponibile supera quella necessaria) o **under-previsioning** (capacità disponibile è inferiore a quella necessaria).

<img src="imgs/capacity_planning.png" width="400">

Prevedere il carico è difficile perché la domanda può avere andamenti diversi: crescente, periodica, eccezionale (se presenta un grande picco legato a un evento) e imprevedibile. Inoltre non varia solo il numero di richieste, ma possono variare anche le prestazioni delle risorse (es. se un server è sovraccarico le richieste possono essere più lente oppure se si hanno problematiche di rete).

<img src="imgs/load_variation.png" width="400">

La soluzione classica è lo **scaling**, i.e. prevedere il carico e modificare le risorse a disposizione. Si parla di **scale in/out** se si aggiungono o rimuovono istanze (es. VM o container) e di **scale up/down** se si aumenta o diminuisce la capacità di un'istanza (es. più CPU, più memoria).  
Nel grafico dello scaling manuale la capacità cresce a gradini, e tra un incremento e il successivo rimane capacità inutilizzata. Una previsione errata può lasciare risorse insufficienti.

<img src="imgs/scaling.png" width="400">

In pratica lo scaling manuale richiede tempo ed è difficile da gestire quando il carico cambia rapidamente.  
La soluzione moderna è l'**autoscaling**: un sistema osserva il carico e modifica automaticamente le risorse secondo una politica (es. se il carico CPU è alto crea una nuova istanza VM per cui ad es. se 120 richieste al secondo con 2 VM 60 ciascuna, con 3 invece 40 ciascuna per cui ogni VM svolge meno lavoro e il carico CPU scende, il lavoro totale resta lo stesso ma viene distribuito su più CPU disponibili).

L'idea è **far seguire alla capacità l'andamento della domanda**

### Elasticity
L'analogia fisica è quella di un materiale che si allunga sotto una forza e si restringe quando la forza viene rimossa. Nel cloud la forza è il carico di lavoro e l'allungamento è l'aumento della capacità: secondo la definizione l'**elasticità** è il grado con cui un sistema riesce ad adattarsi a cambiamenti di carico, rilasciando o allocando risorse autonomamente affinché **la capacità disponibile corrisponda il più possibile alla domanda**.

Comprende quindi entrambe le direzioni: aumentare risorse quando la domanda cresce e rilasciarle quando diminuisce. La distinzione con la scalabilità resta fondamentale: la scalabilità riguarda la capacità di sostenere più carico aggiungendo risorse, mentre l'elasticità riguarda quanto bene e rapidamente le risorse vengono adattate alle variazioni del carico.

Per realizzare l'elasticità servono sistemi software autoadattivi. Un ciclo tipico è:
1. Monitorare il comportamento del sistema e misurare il carico
2. Valutare se la capacità è adeguata
3. Decidere un'eventuale modifica
4. Applicarla e osservarne l'effetto

Tale automazione rende praticabili gestioni di infrastrutture enormi, altrimenti ingestibili manualmente. 

#### Misurare l'elasticità
Come anticipato over-provisioning se la capacità disponibile supera la domanda, under-provisioning se la capacità disponibile è inferiore alla domanda. L'area in blu rappresenta la domanda, in rosso le risorse disponibili.

Una misura dell'elasticità è l'**accuracy** e si calcola come **somma delle aree di over-provisioning $O$ e under-provisioning $U$ nel periodo $T$**.  
Per esprimere matematicamente l'idea, indicando con $d(t)$ la domanda e $c(t)$ la capacità:
$$O = \int_{0}^{T} \max(0, c(t) - d(t)) dt \quad U = \int_{0}^{T} \max(0, d(t) - c(t)) dt$$
Quindi l'accuracy è data da:
$$O + U = \int_{0}^{T} |c(t) - d(t)| dt$$
sarebbe a dire più piccolo è lo scostamento complessivo, più preciso è l'adattamento. Nota che la misura di accuracy è ottimale se pari a 0.

Un'altra metrica è il **timing**: rappresenta il **tempo in cui si rimane in una situazione di scostamento** (i.e. misura la durata totale degli intervalli di under-provisioning e over-provisioning). La differenza rispetto all'accuracy è utile: due sistemi possono rimanere insufficienti per lo stesso tempo, ma uno può mancare di poche risorse e l'altro moltissime. In questo caso l'accuracy è più alta per il secondo sistema (nel senso che il secondo è peggiore, accuracy ottima se pari a 0), ma il timing è lo stesso.

### Load Balancer
Per distribuire il carico tra più repliche dell'applicazione si utilizza un load balancer. Il client contatta un punto di accesso comune, il balancer sceglie una replica (i.e. istanza dell'applicazione) e inoltra la richiesta.  
Il ***VIP Virtual IP** è l'indirizzo virtuale che rappresenta questo punto di ingresso.

Si distinguono diverse possibilità: un balancer di **livello 4** decide usando informazioni del livello di trasporto (TCP/UDP) come indirizzi IP e porte, mentre un balancer di **livello 7** può usare informazioni del livello applicativo (es. URL, intestazioni HTTP, cookie) per decidere a quale replica inoltrare la richiesta.  
Nel caso **one-way**, il bilanciatore inoltra la richiesta a una replica e questa risponde direttamente al client. Nel caso **two-way** invece la risposta del server deve passare di nuovo attraverso il balancer.

Gli **health check** verificano periodicamente lo stato delle repliche e se una replica non risponde o risponde con errore, il bilanciatore la esclude temporaneamente dalla lista di quelle disponibili.

Tra gli obiettivi di Load Balancing: **utilizzare bene le risorse**, **ridurre i tempi di risposta** e **aumentare la throughput** (i.e. numero di richieste completate per unità di tempo).

Per scegliere una replica si possono usare moltissime politiche diverse:
- Random
- Round Robin (distribuisce le richieste a turno)
- Least Loaded
- Weighted Round Robin (assegna più richieste a repliche più potenti)
- Power of Two Choices (sceglie due repliche a caso e inoltra alla meno carica)
- Hashing
- Session Stickiness (per mantenere la sessione di un utente sempre sulla stessa replica)

Random, Round Robin, Weighted Round Robin e Hashing non richiedono di conoscere il carico attuale del server: queste due politiche in uqesto senso sono dette **stateless**. 

Le politiche **client-aware** tengono conto di informazioni relative al client o alla sessione (es. Session Stickiness). Tuttavia l'applicazione della Session Stickiness può rendere meno uniforme la distribuzione del carico.

Nella figura le richieste provenienti da Internet raggiungono un Application Load Balancer che le distribuisce alle istanze EC2. Tali istanze sono appartengono a un Auto Scaling Group e sono distribuite su più Availability Zone (AZ) (zone distinte in una regione, necessario per **limitare gli effetti di guasti locali**). Altri elementi dilimitano e proteggono l'ambiente, come una **VPC** (rete virtuale privata dell'organizzazione cloud), una **subnet** (suddivisioni della rete VPC), un **security group** (insieme di regole per il traffico di rete consentito) e un **target group** (insieme di destinazioni alle quali il balancer può inoltrare le richieste).  
Quando il carico cresce l'autoscaling aggiunge istanze che, una volta pronte e riconosciute come sane, vengono aggiunte al target group e quindi il balancer può distribuirvi le richieste. Quando il carico diminuisce l'autoscaling rimuove istanze dal target group e le termina.

### Service Level Agreement (SLA)
Un **SLA** è un accordo formale tra fornitore e cliente che definisce il livello di servizio previsto. 
Per definire la qualità del servizio servono delle **metriche misurabili** (es. tempo medio di risposta inferiore al secondo, disponibilità del servizio superiore al 99.9%, ecc.); tali condizioni sono note come **Service Level Objectives (SLO)**.  

In generale un SLA contiene più SLO insieme anche ad altre informazioni come:
- Servizi e soggetti coinvolti
- Periodo di validità dell'accordo
- Modalità di misurazione
- Limitazioni e responsabilità del fornitore e del cliente
- Compensazioni previste in caso di mancato rispetto degli SLO

L'SLA infatti non impedisce tecnicamente un guasto, ma stabilisce gli impegni e le conseguenze previste nel caso in cui non vengano rispettati.

In generale il ciclo di vita di un SLA presenta sei fasi:
1. Individuare un Provider
2. Definire il lvl di servizio richiesto
3. Stabilire un accordo
4. Monitorare il servizio e rilevare eventuali violazioni
5. Terminare l'accordo
6. Applicare eventuali penalità previste 

Il punto centrale in questo senso è il monitoraggio durante l'esecuzione, noto come **runtime monitoring**, in modo che il fornitore possa verificare se sta rispettando gli SLO e il cliente possa verificare se il servizio ricevuto è conforme a quanto stabilito.

Un algtro termine importante è **SLI Service Level Indicator**, che indica una metrica specifica usata per misurare un SLO. Quindi in generale:
- SLI == indicatore misurabile della qualità del servizio (es. percentuale mensile di disponibilità, tempo di risposta, throughput e tasso di errore...)
- SLO == condizione da rispettare per un certo SLI (disponibilità >= 99.9%, tempo di risposta <= 1s, throughput >= 1000 req/s, tasso di errore <= 0.1%. In questo senso un SLO si può vedee coem un predicato, nel senso che la condizione alla fine dell'accordo può essere vera o falsa)
- SLA == accordo formale tra fornitore e cliente che definisce gli impegni (SLO) e le conseguenze in caso di mancato rispetto

es. EC2: l'estratto relativo alla singola istanza EC2 indica un obiettivo mensile del 99.5%. Se indichiamo la disponibilità come frazione del tempo otteniamo:
$$\text{Disponibilità} = \frac{\text{Tempo di servizio disponibile}}{\text{Tempo totale del mese}}$$
Per cui si lascia una frazione di indisponibilità dello 0.5%. Per un mese di 30 giorni si parla di 216 minuti, cioè 3 ore e 36 minuti (ciò mostra come una percentuale vicina al 100% possa comunque tradursi in un tempo di inattività significativo).  
In caso di violazione l'estratto prevede service credits: crediti da utilizzare per pagamenti futuro.

Gli SLA presentano diverse difficoltà:
- **Terminologia e definizioni**: due provider possono usare una stessa parola ma intenderla diversamente
- **Le medie nascondono la distribuzione dei problemi**: un'ora di indisponibilità può essere un unico guasto di un'ora come può essere un guasto di 1 minuto al giorno per 60 giorni. 
- **Disponibilità non significa buone prestazioni**: un servizio può essere disponibile ma lento o con errori. Per questo servono metriche diverse per misurare la qualità del servizio
- **Compensazioni limitate**: un credito su pagamenti futuri può essere molto inferiore al danno subito dal cliente
- **Rilevazione e segnalazione**: spesso il cliente deve raccogliere prove e segnalare il problema entro un termine 
- **Scarsa negoziabilità**: i provider spesso offrono SLA standard, senza possibilità di negoziare condizioni diverse.


### Cloud Monitoring
Collegandosi a SLA, il monitoraggio è consiste nel raccogliere e analizzare dati per valutare lo stato dei sistemi e dei servizi. Oltre che a verificare però gli obiettivi di servizio serve a individuare guasti, problemi di prestazioni e inefficienze.

Si distinguono:
- **Metriche di Sistema**: misurano lo stato delle risorse (CPU, memoria utilizzata e libera, operazioni e throughput del disco, traffico di rete, ecc.)
- **Metriche Applicative**: misurano lo stato delle applicazioni (tempo di risposta, tasso di errore, numero di richieste o transazioni, interazioni tra gli utenti)

Le due categorie si completano, ad esempio un tempo di risposta elevato segnala un problema percepito dall'utente. L'utilizzo della CPU, della rete e del disco può aiutare a individuarne la causa.  
La CPU elevata da sola non dimostra un servizio scadente: un sistema può usare bene le risorse e continuare a rispettare i tempi di risposta.

Esistono diversi strumenti e dashboard per raccogliere e visualizzare le metriche, come Amazon CloudWatch o strumenti esterni come Prometheus. Si parla di raccolta delle metriche, grafici del loro andamento, allarmi, analisi dei log... 

<img src="imgs/cloudwatch.png" width="500">

Il collegamento con l'elasticità è diretto: **il monitoraggio fornisce le informazioni grazie alle quali l'autoscaling può decidere se aumentare o ridurre le risorse disponibili**.

### Data center e merda contemporanea
Le infrastrutture fisiche di esecuzione del cloud sono grandi data center noti tipicamente come **WSC warehouse-scale computers**, sistemi di calcolo su scala di un intero edificio. 

Non si tratta solo di una raccola di server, per funzionare come piattaforma servono reti che li colleghino, sistemi di storage, software di gestione e coordinamento, alimentazione e raffreddamento.  
Tale insieme di macchine rende disponibile la capacità del cloud: quando si aggiungono VM il provider deve collocarle su risorse fisiche con capacità utilizzabile.

I server sono organizzati in **rack** (armadi verticali che contengono più server). I rack sono disposti in fila ed ogni rack contiene server, switch di rete, componenti per alimentazione e batterie di backup per gestire brevi eventuali interruzioni. Questa disposizione permette di organizzare molte risorse in modo compatto ma richiede attenzione a cablaggio, manutenzione e dissipazione del calore. 

Un data center consuma molta energia per mantenere in funzione i server.  
L'efficienza energetica è descritta mediante il Power Usage Effectiveness **PUE**:
$$PUE = \frac{\text{Potenza totale assorbita dal data center}}{\text{Potenza assorbita dagli strumenti IT}}$$
Ad esempio se le apparecchiature IT assorbono 10 MW e l'intero impianto 11 MW allora il PUE è 1.1, cioè il 10% dell'energia serve per alimentare e raffreddare l'infrastruttura. Più il PUE è vicino a 1, più efficiente è il data center.  
Comunque il PUE non indica quanta energia consuma complessivamente il data center --> anche un impianto con buon PUE può consumare moltissimo.

I provider distribuiscono i data center in diverse aree geografiche. Pianificare e costruire un nuovo impianto richiede tempi lunghi dato che serve spazio, energia, raffreddamento, connettività, sicurezza e personale specializzato. La posizione geografica influisce sulla latenza.

### Applicazioni cloud
Il cloud viene utilizzato per applicazioni scientifiche, sanitarie, aziendali e rivolte ai consumatori. 

Un'applicazione web cloud è organizzata a più livelli (multi-tier). Nell'esempio se ne vede una a tre livelli logici: il livello di presentazione (front-end) gestisce l'interfaccia con l'utente, il livello applicativo (back-end) gestisce la logica dell'applicazione e il livello di dati gestisce la memorizzazione e l'accesso ai dati. Ogni livello nel cloud chiaramente non corrisponde a un singolo server fisico, ma può essere realizzato da più istanze virtuali distribuite su più server fisici.

Tra le componenti principali nel diagramma AWS abbiamo Route 53 per la risoluzione DNS, Cognito per la gestione autenticazione utenti, CloudFront e S3 per distribuzione e conservazione dei contenuti statici, API Gateway, Application Load Balancer, ECS e Fargate per eseguire e gestire le istanze applicative (container), DynamoDB servizio database NoSQL, ECR per conservare le immagini dei container (con immagine del container si intende il pacchetto software pronto per essere eseguito in un container), CloudWatch per monitoraggio e allarmi. L'applicazione è distribuita su due Availability Zone per limitare gli effetti di guasti locali e utilizza reti private per i componenti interni.  
Il punto dello schema è che un servizio cloud combina molti servizi specializzati.

<img src="imgs/fort.png" width="500">

Un'applicazione **cloud-native** è un'applicazione pensata per essere eseguita proprio in ambiente cloud. Si tratta di applicazioni composte da servizi piccoli, indipendenti e "loosely coupled" (i.e. con pochi vincoli tra loro). Questa organizzazione facilita aggiornamenti indipendenti, scalabilità dei componenti più utilizzati e resilienza ai guasti. Nel corso si approfondiranno microservizi e applicazioni serverless.

Un servizio **stateful** conserva uno stato tra interazioni (es. mantiene dati relativi a una sessione e utili per successive interazioni), mentre un servizio **stateless** non conserva localmente informazioni tra interazioni.  
Un servizio stateless è più semplice da scalare e gestire in quanto ogni richiesta può essere gestita da qualsiasi istanza del servizio. Per lo stateful invece ci si ricollega al Session Stickiness del load balancer o a un database esterno per conservare lo stato.

<img src="imgs/merda.png" width="500">

Altri temi del cloud sono modelli ibridi e multi-cloud (cioè l'uso di più provider cloud); servizi AI, container e serverless (cioè senza server, dove il provider gestisce l'infrastruttura e il cliente si occupa solo del codice da eseguire); Governance e FinOps. 

La **Cloud Governance** riguarda le politiche e le regole per gestire l'uso del cloud, la sicurezza, la conformità e l'ottimizzazione dei costi.  
**FinOps** riguarda invece la gestione economica del cloud, con l'obiettivo di ottimizzare i costi e massimizzare il valore degli investimenti in cloud (es. creare più VM può migliorare le prestazioni ma aumenta la spesa. Occorre valutare entrambe e rilasciare le risorse che non servono più)

Il **multi-cloud** utilizza servizi di più provider. Questo introduce problemi di coordinamento legato alle interfacce e configurazioni diverse, protezione dei dati, monitoraggio distribuito etc... uno degli approcci è **IaC Infrastructure as Code**, cioè descrivere l'infrastruttura e configurazioni attraverso file che gli strumenti del provider possono interpretare per creare e gestire le risorse. Il punto principale dell'IaC è descrivere quindi l'infrastruttura tramite file, **rendendo le operazioni automatizzabili e riproducibili** (tipicamente funziona così: non è il provider a leggere il file IaC ma lo strumento es. Terraform, che lo legge e invia le richieste alle API del provider per creare le risorse).

Il cloud può rendere accessibili risorse e tecnologie specializzate, come AI generativa e applicazioni AI-native, Web3 e architetture decentralizzate come blockhain (Web3 == interagire con applicazioni basate su blockchain), Quantum as a Service per accedere attraverso la rete a risorse di calcolo quantistico.

Tuttavia sempre da ricordare che **la latenza non è mai zero** e che quindi non necessariamente il cloud è adatto ad ogni applicazione.

### Beyond the Cloud
Come detto la latenza non è zero. Per alcune applicazioni questo è accettabile, ma per altre (es. applicazioni industriali, robotica, veicoli autonomi, realtà aumentata e virtuale) la latenza deve essere molto bassa. L'idea è quindi **spostare una parte dell'elaborazione e della memorizzazione più vicino ai luoghi in cui i dati vengono prodotti**. 

Da questo nascono i concetti di **Edge Computing** (elaborazione vicino ai dispositivi) e **Computing continuum** (collaborazione tra risorse vicine e cloud), entrambi rappresentano quindi modelli che si integrano con il cloud.

#### Edge Computing
Si porta la capacità di calcolo e storage alla periferia della rete, vicino a sensori, utenti e dispositivi. Un **edge node** può essere un gateway, un piccolo server, uno smartphone o in generale un dispositivo con capacità di elaborazione.

es. una telecamera produce un flusso video. Un server vicino (edge node) può analizzare il flusso in tempo reale per rilevare oggetti o eventi, inviando al cloud solo risultati o sequenze rilevanti. In genere gli edge node sono posti a "one hop" dal dispositivo del cliente.

**Vantaggi**:
- Riduzione della latenza
- Riduzione della banda necessaria (i dati possono essere filtrati o aggregati localmente, come nell'esempio della telecamera)
- Possibile risparmio energetico (meno dati da trasmettere, anche se si deve considerare il consumo dell'edge node)
- Maggior controllo sui dati (alcuni dati possono restare nel luogo in cui vengono prodotti. Ciò aiuta la privacy purché i dispositivi edge siano sicuri e affidabili)
- Funzionamento con connettività limitata (un sistema può continuare alcune attività locali anche quando il collegamento al cloud è debole o temporaneamente assente)

L'edge AI consiste nell'eseguire modelli di AI direttamente sui dispositivi o nodi edge.  
Le risorse edge possono essere offerte da provider cloud, reti di distribuzione dei contenuti o installate direttamente dall'organizzazione.

**Limiti**:
- Risorse limitate: i nodi edge hanno capacità inferiori rispetto al cloud, quindi non possono eseguire modelli complessi o gestire grandi quantità di dati
- Gestione distribuita: più nodi edge significa più complessità nella gestione, aggiornamento e monitoraggio
- Sicurezza: i nodi edge possono essere più vulnerabili a intrusioni fisiche o attacchi informatici
- Eterogeneità e standardizzazione: Dispositivi e piattaforme possono avere hardware, protocolli e strumenti differenti. 

Una risposta al primo limite è l'**offloading**: spostare parte del carico di lavoro dal nodo edge al cloud o ad altri nodi edge più potenti (es. l'edge node rileva un'anomalia --> il cloud esegue un'analisi approfondita e conserva lo storico).

#### Computing Continuum
Il **computing continuum** integra tutti i livelli visti con il multi layer edge computing, e tenta di porre ognuno di essi in un ambiente nel quale possano essere più adatte.

<img src="imgs/fog.png" width="400">

In generale si vuole verso cloud una maggiore capacità di elaborazione e memorizzazione, mentre verso i dispositivi periferici si vuole la possibilità di rispondere rapidamente a interazioni locali. Il problema centrale diventa quindi decidere dove eseguire ciascuna attività considerando latenza risorse, energia etc...

Vediamo un esempio con monitoraggio ambientale: sensori ambientali distribuiti geograficamente. Un possibile funzionamento del sistema può essere:
1. i sensori raccolgono dati e li inviano a un edge node locale
2. L'edge node filtrano e rilevano eventi urgenti
3. Le risorse regionali aggregano dati di un'area
4. Il cloud conserva lo storico ed esegue analisi globali

L'applicazione può essere rappresentata da un DAG grafo diretto aciclico, dove i nodi rappresentano le attività di elaborazione e gli archi i flussi di dati.

**Difficoltà del computing continuum**:
- Interoperabilità: permettere a dispositivi e piattaforme diverse di comunicare
- Astrazioni di programmazione: offrire modi coerenti per descrivere e distribuire le applicazioni
- Eterogeneità: considerare differenze di hardware, energia e connettività
- Prestazioni e SLA: rispettare gli obiettivi mentre il lavoro attraversa più livelli
- Sicurezza e privacy: proteggere i dati, comunicazioni e accessi lungo tutto il percorso

es. un'attività può essere troppo pesante per il nodo locale, ma spostarla nel cloud può violare il requisito di latenza.



### Benefici del Cloud e altro
Sul piano **tecnico**, il cloud offre:
- Scalabilità e Elasticità
- Accessibilità tramite rete
- Gestione semplificata, soprattutto usando PaaS e SaaS
- Strumenti e competenze di sicurezza del provider
- Rapidità nel predisporre ambienti e servizi

La sicurezza può migliorare, ma dipende anche dalle configurazioni e dalle responsabilità del cliente.

Sul piano economico il cloud riduce la necessità di acquistare l'infrastruttura fisica in anticipo e permette di pagare solo per le risorse effettivamente utilizzate. Si passa da CapEx (spese in conto capitale, i.e. investimento per acquistare beni come server) a OpEx (spese operative per utilizzare servizi).

Per quanto riguarda la **privacy e sicurezza**, il cliente deve sempre porsi molte domande prima di affidare dati a servizi così grandi e complessi. Alcune domande chiave sono:
- Dove sono fisicamente collocati i dati?
- Chi può accedervi?
- Per quali finalità vengono utilizzati?
- Come vengono protetti e recuperati?

Tra gli aspetti tecnici troviamo Cifratura (sia durante la trasmissione che a riposo), Integrità (possibilità di rilevare o prevenire modifiche non autorizzate), Tracciabilità, Recupero.

La **Portabilità** è la possibilità di spostare e adattare dati e applicazioni tra ambienti diversi.  
L'**Interoperabilità** è la capacità di sistemi diversi di comunicare e lavorare insieme.

La portabilità può essere ostacolata dal **vendor lock-in**: dipendenza da API, formati o funzionalità proprietarie.  
I container aiutano a rendere ripetibile l'ambiente software, mentre l'IaC aiuta a ricreare l'infrastruttura. Tuttavia un container **non rende automaticamente portabile un database gestito da un servizio proprietario dal quale l'applicazione dipende**.

Per l'interoperabilità invece occorre coordinare interfacce, autenticazioni e autorizzazioni, formati dei dati... 

Ritorna inoltre il tema degli SLA: le metriche del provider possono non essere sufficienti per valutare la qualità del servizio percepita dal cliente. **In questo senso il monitoraggio da parte del cliente è fondamentale**.

Si torna poi a Scalabilità e elasticità, due domande fondamentali: 
- **l'applicazione è progettata per scalare?** Aggiungere VM aiuta se il lavoro può essere distribuito, ma se tutte le repliche dipendono da un database già saturo quello può rimanere collo di bottiglia.
- **il provider riesce a fornire rapidamente le risorse necessarie?** La creazione e avvio delle istanze richiede tempo.

Quindi per ottenere elasticità servono sia un'applicazione capace di utilizzare risorse aggiuntive sia un'infrastruttura capace di renderle disponibili rapidamente. I carichi **stateful** sono quelli più difficili da scalare in quanto bisogna gestire la coerenza dello stato tra repliche.

### Cloud outages e guasti a cascata, le difficoltà del provider
Anche grandi infrastrutture possono diventare indisponibili. Le cause possono comprendere esaurimento risorse, errori software o di configurazione, aggiornamenti difettosi, problemi fisici del data center.  
Il concetto principale è il **cascading failure** (guasto a cascata): se un servizio condiviso semtte di funzionare, questo può compromettere tutti quelli che dipendono da esso e così a cascata. Avere più repliche quindi non è sempre sufficiente, se tutte dipendono dallo stesso componente guasto siamo fottuti tutti --> **la resilienza richiede di considerare anche le dipendenze comuni oltre ai singoli server**.

Tra le difficoltà del provider:
- Consumo energetico (tra calcolo, storage e raffreddamento)
- Domanda incerta (il provider deve avere capacità sufficiente per i clienti, limitando risorse inutilizzate)
- Gestione SLA (rispettare gli impegni presi con i clienti)
- Interoperabilità (fornire strumenti e interfacce che permettano integrazione e gestione)
- Competenze specialistiche

In definitiva il cloud modifica il modo in cui i servizi vengono progettati ed erogati, permettendo di ottenere risorse tramite servizi delegano parte della gestione. La complessità dei sistemi distribuiti rimane però presente.
