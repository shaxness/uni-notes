## Introduction to Deep Learning
We only briefly introduce deep learning here. For a more comprehensive understanding, wait for the upcoming Deep Learning courses.

Deep learning is mostly linear algebra. The basic element in deep learning is the **perceptron** (aka **neuron**), note that it started as a biological ispiration in the '60, but the functionality of today's models is very different from biological neurons and not ispired by them anymore.

A **perceptron is nothing more than a function that takes one or more features as input and outputs a single value**. The output is computed as a weighted sum of the inputs, passed through an activation function.  
$ y = f(w_0 + w_1x_1 + w_2x_2 + ... + w_nx_n) $  
Aside from the application of f, the values in input are nothing more than what we have seen in linear regression -> the perceptron is a generalization of linear regression, where we don't only take the weighted sum of the inputs, but we also apply a function f (which we'll see, is non-linear) to it.

Below the classical representation of a perceptron: we have n inputs (x1, x2, ..., xn), each one multiplied by a weight (w1, w2, ..., wn). The perceptron is the weighted sum of the inputs plus a bias (w0) which is another weight that we have to figure out ourselves. Then everything is passed through an activation function f, which produces the output y.

<img src="img_teoria/perceptron.png" alt="Perceptron" width="600"/>

The interesting part is how to choose the weights (including the bias). Let's imagine to have only two features (x1 and x2) -> two weights (w1 and w2) plus the bias (w0). The perceptron is a *family of models*, since each different choice of weights and bias produces a different model. So different choices of w0, w1 and w2 will produce a different function that maps the inputs (x1, x2) to the output y.

### Activation Functions
The main reasons because we apply an activation function f to the weighted sum are:
- **To enforce properties on perceptron's output**: for example, in binary classification we want the output to be between 0 and 1, so we can interpret it as a probability. In this case, we can use the sigmoid activation function.
- **To introduce non-linearity**: if we didn't apply any activation function, the perceptron would be just a linear model (weighted sum of inputs). By applying a non-linear activation function, we can model more complex relationships between inputs and output.

Commonly adopted functions are *ReLU, Sigmoid, Leaky ReLU, Tanh, Softmax, Linear, GeLU*, each of these is non linear and has its own characteristics that make it more or less suitable for specific tasks. Also all these functions are of the kind $f: R -> R$ (they take a single real number as input and output a single real number).

#### Enforce properties on perceptron's output
Let's imagine to deal with a binary classification problem, where we want the output of the perceptron to be between 0 and 1, so we can interpret it as a probability. 

Imaginining to have only two features (x1 and x2), the perceptron would have only two weights (w1 and w2) plus the bias (w0). Normally, with SDV, Classification trees etc.. we would try to find the line that best separates the two classes. Here, instead, given in input $x \in R^2$ (the two features), we want to output a value between 0 and 1 that represents the probability that x belongs to class 1 (vs class 0).

For binary classification, we must ensure that $Pr(positive|x) \in [0, 1]$ and $Pr(negative|x) = 1 - Pr(positive|x)$. Since, by construction (it is a binary problem), $Pr(positive|x) + Pr(negative|x) = 1$, we only need to ensure that $Pr(positive|x) \in [0, 1]$.

To do so (since normally with the weighted sum we get a number that is in between $(-\infty, +\infty)$), we can use the **sigmoid activation function** (see image below). The sigmoid function takes any real number as input and outputs a value between 0 and 1, the bigger the input, the closer the output is to 1, the smaller the input, the closer the output is to 0. The function is defined as:

$ f(x) = \frac{1}{1 + e^{-x}} $

As we change the weights, different functions i.e. different probabilities for each input point are produced.

#### Adding more perceptrons
We have seen only the case of a single perceptron, but deep learning models are made of many perceptrons connected together in layers. Each perceptron is put one in top of the other, takes the same inputs and produces different outputs (since each perceptron has its own weights and bias). The outputs of all the perceptrons in a layer are then passed as input to the next layer of perceptrons, and so on.
For each perceptron, we have an output.

The set of perceptron that take the same inputs is called a **linear layer** (aka **fully connected layer**, since each input is connected to every neuron and in the following layer each neuron is connected to every output of the previous layer). A deep learning model is made of many layers of perceptrons connected together. The first layer is called the **input layer**, the last layer is called the **output layer**, and all the layers in between are called **hidden layers**.

In the first image below we see only one layer of perceptrons (that is both the input and output layer), in the second image we see two layers connected together (input and output respectively), in the third image a general case (also hidden layers).

For the first image, the output of each perceptron is nothing more than the dot product between the weight vector transposed and the input vector:
$ y_i = f(w_{i}^T x) $  
where i is the index of the perceptron in the layer.

Considering all perceptrons, we are actually applying a **matrix-vector multiplication** between the weight matrix W transposed (where each row is the weight vector of a perceptron, and the number of columns is the number of features) and the input vector x:
$ y = f(W^T x) $

The number of weights is known as the number of **parameters** of the model (size of the model). The more parameters we have, the more complex the model is, and the more data we need to train it properly.

#### Introduce non-linearity
Another important reason to apply an activation function is to introduce non-linearity in the model. If we didn't apply any activation function, the perceptron would be just a linear model (weighted sum of inputs), since a linear combination of linear functions is still a linear function.

This would mean that using multiple layers of perceptrons without activation functions would be equivalent to using a single layer of perceptrons!

In the example above, if f(x) = x (i.e. no activation function), then: $z = s^T W^T x$.  
This implies that we could've used $W' = W s$ as weight matrix for a single layer of perceptrons, obtaining the same result -> we wouldn't have needed the second layer at all! And we would have built a model that is still linear. (so if the function is linear no matter how many layers we stack, the overall function is still linear and we could've used only one layer with certain weights to build it).

So in general we use non-linear activation functions (like ReLU, Sigmoid, Tanh, etc..) to be able to build more complex models that can capture non-linear relationships between inputs and outputs.

Below: if we apply a linear activation function (i.e. no activation function), and in the end the sigmoid to map the output between 0 and 1, the overall function is still linear (left), we can at most separate the two classes with a line. If we apply a non-linear activation function (right), we can build a more complex function (*decision boundary*) that can separate the two classes better.

### Multi-layer perceptron models
A deep learning model made of multiple layers of perceptrons is called a **multi-layer perceptron (MLP)**. 

An interesting property of MLPs is in the **Universal approximation Theorem**: for any continuous function g defined on a compact subset of R^n and for any ε > 0, there exists a feedforward neural network with a single hidden layer and a finite number of neurons (perceptrons) that can approximate g within an arbitrary degree of accuracy ε.

(sarebbe a dire, usando un MLP con una singola hidden layer e un numero finito di neuroni, possiamo approssimare qualsiasi funzione continua definita in un sottoinsieme compatto di R^n*°* con una precisione arbitraria ε).
*°*NB. questa funzione è quella complessiva, diversa da quelle di attivazione definite di R in R, per questo definita su R^n.
Inoltre il sottinsieme compatto di R^n è un insieme chiuso e limitato in R^n, serve per garantire che la funzione g sia ben definita e limitata all'interno di questo insieme ed evitare problemi di divergenza all'infinito.

Ci sono due "problemi" con questo teorema:
1. Non dice nulla sul numero di neuroni necessari per ottenere una certa precisione ε. In pratica, per approssimare funzioni complesse, potremmo aver bisogno di un numero molto grande di neuroni, rendendo il modello computazionalmente costoso e difficile da addestrare.
2. Non dice nulla su come trovare i pesi e i bias del modello per ottenere l'approssimazione desiderata.

Quello che si fa generalmente, dato che usare un solo layer potrebbe richiedere un numero enorme di neuroni, **è usare più hidden layers (da qui deep learning) con un numero minore di neuroni per layer**. In questo modo si possono costruire modelli più efficienti e capaci di catturare relazioni complesse tra input e output.

### Activation functions for classification
We talked only about the sigmoid activation function for binary classification, but that doesn't work for multi-class classification (where we have more than two classes), since we need to output a probability distribution over all classes (i.e. the sum of probabilities for all classes must be equal to 1).  
ex. given an image, we want to classify it as either a cat, dog or bird. The output of the model should be a vector of three probabilities, one for each class, that sum up to 1.

So the sigmoid is no more useful here, since it outputs a single value between 0 and 1 (and we can't expect that the sum of multiple sigmoids is equal to 1).

We define as **logit** the output of the model before applying the activation function (i.e. the weighted sum of inputs, in our case before applying the softmax function which will turn the logits into probabilities). Note that of course the logit is only defined in the last layer of the model.

<img src="img_teoria/logit.png" alt="Logit Definition" width="400"/>

If we have a binary classification, then we have only two outputs (one for each class), and we can use the sigmoid activation function to turn the logit into a probability for the positive class (the probability for the negative class is just 1 minus that).  
We can think in this case of the model as something that takes the input features x and produces a logit z to which we apply the sigmoid function to get the probability of the positive class:

$ \sigma(z) = \frac{1}{1 + e^{-z}} = Pr(positive|x) $

This works since the probability of negative class is just $1-Pr(positive|x)$ -> both are in [0, 1] and sum up to 1.

For multi-class classification, we need to output a probability distribution over all n classes. The number of logits, in this case, is equal to the number of classes and can be seen as a vector $\mathbf{z} = [z_1, z_2, \ldots, z_n]$. To turn the logits (which could be any number, they are in $(-\infty, +\infty)$) into probabilities, we use the **softmax activation function**:

$ softmax(z_i) = Pr(c_i|x) = \frac{e^{z_i}}{\sum_{j=1}^{n} e^{z_j}} $  for i = 1, 2, ..., n

(i.e. for each class i, the probability of class i given input x is equal to the exponential of the logit for class i divided by the sum of exponentials of all logits. Note that exponentiation guarantees that the output is positive, and dividing by the sum ensures that the outputs sum to 1 and the number is in (0, 1)).

With softmax, it can be easily verified that:
- $ Pr(c_i|x) \in [0, 1] $ for all i
- $ \sum_{i=1}^{n} Pr(c_i|x) = 1 $

### Activation functions for regression
In regression problems, we typically want the output of the model to be a continuous value that can take any real number (i.e. in $(-\infty, +\infty)$). Therefore, since perceptrons natively produce as output real numbers (the weighted sum of inputs), there is no need to enforce any specific property on the output, hence we can use a **linear activation function** (i.e. no activation function at all):
$ f(x) = x $

NB. l'uso della linear activation function non implica che il modello sia lineare, perché in questo caso come activation function si intende la funzione usata solo nell'ultimo layer del modello. Infatti, anche usando una linear activation function nell'output layer, se il modello ha più hidden layers con funzioni di attivazione non lineari, l'intero modello può essere non lineare.

### Defining weights and bias
Let's call the weights and biases of perceptrons as $\theta$. So far, we assumed they were known. We need a way to find the best values for them.

The approach is similar to what we did for linear regression: *we define a loss function that measures how well the model is performing (i.e. how close the predictions are to the true values), and then we use optimization algorithms (like gradient descent) to find the values of $\theta$ that minimize the loss function*.

So the idea is to pick a function (objective, or loss), let's call it $\mathcal{L}(\theta)$, that we want to minimize, and find $\theta$ that minimizes it. In linear regression we used MSE since we proved that it is a quadratic function of the weights, hence it has a single global minimum that can be found analytically. In deep learning, however, the loss functions are generally non-convex (due to the non-linear activation functions and the complex structure of the models), hence we cannot find the minimum analytically. Therefore, we use iterative optimization algorithms like gradient descent to find a local minimum of the loss function.

So, with deep learning (since we don't have a closed form solution as in linear regression), the idea is to try a bunch of different values for $\theta$ iteratively following an algorithm, compute the loss function for each set of values, and update $\theta$ in the direction that reduces the loss function the most (i.e. the negative gradient direction for gradient descent), until we reach a point that we consider a "good solution" (i.e. a local minimum of the loss function or a solution that satisfies some stopping criteria).

We start with a random $\theta$ (i.e. random weights and biases), and then "move around" with "some policy".

With "move around", we mean that we update the weights and biases ($\theta$) iteratively:

$ \theta_{t+1} = \theta_t + update $

With "some policy", we mean that we take a small step in the direction where the function decreases locally, i.e. the negative gradient direction (we choose the gradient since in one dimension the gradient is just the derivative, which tells us the slope of the function at that point, and the negative gradient tells us the direction where the function decreases):

$ update = - \alpha \nabla_{\theta} \mathcal{L}(\theta_t) $

That is exactly how **gradient descent** works, where $\alpha$ is the *learning rate* (a small positive number that controls the step size, we don't want to make it too large because we could overshoot the minimum and diverge, and not too small because we could take too long to converge).

The main limitation of gradient descent is **sensitivity to weight initialization**: if we start with a bad initialization, we could end up in a poor local minimum or take too long to converge.
(in the example, if we start from that point, the minus gradient goes in the direction in which the function decreases -> we end up in a local minimum that is not the global minimum).

<img src="img_teoria/sfigat.png" alt="Gradient Descent Sensitivity to Initialization" width="400"/>

To help and prevent local minima or too slow convergence, we can use some techniques (just introduced here, will be covered in depth in other courses):
- **Adaptive learning rates**: instead of using a fixed learning rate, we can adapt it during training based on the progress made. For example, we can decrease the learning rate when the loss function stops decreasing significantly.
- **Adding momentum**: instead of stopping at the current gradient, we can accumulate a velocity vector that takes into account past gradients, allowing us to escape shallow local minima.
- **Learning rate schedules**: we can define a schedule to decrease the learning rate over time, allowing for larger steps at the beginning of training and smaller steps as we approach convergence.

#### Backpropagation
So far we assumed we were able to compute the gradient of the loss function with respect to the weights and biases ($\nabla_{\theta} \mathcal{L}(\theta)$). But how do we compute it efficiently?

The idea is to use **backpropagation**, which is an efficient algorithm to compute the gradients of the loss function with respect to the weights and biases of the model. It is based on the **chain rule of calculus**, which allows us to compute the gradient of a composite function by multiplying the gradients of its components.  
In general the chain rule states that if we have a function $h(x) = f(g(x))$, then the derivative of h with respect to x is given by:
$ h'(x) = f'(g(x)) \cdot g'(x) $  
The idea is to apply this rule from the end of the computational graph (i.e. the loss function) back to the beginning (i.e. the weights and biases).

A computational graph is a directed graph where each node represents an operation (like addition, multiplication, activation function, etc.), and the edges represent the flow of data between the nodes. The computational graph allows us to visualize the sequence of operations that are performed to compute the output of the model given the inputs and the parameters. A simple example below, where we want to compute $y = wx + b$. First we compute the product $wx$, then we add the bias b to get the final output y.

<img src="img_teoria/comp_graph.png" alt="Computational Graph Example" width="400"/>

The backpropagation algorithm consists of two main steps:

1. **Forward Pass**: Compute the output of the network and the loss, by building the computational graph.
2. **Backward Pass**: Compute the gradients of the loss with respect to the weights and biases using the chain rule.

During the forward pass, we compute the output of each layer and store the intermediate values (activations) for use in the backward pass (sarebbe a dire si salvano gli output delle operazioni necessarie per applicare la chain rule). During the backward pass, we start from the output layer and propagate the gradients backward through the network, layer by layer, using the chain rule to compute the gradients of the loss with respect to the weights and biases.

This process allows us to efficiently compute the gradients for all parameters in the network with a single backward pass, making it much more efficient than computing the gradients individually for each parameter.

Let's see an example: imagine to have dataset with one point (x, y) and a (weird) model with two parameters ($\theta_1$ and $\theta_2$), that predicts $\theta_1 \theta_2 x$. We want to minimize the MSE loss function:
$ \mathcal{L} = (\theta_1 \theta_2 x - y)^2 $

So first we do the forward pass, by building the computational graph with all the intermediate values:
- a = $\theta_1 \theta_2$
- b = $a x$ = $\theta_1 \theta_2 x$
- c = $b - y$ = $a x - y$ = $\theta_1 \theta_2 x - y$
- $\mathcal{L} = c^2$ = $(b - y)^2$ = $(a x - y)^2$ = $(\theta_1 \theta_2 x - y)^2$

We want to compute $\frac{\partial \mathcal{L}}{\partial \theta_1}$ and $\frac{\partial \mathcal{L}}{\partial \theta_2}$, in order to compute the gradients and update the parameters using gradient descent.

After the forward pass, we do the backward pass, by applying the chain rule to compute the gradients:
- $\frac{\partial \mathcal{L}}{\partial c} = 2c$
- $\frac{\partial c}{\partial b} = 1$
- $\frac{\partial b}{\partial a} = x$
- $\frac{\partial a}{\partial \theta_1} = \theta_2$
- $\frac{\partial a}{\partial \theta_2} = \theta_1$

Using the chain rule, we can compute the gradients:

$ \frac{\partial \mathcal{L}}{\partial \theta_1} = \frac{\partial \mathcal{L}}{\partial c} \cdot \frac{\partial c}{\partial b} \cdot \frac{\partial b}{\partial a} \cdot \frac{\partial a}{\partial \theta_1} = 2c \cdot 1 \cdot x \cdot \theta_2 = 2(\theta_1 \theta_2 x - y) x \theta_2 $

$ \frac{\partial \mathcal{L}}{\partial \theta_2} = \frac{\partial \mathcal{L}}{\partial c} \cdot \frac{\partial c}{\partial b} \cdot \frac{\partial b}{\partial a} \cdot \frac{\partial a}{\partial \theta_2} = 2c \cdot 1 \cdot x \cdot \theta_1 = 2(\theta_1 \theta_2 x - y) x \theta_1 $

### Loss functions

#### Loss functions for regression
For Regression, we have already seen that we can use as loss function the Mean Squared Error (MSE) and the Mean Absolute Error (MAE):
- $ \mathcal{L}_{MSE} = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2 $  

- $ \mathcal{L}_{MAE} = \frac{1}{N} \sum_{i=1}^{N} |y_i - \hat{y}_i| $  

where N is the number of data points, $y_i$ is the true value for point i, and $\hat{y}_i$ is the predicted value for point i.

#### Loss functions for classification
Slightly more interesting are the loss functions for Classification problems.

For binary classification, we want a function that gives us a large number if the prediction is bad and a small number if the prediction is good. A commonly used loss function for binary classification is the **Binary Cross-Entropy Loss**. We first consider:
- y = true label (0 or 1), "ground truth"
- $\hat{y}$ = predicted probability of the positive class (output of the sigmoid function, in [0, 1])

Then, the Binary Cross-Entropy Loss is defined as:
$ \mathcal{L}_{BCE} = -[y \cdot \log(\hat{y}) + (1 - y) \cdot \log(1 - \hat{y})] $
So the ground truth y acts as a switch of the loss term to be applied:
- If y = 1 (positive class), the loss is $ -\log(\hat{y}) $: this means that if the predicted probability $\hat{y}$ is close to 1 (correct prediction), the loss will be small, while if $\hat{y}$ is close to 0 (wrong prediction), the loss will be large.
- If y = 0 (negative class), the loss is $ -\log(1 - \hat{y}) $: this means that if the predicted probability $\hat{y}$ is close to 0 (correct prediction), the loss will be small, while if $\hat{y}$ is close to 1 (wrong prediction), the loss will be large.

<img src="img_teoria/bce.png" alt="Binary Cross-Entropy Loss" width="200"/>

For multi-class classification, we can use the **Categorical Cross-Entropy Loss**. It is a generalization of the binary cross-entropy loss to multiple classes. We first consider:
- $y_i$ = true label for class i (1 if the true class is i, 0 otherwise)
- $\hat{y}_i$ = predicted probability for class i (output of the softmax function, in [0, 1])

Then, the Categorical Cross-Entropy Loss is defined as:
$ \mathcal{L}_{CCE} = - \sum_{i=1}^{n} y_i \cdot \log(\hat{y}_i) $
where n is the number of classes.  
Again, the ground truth $y_i$ acts as a switch: only the term corresponding to the true class contributes to the loss, while all other terms are zero.
