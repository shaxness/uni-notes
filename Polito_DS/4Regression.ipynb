## Regression
Differently from Classification, the idea with regression is to predict a continuous value instead of a discrete label (ex. predict the price of a house instead of whether it is expensive or cheap).

In general, we want to predict a continuous outcome variable (also called **dependent variable** or target variable) based on one or more predictor variables (also called **independent variables** or features), i.e. we learn a mapping (**function**) from input features to a continuous output $f: X \rightarrow Y$ where $Y \in \mathbb{R}$.

Regression is useful not only for predicting continuous values, but also for understanding relationships between variables and identifying significant predictors.

### Linear Regression
Linear Regression is the simplest form of regression, it is used to model the relationship between a dependent variable and one or more independent variables by fitting a linear equation to observed data.

#### Simple Linear Regression
Per ora focalizziamoci sul caso di una singola variabile indipendente x (**Simple linear regression**).  
Si assume che tra x (variabile indipendente) e y (variabile dipendente) ci sia una relazione lineare, che può essere rappresentata dall'equazione della retta:
$ y = \theta_0 + \theta_1 x + \epsilon $
dove $\theta_0$ è l'intercetta (valore di y quando x=0), $\theta_1$ è il coefficiente angolare (indica quanto y cambia per ogni unità di cambiamento in x) e $\epsilon$ è l'errore o residuo, che rappresenta la "parte casuale" che il modello non riesce a spiegare usando una relazione lineare (rumore nei dati, i record che raccogliamo e utilizziamo per loro natura non seguono perfettamente la relazione lineare ma sono "sparsi" attorno alla retta).

L'obiettivo della regressione lineare è stimare i parametri $\theta_0$ e $\theta_1$, per poi fare la previsione di y dato un nuovo valore di x come:
$ \hat{y} = \hat{\theta_0} + \hat{\theta_1} x $.
Il valore del residuo $\epsilon$ rappresenta quindi la differenza tra il valore osservato di y e il valore predetto $\hat{y}$.

In generale per un buon modello di regressione lineare, i residui dovrebbero essere:
- **Normally Distributed** (seguire approssimativamente una distribuzione normale con media 0). Ciò significa che quindi la maggior parte dei residui dovrebbe essere vicina a 0, con pochi valori estremi.
- **Homoscedastic**: ossia la varianza dei residui dovrebbe essere costante per tutti i valori di x. In altre parole, la dispersione dei residui attorno alla retta di regressione dovrebbe essere simile per tutti i valori di x.

Data una serie di dati $(x_i, y_i)$ per $i=1,2,...,n$, possiamo stimare i parametri $\theta_0$ e $\theta_1$ utilizzando il metodo dei minimi quadrati, che minimizza la somma dei quadrati dei residui (differenza tra i valori osservati e quelli predetti):
$ MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y_i})^2 $

Utilizzare lo squared error è utile per diversi motivi:
- penalizza maggiormente gli errori più grandi (differenze più grandi tra y e $\hat{y}$) rispetto a quelli più piccoli. Se l'errore è <= 1, lo squared error è ancora più piccolo, mentre se l'errore è > 1, lo squared error diventa molto più grande.
- è matematicamente conveniente, in quanto è differenziabile
- sempre positivo, quindi non ci sono problemi di segno
- quantifica la qualità complessiva del modello, più è basso più il modello è accurato

Dal momento che 
$ MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - (\theta_0 + \theta_1 x_i))^2 $ è funzione quadratica in $\theta_0$ e $\theta_1$, questa ha un solo minimo globale. Quindi
possiamo calcolare le derivate parziali rispetto a $\theta_0$ e $\theta_1$, porle uguali a 0 e risolvere il sistema di equazioni per trovare facilmente i valori ottimali di $\theta_0$ e $\theta_1$ che minimizzano l'MSE. Si ottengono le seguenti formule:
- $ \hat{\theta_0} = \bar{y} - \hat{\theta_1} \bar{x} $

- $ \hat{\theta_1} = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n} (x_i - \bar{x})^2} $

Dove $\bar{x}$ e $\bar{y}$ sono le medie di x e y rispettivamente.

#### Multivariate Linear Regression
Quando abbiamo più di una variabile indipendente, parliamo di **Multivariate Linear Regression**.
In questo caso non si parla più di retta ma di iperpiano (o ipersuperficie) che cerca di adattarsi ai dati in uno spazio multidimensionale.

Sia $X = [x_1, x_2, ..., x_n]$ l'insieme delle n variabili indipendenti (features), y la variabile dipendente, e $\hat{y}$ la previsione della variabile dipendente, la relazione lineare può essere rappresentata come:
$ \hat{y} = \theta_0 + \theta_1 x_1 + \theta_2 x_2 + ... + \theta_n x_n $
o, in forma vettoriale, con il prodotto scalare:
$ \hat{y} = \theta^T X$ 

In questo caso, la soluzione per stimare i parametri $\theta$ che minimizzano l'MSE è data dalla formula:
$ \hat{\theta} = (X^T X)^{-1} X^T Y $
(qui Y non è un singolo valore ma il vettore delle osservazioni della variabile dipendente, ossia Y = [y1, y2, ..., ym] dove m è il numero di record).

(qui sotto l'MSE nel caso multivariabile)

<img src="img_teoria/multivariate_linear_regression.png" alt="multivariate linear regression" width="400"/>

Si osservi come i coefficienti $\theta_i$ aiutano a capire le relazioni esistenti tra variabili indipendenti e dipendenti. In particolare, essi rappresentano l'effetto marginale di ciascuna variabile indipendente sulla variabile dipendente, mantenendo costanti tutte le altre variabili indipendenti. Ad esempio, $\theta_1$ indica quanto ci si aspetta che y cambi per ogni unità di aumento in $x_1$, assumendo che tutte le altre variabili indipendenti rimangano costanti.


#### Non-Linear Relationships
In molti casi, la relazione tra le variabili indipendenti e la variabile dipendente potrebbe non essere lineare. In questi casi, un approccio comune è aggiungere nuove feature che rappresentano trasformazioni non lineari delle variabili indipendenti originali (es. polinomiali, logaritmiche, esponenziali, ecc.), per poi utilizzare la regressione lineare su questo nuovo insieme di feature trasformate.

E.g. se sappiamo che la relazione tra x e y è inversa e quadratica, possiamo aggiungere una nuova feature $\frac{1}{x^2}$ e utilizzare la seguente equazione:
$ \hat{y} = \theta_0 + \theta_1 x + \theta_2 \frac{1}{x^2} $

Funziona perché i punti vengono mappati in uno spazio multidimensionale (con una nuova dimensione, nel nostro caso), dove la relazione tra le nuove feature e y può essere approssimata come lineare.


#### Residual Plots 
I residual plots sono uno strumento per valutare se una regressione lineare è appropriata per i dati. Essi rappresentano un grafico dove:
- Sull'asse x si hanno i valori predetti dal modello ($\hat{y}$) oppure i valori della variabile indipendente (x) (come nel nostro caso)
- Sull'asse y si hanno i residui (differenza tra i valori osservati e quelli predetti, $y - \hat{y}$)

Un buon residual plot dovrebbe mostrare residui **sparsi casualmente intorno a 0**, senza mostrare forme, pattern o strutture evidenti. La loro varianza deve essere costante (**homoscedasticity**). Nel caso in cui ciò si verifichi, allora il modello di regressione lineare è appropriato per i dati.

Al contrario, come si vede nel grafico in alto a destra, se i residui mostrano un pattern (es. una curva), allora il modello di regressione lineare non è adatto, in quanto non riesce a catturare la relazione tra le variabili indipendenti e la variabile dipendente.

### Polynomial Regression
Nel caso in cui non sappiamo la forma della relazione tra le variabili indipendenti e la variabile dipendente, possiamo utilizzare la **Polynomial Regression**. In questo caso non si vuole più predire y come una combinazione lineare delle feature (usando una retta), ma come una combinazione polinomiale di esse (usando un polinomio).

Quindi, nel simple case con una sola feature x, avremo un'equazione del tipo:
$ \hat{y} = \theta_0 + \theta_1 x + \theta_2 x^2 + ... + \theta_d x^d $

Per problemi multivariati, possiamo aggiungere potenze polinomiali di ciascuna feature e anche termini di interazione tra le feature per catturare relazioni più complesse tra le variabili indipendenti e la variabile dipendente!  
E.g. Con 3 feature x1, x2, x3 e un polinomio di grado 2, avremo: $x1^2, x2^2, x3^2, x1*x2, x1*x3, x2*x3$ come nuove feature.

Quando si usa Polynomial Regression, bisogna quindi porre attenzione al numero di feature che si generano, esse infatti crescono "combinatorially", ossia esponenzialmente/fattorialmente, con il numero di feature originali e il grado del polinomio scelto. (ogni volta che aggiungo una feature o un grado in più, il numero di nuove feature cresce esponenzialmente).

### Evaluation Metrics for Regression
Le principali metriche per valutare le performance di un modello di regressione sono:
- **Mean Squared Error (MSE)**: misura la media dei quadrati delle differenze tra i valori osservati e quelli predetti. Formula: $ MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y_i})^2 $  
A volte potremmo trovarlo non normalizzato -> **Sum of Squared Errors (SSE)**.
- **Root Mean Squared Error (RMSE)**: è la radice quadrata dell'MSE, fornisce una misura dell'errore nella stessa unità di misura della variabile dipendente. Formula: $ RMSE = \sqrt{MSE} $ (simile al ragionamento che si fa per lo standard deviation rispetto alla variance).
- **Mean Absolute Error (MAE)**: misura la media delle differenze assolute tra i valori osservati e quelli predetti. Formula: $ MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y_i}| $  
A differenza di MSE, MAE penalizza gli errori piccoli "di più", ed è meno sensibile agli outlier.
- **R-squared (R²)**: rappresenta la proporzione della varianza nella variabile dipendente che è spiegata dalle variabili indipendenti nel modello. Formula: $ R^2 = 1 - \frac{MSE}{\sigma^2} $ dove $\sigma^2$ è la varianza dei valori osservati di y. 
1. Se MSE = 0 (predizioni perfette), allora R² = 1. Questo è l'upper bound di R².
2. Se il modello predice sempre la media di y (cioè non usa le feature per fare previsioni), allora MSE = $\sigma^2$**°** e R² = 0.
3. Se il modello è peggiore di predire sempre la media di y, allora MSE > $\sigma^2$ e R² < 0.

**°**(perché sigma^2 è la varianza dei valori osservati esistente di base, MSE usando la media per ogni valore è esattamente la definizione di varianza campionaria $\sigma^2 = \frac{1}{n} \sum_{i=1}^{n} (y_i - \bar{y})^2$)

### Regularization
To understand regularization, we first need to understand the concept of **overfitting** and **underfitting**:
- **Overfitting**: the model is too complex and fits the training data too closely -> *high variance*. High overfitting means poor performance on test data, and may occur when using higher degree polynomials. In the example below, the polynomial regression of degree 15 perfectly fits the training data, but fails to generalize to new data points (test set), resulting in poor performance. 
- **Underfitting**: the model is too simple and fails to capture the underlying trend of the data -> *high bias*. Underfitting means poor performance on both training and test data, and may occur when using too low degree polynomials. In the example below, the polynomial regression of degree 1 fails to capture the trend of the data, resulting in poor performance.

It differs a lot, from little to very high, depending on the training data used.

On the other hand we define underfitting as **high bias** because the model is too simple to capture the underlying patterns in the data. In the image below, on the left, we see how the model consistently fails to capture the trend of the data, resulting in high error regardless of the training data used, and all these errors are similar to each other.

<img src="img_teoria/bias_variance.png" alt="bias and variance" width="600"/>

In general, we can prevent overfitting by:
- **Reducing model capacity**: using simpler models with fewer parameters (e.g., lower degree polynomials)
- **Increasing the dataset size**: more data can help the model generalize better
- **Using Regularization techniques**: adding a penalty term to the loss function to discourage overly complex models

Of course reducing the model capacity too much can lead to underfitting, and not always having more data is possible, so regularization is often the most practical approach to control overfitting.

#### Regularization techniques
Prima di capire la regularization, è importante capire che un modello troppo complesso (che potrebbe quindi causare overfitting) è un modello che non solo utilizza molte feature (high degree polynomials), ma che ha anche **coefficienti molto grandi** (in valore assoluto) associati a queste feature (le vediamo nelle figure sopra nei casi di overfitting: un coefficiente grande rappresenta una pendenza molto ripida, quelle "parabole" che vediamo nei casi di overfitting sono dovute a coefficienti molto grandi).

The idea behind regularization is to *add a penalty term to the loss function* (e.g., MSE) that discourages overly complex models by penalizing large coefficients. This helps to prevent overfitting by constraining the model's complexity.  
Quindi non si minimizza più solo l'MSE, ma si minimizza l'MSE + un termine di penalizzazione (regularization term) -> si scoraggia il modello ad utilizzare coefficienti troppo grandi.

Esistono due principali tecniche di regularization:
- **Lasso Regression (L1 Regularization)**: aggiunge un termine di penalizzazione basato sulla somma dei valori assoluti dei coefficienti. La funzione di perdita diventa:
$ Loss = MSE + \lambda ||\theta||_1 $
- **Ridge Regression (L2 Regularization)**: aggiunge un termine di penalizzazione basato sulla somma dei quadrati dei coefficienti. La funzione di perdita diventa:
$ Loss = MSE + \lambda ||\theta||_2 $



Dove $\lambda$ è un iperparametro che controlla la forza della penalizzazione (maggiore è $\lambda$, maggiore è la penalizzazione per i coefficienti grandi).  

Per Lasso, usando norma 1 (somma dei valori assoluti degli elementi), la penalizzazione è lineare, portando quindi i valori di $\theta$ a diventare 0 nel caso in cui non sono importanti per la predizione.  (in pratica, il modello riconosce che una certa variabile $x_j$ non aiuta a predire y, e quindi invece di tenerla con un peso piccolo lo elimina ($\theta_j = 0$)).

Per Ridge, usando norma 2 (somma dei quadrati degli elementi, tutto sotto radice), la penalizzazione è quadratica, quindi i coefficienti vengono ridotti ma non portati a 0. Questo significa che tutte le feature rimangono nel modello, ma con pesi più piccoli, il che aiuta a ridurre l'overfitting senza eliminare completamente nessuna feature.

(il perché Lasso fa feature selection (i.e. assegna 0 a certe feature) mentre Ridge no, lo si può capire osservando la forma geometrica delle due penalizzazioni: la norma 1 (Lasso) ha una forma a "diamante" con angoli acuti, mentre la norma 2 (Ridge) ha una forma circolare. Quando si minimizza la funzione di perdita con queste penalizzazioni, la soluzione ottimale per Lasso tende a cadere sugli assi (dove alcuni coefficienti sono esattamente 0), mentre per Ridge tende a cadere all'interno della regione circolare, mantenendo tutti i coefficienti non nulli.) -> motivo geometrico, non serve saperlo

### Other Regressors 
Possiamo costruire Decision Trees per la regressione, chiamati appunto **Decision Tree Regressors**. Essi funzionano in modo simile ai Decision Tree per la classificazione, ma invece di predire una classe, predicono un valore continuo.

L'idea è quindi la stessa dei decision tree: si dividono ricorsivamente i dati in modo da formare gruppi il più omogenei possibile rispetto al valore della variabile dipendente (y). Ad ogni passo si sceglie quindi una feature (ex. X[2]) e un valore di split (ex. 2.45). I dati saranno divisi in due gruppi: a sx quelli con X[2] <= 2.45 e a dx quelli con X[2] > 2.45. Si continua a dividere i dati in questo modo fino a quando non si raggiungono certi criteri di arresto (es. profondità massima dell'albero, numero minimo di campioni in un nodo, ecc.).

La cosa diversa dal decision tree classico è che stavolta non si calcola più la node impurity tramite Gini o Entropy, ma si considera la Node Impurity come la varianza dei valori di y nei dati presenti in quel nodo. L'obiettivo è quindi minimizzare la varianza all'interno di ciascun nodo figlio dopo lo split, in modo da avere nodi il più omogenei possibile rispetto ai valori di y.

Arrivati alle foglie, **la predizione per un nuovo campione che arriva in quella foglia sarà la media dei valori di y dei campioni presenti in quella foglia**.

Quindi in pratica lo split migliore è quello che minimizza la somma delle varianze dei due nodi figli (si vuole minimizzare l'MSE dei due nodi figli).

Ulteriori tecniche di regressione includono:
- **Random Forest Regressors**: semplicemente una collezione di decision tree regressors, dove la predizione finale è la media delle predizioni di tutti gli alberi nel forest. Questo aiuta a ridurre l'overfitting rispetto a un singolo albero.
- **KNN Regressors**: l'idea è simile a KNN per la classificazione, ma invece di predire la classe più comune tra i k-nearest neighbors, si predice la media dei valori target dei k-nearest neighbors (vicini più prossimi). Questa media può essere calcolata come media semplice o come media pesata, dove i vicini più vicini hanno un peso maggiore nella media rispetto a quelli più lontani (esattamente come per KNN classification)
- **Neural Networks**: le reti neurali predicono valori continui nativamente, come vedremo.
