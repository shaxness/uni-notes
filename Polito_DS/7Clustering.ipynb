## Introduction to Clustering (Definitions)

Still talking about unsupervised learning, we are not talking about association rules but about the will of grouping data (points) based on their similarity in clusters.

Ideally, we'd like the intra-cluster distances minimized and the inter-cluster distances maximized.

<img src="img_teoria/clusters.png" alt="clustering" width="400"/>

Using clustering is useful to understand the structure of the data, to summarize and compress it (reduce the size of large datasets by working separately only in clusters of them), to detect outliers (points that do not belong to any cluster).

Notion of a cluster can be ambiguous (given a set of points, how many clusters are there? Which points belong to which cluster? Usually there is not an evident answer). For these reasons, different clustering algorithms can give different results on the same dataset, also we'll see that evaluating the quality of a clustering is not trivial and there are different metrics based on the context (similar to classification problems, but here is more difficult because we don't know what we want to predict).

We define as a **Clustering** a set of clusters, where each cluster is a subset of the data points.  
It is important to distinguish to kinds of clustering:
- **Partitional Clustering**: each data point belongs to exactly one cluster (hard clustering). So the data objects are divided in non-overlapping clusters.
- **Hierarchical Clustering**: data points can belong to multiple clusters (soft clustering). Clusters are organized in a hierarchy, where each cluster can have sub-clusters. These clusters are often represented as a tree (*dendrogram*).

Some other definitions:
- **Exclusive vs Non-Exclusive Clustering**: in exclusive clustering, each data point belongs to exactly one cluster. In non-exclusive clustering, data points can belong to multiple clusters. (it's similar to partitional vs hierarchical, but more general)
- **Fuzzy vs non-Fuzzy Clustering**: in fuzzy clustering, each data point can belong to multiple clusters with a certain weight between 0 and 1. Each of the weight for a single point must sum up to 1. In non-fuzzy clustering, data points either belong to a cluster or not (binary membership).
- **Partial versus Complete Clustering**: in partial clustering, some data points can be left unclustered (not assigned to any cluster). In complete clustering, all data points must be assigned to a cluster.
- **Heterogeneous vs Homogeneous Clustering**: in heterogeneous clustering, clusters can have different shapes, sizes, and densities. In homogeneous clustering, clusters are assumed to have similar characteristics.

### Types of Clusters
Going back to the fact that clustering is ambiguous, we can distinguish between many types of clusters:

- **Well separated clusters**: clusters that are clearly separated from each other, with a large distance between them. Each point is closer to the points in its own cluster than to points in other clusters. This is the ideal case of clustering, and by applying any clustering algorithm we will be able to identify these clusters easily.

<img src="img_teoria/well_separated.png" alt="well separated clusters" width="300"/>

- **Center-based clusters**: clusters that are defined by a central point (typically the centroid) and a radius. Each point in the cluster is closer to the centroid than to any other centroid, so its a weaker assumption than well separated because now points of a cluster can be closer to points of other clusters than to points in their own cluster. The center of the cluster is often a **centroid**, which is the mean of all the points in the cluster.  
The centroid is not necessarily a point in the dataset, and for some algorithms we need an actual point as center, so the idea is to choose the point that is closest to the centroid as the center of the cluster (this point is called **medoid**, it is the most representative point of a cluster).  Center based clusters are of course spherical in shape.

<img src="img_teoria/center_based.png" alt="center based clusters" width="300"/>

- **Contiguous clusters** (AKA nearest neighbor clusters): clusters, in general, don't have to be necessarily be spherical. With contiguous clusters we define a cluster as a set of points s.t. a point in a cluster is closer to one or more points in the same cluster than to any point in other clusters. This allows for clusters of arbitrary shapes, as long as the points are connected to each other. Contiguous clusters can be identified using graph-based algorithms, where points are nodes and edges connect points that are close to each other. This will make more sense with hierarchical clustering.

<img src="img_teoria/contiguous.png" alt="contiguous clusters" width="300"/>

- **Density-based clusters**: it is an extension of the previous concept of contiguous clusters. Here, a cluster is defined as a dense region of points separated by regions of low density. A point belongs to a cluster if it is in a region where the density of points is above a certain threshold. This allows for clusters of arbitrary shapes and sizes, and can also identify noise points that do not belong to any cluster. We will see an example of density-based clustering with DBSCAN. Guarda sotto l'immagine tipo calamita: mentre in continuity based si consideravano i punti che collegavano i due cerchietti, qua no perché la densità di punti tra i due cerchietti è bassa.

<img src="img_teoria/density_based.png" alt="density based clusters" width="300"/>

- Property or Conceptual clusters: clusters that are defined by some shared properties or concepts among the data points, rather than by their spatial proximity. (we already know the shape that the clusters should have, based on some prior knowledge of the data, and we use this knowledge to define the clusters). For example, in a dataset of animals, we might define clusters based on properties such as habitat, diet, or behavior, rather than their physical location.

<img src="img_teoria/property_based.png" alt="property based clusters" width="300"/>

- Described by an objective function: clusters that are defined by optimizing a certain objective function, such as minimizing the sum of squared distances within clusters or maximizing the separation between clusters


## Clustering Algorithms 

We'll discuss three algorithms for clustering:
- K-Means and its variants (partitional, it's very fast)
- DBSCAN (density-based)
- Hierarchical Clustering (agglomerative)

### K-Means
K-Means is a partitional clustering algorithm that aims to partition the data points into K clusters, where K is a user-defined parameter. With this algorithm each cluster is associated with a centroid. The algorithm works by iteratively assigning each data point to the nearest cluster centroid and then updating the centroids based on the assigned points. The process continues until convergence, which occurs when the centroids stabilize. A limitation of the algorithm is that the number of clusters K must be specified in advance, which can be challenging if the optimal number of clusters is not known. 

<img src="img_teoria/kmeans.png" alt="k-means clustering" width="400"/>

Some details about the algorithm: the initial centroids are often choosen randomly from the data points, but this could lead to problems and so there are more advanced methods like K-Means++. The centroid is typically calculated as the mean of all the points assigned to the cluster. The distance is generally calculated using Euclidean distance, which is convenient because the centroid is already the point that minimizes the sum of squared euclidean distances for all points in the cluster. Also, most of the convergence happens in the first few iterations, so usually the stopping condition is not "until reaching convergence" but "until relatively few points change clusters".  

The complexity of the algorithm is **O(n * k * I * d)**, where n is the number of data points, k is the number of clusters, I is the number of iterations until convergence, and d is the number of dimensions (features) in the dataset. In fact for each iteration we have to compute the distance between all points to all centroids, and since we have K centroids n*K, we must do this for each iteration I and to compute the distances we must compare all values for each dimension -> d number of attributes.  Differently from the other algorithms we'll see, K-Means is very fast (linear) and scalable to large datasets, but has some limitations (we'll see them later).

#### Limitations of K-Means (1)
**Choosing the starting centroids**: since the initial centroids are often chosen randomly, the algorithm can converge to different solutions depending on the initial choice. Let's see two different K-means clusterings:

<img src="img_teoria/kmean_limi1.png" alt="k-means limitations" width="400"/>

We could get both clusterings depending on the initial centroids, below we see the initialization that leads to the "bad" clustering:

<img src="img_teoria/kmean_limi2.png" alt="k-means limitations 2" width="300"/>

<img src="img_teoria/kmean_limi3.png" alt="k-means limitations 3" width="300"/>

This is because the centroids were initialized too close to each other, leading to a poor clustering.  We will see later how to evaluate if a clustering is good or bad, but for now lets define the most common measure: **Sum of Squared Errors (SSE)**.

**SSE** is defined as the sum of the squared distances between each data point and its assigned cluster centroid:

$ SSE = \sum_{i=1}^{k} \sum_{x \in C_i} dist^2(m_i, x) $

where k is the number of clusters, C_i is the set of points in cluster i, x is a data point in cluster i, and m_i is the centroid of cluster i. The second bad clustering of before has a higher SSE than the first good clustering, because the points are farther from their centroids.

**Solutions to Initial Centroids Problem**:
- Of course we can't test all possible initializations, but we can **run the algorithm multiple times with different random initializations** and choose the best result based on SSE. Running the algorithm multiple times helps since it increases the chances of finding a better clustering solution.
- Another idea is to use hierarchical clustering to **preprocess the data and find initial centroids**. We can use hierarchical clustering to group the data points into K clusters, and then use the centroids of these clusters as the initial centroids for K-Means. This can help to find better initial centroids that are more representative of the data. (nota però che questo approccio può essere computazionalmente costoso, specialmente per grandi dataset, poiché l'algoritmo gerarchico ha una complessità più elevata rispetto a K-Means).
- A more advanced method is **K-Means++**, which selects initial centroids in a way that spreads them out across the data space. The idea is to choose the first centroid randomly from the data points, and then choose the others to be "distant" from the prevuious ones.
- Do **Postprocessing**: after running K-Means, we can analyze the resulting clusters and merge or split them if necessary. For example, if two clusters are very close to each other, we can merge them into a single cluster. Conversely, if a cluster is too large or contains points that are far from the centroid, we can split it into smaller clusters.
- **Bisecting K-means**: it is another algorithm with the idea to iteratively build clusters by applying k-means only to 2 clusters at a time.

Before moving on, let's see how to **set the parameter K**. By computing the SSE for different values of K, we can plot the SSE against K and look for an "elbow" point where the decrease in SSE starts to level off. This point indicates a good trade-off between the number of clusters and the quality of the clustering. (if too many clusters -> SSE doesn't decrease much, if too few clusters -> SSE is high). Vedi fine file per capire meglio esattamente quale k scegliere.

Of course the graph must be computed for the specific dataset.

<img src="img_teoria/elbow_method.png" alt="elbow method" width="400"/>

**Handling Empty CLusters**: during the K-Means algorithm, it is possible that some clusters may end up with no data points assigned to them. To handle this situation, we must reassign the empty cluster's centroid to a new location. We could:
- choose as the new centroid the point that contributes the most to the overall SSE (the point that is farthest from its current centroid)
- choose a point from the cluster with the highest SSE 

Some more important things:
- **Preprocessing**: we need for k-Means to normalize the data before applying the algorithm, because K-Means is sensitive to the scale of the features. If the features have different scales, the distance calculations will be dominated by the features with larger scales, leading to biased clustering results. We also need to eliminate outliers, because they can significantly affect the centroids and lead to poor clustering results.
- **Postprocessing**: 
    - we can eliminate clusters that may represent noise or outliers, by setting a minimum size threshold for clusters. Clusters with fewer points than the threshold can be removed from the final clustering result. 
    - we can split "loose" clusters, i.e. custers with high SSE (variance) into smaller clusters with less SSE.
    - we can merge clusters that are "close" and have relatively low SSE when combined.

#### Limitations of K-Means (2)
K-means has problems when clusters differ in:
- **Sizes** 
- **Densities**
- **non globular shapes** (because K-means uses the mean of the points in a cluster as the centroid, so it "assumes" spherical clusters -> **kmeans tends to build globular clusters**)

### Hierarchical Clustering
It has a completely different approach compared to k-means. It builds a set of nested clusters organized as a hierarchical tree (dendrogram). We'll see that approaching clustering in a hierarchical way allows us to overcome some of the limitations of k-means (like non globular shapes, different sizes and densities of clusters), but it is computationally more expensive.

<img src="img_teoria/dendrogram1.png" alt="dendrogram" width="400px"/>

The idea to build that dendrogram: we start by seeing the closest pair of points in the dataset and we group them into a cluster (in our case 1,3). Then, we look for the next closest point or cluster and we group them together (2,5). Then the following pair is 4,5, but 5 is already in a cluster with 2, so we group 4 with the cluster (2,5). Then we see that 2 is closest to the cluster (1,3), so again a new cluster. Finally we have only 6 and group it with the rest. (this is just an idea... we'll see the actual algorithm later)

After having built the dendrogram, how can we extract a flat clustering from it? We can "cut" the dendrogram at a certain height, and all the subtrees below that height will be our clusters. The choice of the cutting height determines the number of clusters we obtain. For example, if we want to obtain 4 clusters, we can cut the dendrogram at a height that results in 4 subtrees. We would get in our case the clusters: (1,3), (2,5), (4), (6).

**So, if we obtain the dendrogram, we can get any clustering we want just by cutting it at the right height.** (while with k-means we must set k in advance and we get only one clustering).

<img src="img_teoria/cutt.png" alt="dendrogram cut" width="300px"/>

So the **strengths** of hierarchical clustering are:
- we don't have to assume any particular number of clusters in advance (any desired clustering can be obtained from the dendrogram by cutting it at the right height)
- they may correspond to meaningful taxonomies (ex. biological classifications, like animal kingdoms where species are grouped based on shared characteristics)

There exist two main types of hierarchical clustering:
- **Agglomerative Hierarchical Clustering**: it is a bottom-up approach, where we start with each data point as individual clusters. At each step, we merge the **two closest clusters** until we have a single cluster containing all data points or until a stopping criterion is met (ex. we arrived to a desired number of clusters k).
- **Divisive Hierarchical Clustering**: it is a top-down approach, where we start with all data points in a single cluster. At each step, we split the cluster into smaller clusters until each data point is in its own cluster or there are k desired clusters.

Tipically, agglomerative clustering is more commonly used due to its simplicity and effectiveness, and we'll focus on it.  Traditional hierarchical algorithms use a similarity or distance matrix to keep track of the distances between clusters. The distance between two clusters can be defined in different ways (as we'll see later).

The basic agglomerative hierarchical clustering algorithm works as follows:
1. Compute the proximity matrix for all data points (i.e. the distance between each pair of points).
2. Let each data point be its own cluster.
3. Repeat:
    - Merge the two closest clusters based on the proximity matrix, and update the proximity matrix to reflect the distances between the new cluster and the remaining clusters.
4. Until only one cluster remains (or until a stopping criterion is met, such as reaching a desired number of clusters k).

Here the **key operation is to compute the proximity between clusters**.

As said, we start with a n*n proximity matrix, where n is the number of data points. At each step, we merge two clusters and update the proximity matrix, which will be of lower dimension (n-1 * n-1) since now two points form a single cluster etc...
If we use a distance which is symmetric, then the proximity matrix is also symmetric, so we can store only half of it (upper or lower triangular matrix).

#### Inter-Cluster Distance Measures
Now the question is: how to compute the distance between two clusters? There are different methods:
- **MIN Distance** (aka *single linkage*): the distance between two clusters is defined as the minimum distance between any pair of points, one from each cluster. This method tends to produce long, thin clusters, as it focuses on the closest points between clusters. It is an "optimistic" approach, because it tells us that two clusters are close if there is at least one pair of points that are close, not considering the rest of the points in the clusters.
<img src="img_teoria/min_distance.png" alt="min distance" width="200px"/>

- **MAX Distance** (aka *complete linkage*): the distance between two clusters is defined as the maximum distance between any pair of points, one from each cluster. This method tends to produce compact, spherical clusters, as it focuses on the farthest points between clusters. It is a "pessimistic" approach, because it tells us that two clusters are far apart if there is at least one pair of points that are far apart, not considering the rest of the points in the clusters.
<img src="img_teoria/max_distance.png" alt="max distance" width="200px"/>

- **Group Average**: the distance between two clusters is defined as the average distance between all pairs of points, one from each cluster. This method balances the effects of MIN and MAX distances, and tends to produce clusters that are neither too long nor too compact.
<img src="img_teoria/avg_distance.png" alt="average distance" width="200px"/>

- **Centroid Distance**: the distance between two clusters is defined as the distance between their centroids (mean points).  
<img src="img_teoria/centroid_distance.png" alt="centroid distance" width="200px"/>

- **Other methods driven by an objective function**: there are other methods to compute inter-cluster distances, such as Ward's method, which uses squared error minimization to determine the distance between clusters (we'll see later).

##### Group Average
The proximity of two clusters with group average is the pairwise proximity average between all points in the two clusters.

<img src="img_teoria/group_averag1.png" alt="group average distance" width="300px"/>

The formula tells us that to compute the distance between two clusters A and B, we must compute the distance between each point in cluster A and each point in cluster B, sum all these distances and divide by the total number of pairs of points (i.e. the product of the number of points in cluster A and the number of points in cluster B).

We can see from the example how we get a in between behavior compared to min and max linkage. In fact we obtain cluster 3 that is the same obtained with max linkage, but then cluster 4 that is less globular, the same that we would obtain with min linkage at that point.

<img src="img_teoria/group_averag2.png" alt="group average example" width="300px"/>


##### Ward's Method
The idea here is that *similarity of two clusters is based on the increase in the overall squared error when the two are merged*. **So we want to merge the two clusters that will cause the smallest increase in the overall SSE** (i.e. the ones that give the least increasing variance). We remember that SSE is defined as the sum of squared distances between each data point and its assigned cluster centroid.

So we know that each cluster has a SSE, and when we merge two clusters A and B, the new cluster C will have a new SSE that is necessarily >= than the sum of the SSEs of A and B. We want to find the two clusters A and B such that the increase in SSE is minimized.

This approach is different from the others, we don't have to store a proximity matrix. Below an example of clustering using Ward's method, not how the cluster 4 is formed because it is the one that, once formed, minimizes the increase in SSE.

<img src="img_teoria/wards.png" alt="ward's method" width="400px"/>

##### Computational Cost
- **Space Complexity O(n^2)**: we need to store the proximity matrix, which has size n*n.
- **Time Complexity O(n^3)** in many cases: there are n steps (we must merge n-1 times to go from n clusters to 1), and in each step we must consult the proximity matrix to find the closest clusters (O(n^2)) and then update the proximity matrix (O(n), devo aggiornare solo una riga/colonna di lunghezza n), -> O(n^3).  
Time complexity can be reduced to **O(n^2 log n)** using more advanced data structures like priority queues.

So in general a superlinear complexity... for large datasets it's unfeasible to use hierarchical clustering directly. A solution could be to use hierarchical clustering on a sample of the data to build the dendrogram, a way to aggregate data could be by running k-means to build small clusters and then using hierarchical clustering on the centroids of these clusters. **So sometimes hierarchical clustering is used in combination with other clustering algorithms to improve scalability and efficiency.**

### DBSCAN
DBSCAN (Density-Based Spatial Clustering of Applications with Noise) is a density-based clustering algorithm that groups together data points that are closely packed together, while marking points that are in low-density regions as outliers. 

We remember that density-based clusters are defined as dense regions of points separated by regions of low density. A point belongs to a cluster if it is in a region where the density of points is above a certain threshold.

So, we define **Density** as the number of points within a certain radius (epsilon) around a point.  DBSCAN introduces three types of points:
- **Core Points** a point is a core point if there are more than a specified number of points (MinPts) within its eps. (con "its eps" si intende il cerchio di raggio eps centrato nel punto)
- **Border Points** a point is a border point if it has fewer than MinPts within its eps, but it is within the eps of a core point.
- **Noise Points** a point is a noise point if it is neither a core point nor a border point (so not enough points in its eps and not in the eps of a core point -> low density region).

So if a region has less than MinPts points in its eps, it is considered a low-density region and the points in that region are marked as noise points. We see an example below, where MinPts=4 and eps is the radius of the circles. Border Point has less than 4 points in its eps, but is within the eps of a core point, while Noise Point has less than 4 points in its eps and is not within the eps of any core point.

<img src="img_teoria/dbscan_points.png" alt="dbscan points" width="300px"/>

Based on these definitions, DBSCAN works as follows:
1. Eliminate all noise points from the dataset (**so DBSCAN performs a clustering only on core and border points**).
2. For each core point, if it has no cluster then assign it to a cluster (we label it with current_cluster_label)
   - For each point in its eps, if the point has no cluster, assign it to the current cluster

Quindi in pratica partiamo da un core point non ancora assegnato a nessun cluster, lo assegniamo ad un nuovo cluster, e assegniamo a quel cluster tutti i punti nel suo eps che non sono ancora stati assegnati a nessun cluster (sia core che border points). Dopodiché passiamo al prossimo core point:
- se questo non è ancora stato assegnato a nessun cluster, lo assegniamo ad un nuovo cluster e assegniamo a quel cluster tutti i punti nel suo eps che non sono ancora stati assegnati a nessun cluster. 
- Se invece il core point è già stato assegnato a un cluster, allora tutti i punti nel suo eps che non sono ancora stati assegnati a nessun cluster li assegniamo a quel cluster.

<img src="img_teoria/dbscan_alg.png" alt="dbscan algorithm" width="400px"/>

Nell'esempio sotto abbiamo chameleon dataset. In pratica DBSCAN procede "propagando" i cluster a partire dai core points, per cui ad esempio se parto da un core point in una regione densa dell'esempio assegnerò a tutti i punti nel suo raggio eps lo stesso cluster, dopodiché passerò al prossimo core point che se aveva lo stesso cluster allora assegnerà a tutti i punti nel suo eps lo stesso cluster, e così via finché non avrò assegnato a tutti i punti della regione densa lo stesso cluster. Dopodiché passerò alla prossima regione densa e ripeterò il procedimento, distinguendo così i vari cluster.

<img src="img_teoria/dbscan_clusters.png" alt="dbscan clusters" width="300px"/>

**Strengths of DBSCAN**:
- It is resistant to noise and outliers, as it explicitly identifies and marks them as noise points.
- It can handle clusters of arbitrary shapes and sizes, as it does not assume any specific cluster shape.

**DBSCAN Limitations**: it doesn't work well when we have points with different densities. This is because DBSCAN uses a fixed radius (eps) and a fixed minimum number of points (MinPts) to define dense regions. If the dataset contains clusters with varying densities, a single eps and MinPts may not be suitable for all clusters. In the example below, we could fix eps and minpts so that we have 3 big clusters on the up-right, or 3 smallest (but way more dens) clusters on the bottom-right.

<img src="img_teoria/dbscan_limitations.png" alt="dbscan limitations" width="400px"/>

There are ways to overcome this limitation, we could for example first use MinPts = 4 and Eps = 9.62 to identify the three high dense regions of points, and then "remove" them (we take them away for the moment, we determined with the first dbscan that they are clusters) and apply dbscan again with different parameters (that define density less strictly) to identify the clusters in the less dense regions. 

**So we can use DBSCAN in multiple passes with different parameters to identify clusters of varying densities.**

## Clustering Validation
Clustering validation is the process of evaluating the quality and effectiveness of clustering results. It is messy because we don't have a single definition of what a "good" clustering is (**cluster quality depends very much on what we are looking for), hence there are many different ways to evaluate clustering results.

For example, we could cluster random points with the algorithms seen before, and we would get clusters, which probably could get good results according to some metrics, but they would be meaningless clusters -> validation makes sense only if we have some idea of what we are looking for in the data.

<img src="img_teoria/validation.png" alt="clustering validation" width="300px"/>

So the numerical measures we use to evaluate clustering results can be divided into three main categories:
- **External Index**: these are used to measure the extent to which cluster labels match known, externally suppplied class labels. Sarebbe a dire che, per valutare il clustering, facciamo riferimento a delle etichette "vere" che sappiamo essere corrette (es. in un dataset di animali sappiamo che ci sono cani, gatti, uccelli etc... e vogliamo vedere se il clustering riesce a raggruppare correttamente questi animali). Quindi queste misure confrontano i risultati del clustering con queste etichette "vere" per valutare la qualità del clustering. (ex. purity, entropy ... perché misurano quanto i cluster ottenuti sono "puri" rispetto alle etichette vere).
**The limitation of external indices** is that they require the availability of ground truth labels, which may not always be available or reliable in real-world scenarios. Also, if we know the true labels, we may not need to perform clustering at all (because clustering is unsupervised learning, so we don't have labels, and also clustering is used to predict labels for new data points, so if we already have labels why don't we use classification algorithms directly?). NB with labels we mean class label, ossia l'etichetta di appartenenza di un elemento del dataset con le sue feature ad una classe (es. cane, gatto, uccello etc...). Con unsupervised learning abbiamo solo le feature, senza etichette di classe.
- **Internal Index**: these are used to evaluate the quality of clustering results **without** reference to external labels. Queste misure valutano la coesione all'interno dei cluster (quanto i punti in un cluster sono vicini tra loro) e la separazione tra i cluster (quanto i cluster sono distinti l'uno dall'altro). (ex. SSE, Silhouette coefficient, cluster separation, cluster cohesion ...)
- **Relative Index**: these are used to compare the quality of different clustering results on the same dataset. Queste misure confrontano i risultati di diversi algoritmi di clustering o diverse configurazioni dello stesso algoritmo per determinare quale produce i risultati migliori. Tipicamente per capire quali clustering sono migliori si usano Internal Index (es. SSE).

(Remember that external indices require ground truth labels, and because of that usually they are not applicable (if we had the labels we wouldn't need clustering, we'd use classification instead)).

An example about external measures: we observe that in the clustering, cluster three has 671 points labeled as "Sports" and only few points with other labels -> high purity and low entropy! Of course to compute these measures we need to know the true labels of the points. On the other hand cluster 4 has a mix of different labels -> lower purity and high entropy.

<img src="img_teoria/external_measures.png" alt="external measures example" width="400px"/>

### Internal Measures
- **Cluster Cohesion**: it measures how closely related the data points in a cluster are to each other. A common measure of cohesion is **Within Cluster Sum of Squares (WSS)**, which is the sum of squared distances between each point in a cluster and the centroid of that cluster. Lower WCSS values indicate higher cohesion, as points are closer to the centroid.  
Nella formula si fa la somma per ogni cluster della distanza quadratica di ogni punto del cluster dal suo centroide m_i.  
<img src="img_teoria/cohesion.png" alt="cluster cohesion" width="200px"/>  

- **Cluster Separation**: it measures how distinct or well-separated different clusters are from each other. A common measure of separation is **Between Cluster Sum of Squares (BSS)**, which is the sum of squared distances between the centroids of different clusters. In the formula we sum for each cluster i the squared distance between its centroid m_i and the overall mean m of the dataset, multiplied by the number of points in the cluster |C_i|.  
<img src="img_teoria/separation.png" alt="cluster separation" width="200px"/>

- **Silhouette** it summarizes both cohesion and separation. The nice thing about silhouette is that it can be computed for each point in the dataset, and then we can average the silhouette values of all points to get an overall measure of clustering quality. The silhouette of a point tells us if the point is well clustered (close to its own cluster and far from other clusters) or poorly clustered (far from its own cluster and close to other clusters).  Of course we can compute the silhouette for each cluster by averaging the silhouette values of all points in that cluster, and the overall silhouette by averaging the silhouette values of all points in the dataset.


#### Silhouette 
To compute the silhouette of a point i, we need to calculate two values:
- **a(i)**: it is the average distance between point i and all other points in the same cluster. It measures how well point i is clustered with its own cluster. **A lower value of a(i) indicates that point i is closer to other points in its cluster, which is desirable**. (it tells us cohesion)
- **b(i)**: it is the minimum average distance between point i and all points in other clusters (cioè dato un cluster diverso da quello in cui si trova i calcolo la distanza tra tutti i suoi punti e i, e ne faccio la media. b(i) sarà il minimo tra tutte le medie). It measures how well point i is separated from other clusters. **A higher value of b(i) indicates that point i is farther from points in other clusters, which is desirable**. (it tells us separation)

The **silhouette s(i)** of point i is then computed as:

$ s(i) = \frac{b(i) - a(i)}{max(a(i), b(i))} $

So s(i) ranges from -1 to 1, tipically it is between 0 and 1. The closer s(i) is to 1, the better clustered point i is. If s(i) is close to 0, it means that point i is on or very close to the decision boundary between two neighboring clusters. If s(i) is negative, it means that point i may have been assigned to the wrong cluster, since it means that a(i) > b(i) (the average distance to points in its own cluster is greater than the average distance to points in another cluster).

- *The avg s(i) for all points in a cluster gives us the silhouette of that cluster, it measures how well the points in that cluster are clustered together and separated from other clusters.*
- *The avg s(i) for all points in the dataset gives us the overall silhouette of the clustering result, it measures the overall quality of the clustering result.*

**Since the silhouette is based on distances, it assumes that the clusters are globular in shape (like k-means).**

NB. **Silhouette costs O(n^2)** because for each point we must compute distances to all other points to compute a(i) and b(i).

### Rand Index
Can be used both as relative and external metric. It tells us if we assign the same pair of points to the same cluster in two different clusterings.

Let's assume that we have a ground truth (labels a, b c...). Of course clustering doesn't know about these labels, and it produces its own clusters that are hopefully s.t. points with the same label are in the same cluster. So the idea is exactly this: any two objects that are in the same cluster should have the same label, and viceversa.

So: I consider all pairs of points in the dataset, and I count:
- $f_{00}$: number of pairs having different class labels and assigned to different clusters
- $f_{11}$: number of pairs having same class labels and assigned to same clusters
- $f_{01}$: number of pairs having different class labels but assigned to same clusters
- $f_{10}$: number of pairs having same class labels but assigned to different clusters

And I compute the **Rand Index** as:

$ RI = \frac{f_{00} + f_{11}}{f_{00} + f_{11} + f_{01} + f_{10}} $

This works for external validation. For relative validation, we can compute the Rand Index between two different clustering results on the same dataset (without knowing the true labels), to see how similar the two clusterings are. In that case, f00 is the number of pairs assigned to different clusters in both clusterings, and f11 is the number of pairs assigned to the same clusters in both clusterings.

In scikit learn it is implemented the adjusted Rand Index, which is just a normalized version of the Rand Index, for which 0 if we have something equal to random cluster assignment and 1 if the clustering is perfect in terms of matching the true labels.

**How to choose Eps and MinPts in DBSCAN?**
- We want the distance for each point from its MinPts-th nearest neighbor.
- First we decide to try a MinPts value, for example k=5. Then we compute the euclidean distance and each other point, then we sort them getting for each point the ordered array of distances from other points. Then we take the k+1th nearest neighbor distance (k+1 e non k because we exclude the first, i.e. the distance of the point from itself which is 0). So we get an array containing for each point the distance from its k-th nearest neighbor.
<img src="img_teoria/kth-nn.png" alt="kth nearest neighbor" width="400px"/>

- By sorting and then plotting the array, I get a curve where in the x-axis I have the points ordered by increasing distance from their k+1-th nearest neighbor, and in the y-axis I have the distance from the k+1-th nearest neighbor.

<img src="img_teoria/eps_minpts.png" alt="eps minpts" width="400px"/>

In the image we observe how, up to around the 1750th point, the distance from the kth nearest neighbor increases slowly, while after that it increases rapidly. So the majority of the dataset has a k-th nearest neighbor with distance below 20, precisely around 88% of the dataset (1750/2000) has a distance from its kth nearest neighbor below 18. This means that if we set eps=18, then 88% of the points will have at least k=5 points in their eps, i.e. they will be core points (while the points with distance > eps will be border or noise points).

So in general, fixed a MinPts=k, we look at the curve obtained and we choose eps as the distance value at the "elbow" point of the curve, i.e. the point where the distance starts increasing rapidly. This way we can identify a suitable eps value that captures the density of points in the dataset.

Now, by varying eps and looking at the graph of chameleon, we'll get to the point where we identify a good clustering.

Questo non è dovuto al fatto che DBSCAN ha fatto peggio di k-means, ma al fatto che la silhouette non è una buona metrica per valutare DBSCAN, perché silhouette si basa sulle distanze e quindi assume che i cluster siano globulari, mentre DBSCAN può trovare cluster di forme arbitrarie.
