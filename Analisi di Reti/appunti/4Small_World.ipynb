Lo psicologo sociale Stanley Milgram condusse un esperimento nel '67: scelse una persona (*destinatario*) alla quale doveva essere recapitata una lettera della quale lo stesso Milgram era il *mittente*. Tale obiettivo risiedeva molto lontano da Milgram (es. coste opposte degli Stati Unti). 

Milgram ha poi scelto a caso un insieme di *iniziatori* e ha consegnato a ciascuno di essi una copia della lettera, fornendo anche ad ogni iniziatore una serie di informazioni sul destinatario (nome, indirizzo, occupazione, info personali come interessi, passatempi...).

L'obiettivo di ogni iniziatore era quello di far giungere la lettera al destinatario, ma con il vincolo di non poterla inviare direttamente: **ogni iniziatore doveva scrivere il suo nome sulla lettera e, con l'obiettivo di far giungere la lettera a destinazione in meno passi possibilie, consegnarla a un diretto conoscente chiedendogli di fare lo stesso.**

Ciò che ha osservato Milgram con questo esperimento è stato che:
- circa un terzo delle lettere hanno raggiunto il destinatario
- le lettere che lo hanno raggiunto hanno impiegato in media solo sei passi (i famosi 6 gradi di separazione)

In base all'esito dell'esperimento possiamo quindi trarre due conclusioni:
1. in una rete sociale è presente una moltitudine di percorsi molto brevi che connettono qualunque coppia di nodi --> fenomeno **Small World**
2. i percorsi brevi non solo esistono, ma possono essere trovati con facilità (i.e. da nodi che non conoscono la struttura della rete, ma solo i propri vicini)

Che spiegazione intuitiva si può dare al fatto che esistono così tanti shortest path tra coppie di nodi in una rete sociale? Possiamo vederla così:  
se io ho 100 conoscenze, e ciascuno di essi ha 100 conoscenze, e ciascuno di essi a loro volta 100 conoscenze e così via --> il grafo delle relazioni in questo gruppo contiene 10k percorsi di lunghezza 2 da me ad altre persone della rete, 1kk di percorsi di lunghezza 3 da me ad altre persone della rete e così via i.e. *sono collegato tramite percorsi molto brevi a una fracca di gente.  
Poiché il ragionamento può essere ripetuto per qualsiasi individuo che popola la rete, ecco la spiegazione all'esistenza di tanti percorsi brevi in una rete sociale.

Questo ragionamento è intuitivo e facile, ma nascone una pecca: non tiene conto della **chiusura triadica**, ossia del fatto che in una rete sociale esistono tanti "triangoli" di relazioni, per cui se un individuo conosce $b$ e $c$ allora è probabile che $b$ e $c$ si conoscano a loro volta.  
Quindi, fra i 100 amici dei miei amici, si troveranno diversi di quelli che sono già miei amici, e il grafo della relazione somiglierà alla figura a dx più che quella a sx:

### Il modello Watts-Strogatz
Premesso ciò, ci occupiamo di studiare un modello generativo di grafi aleatori che generi degli **Small Worlds** che **contengano molte chiusure triadiche**.  

Il modello proposto da Watts e Strogatz nel 1998 consiste di un grafo fissato deterministicamente e di un insieme di archi casuali.  
Il grafo fissato è una griglia "arricchita", che corrisponde sostanzialmente a uno *Unit Disk Graph*: considerano i nodi come punti in uno spazio metrico bidimensionale, questi *sono disposti sui punti a coordinate intere di un quadrato centrato nell'origine e ogni nodo è collegato a ciascuno dei nodi vicini in orizzontale, verticale e diagonale.* 

Formalmente: fissato $n \in \mathbb{N}$, consideriamo il grafo $G = (V, E)$ con $V = \{(i, j): 0 \le i \le n \cap 0 \le j \le n\}$ e ciascun nodo $(i, j)$ con $0 \lt u \lt n$ e $0 \lt j \lt n$ è collegato ai nodi $(i-1, j-1), (i-1, j), (i-1, j+1), (i, j-1), (i, j+1), (i+1, j-1), (i+1, j), (i+1, j+1)$  
e analogamente per i nodi $(i, j)$ con $i \in \{0, n\}$ e/o $j \in \{0, n\}$ (senza uscire dai confini del quadrato).

Una volta fissato deterministicamente il grafo come descritto sopra, si fissa un valore $k$ e **ogni nodo sceglie u.a.r. $k$ altri nodi del grafo, che diventano suoi ulteriori vicini** (da qui **griglia arricchita**).

<img src="imgs/4_2.png" alt="Watts-Strogatz" width="400"/>

Più che a una griglia su una superficie piana, dobbiamo in realtà pensare a **una griglia poggiata su una superficie sferica**, che si richiude su se stessa e che chiameremo **wrapped** (il senso è che in questo modo non ci sono nodi "al bordo" della griglia, se vado a destra oltre il confine torno a sinistra)

Osserviamo ora un grafo generato secondo Watts-Strogatz:
1. Possiamo individuare una sorta di dicotomia tra relazioni locali e relazioni a distanza rispetto agli archi deterministici e quelli random: gli archi della griglia determinisitici rappresentano infati le relazioni tra nodi "fisicamente" vicini; mentre quelli random esprimono relazioni tra quelli "fisicamente" lontani.
2. Si può quindi immaginare che i nodi "fisicamente" vicini abbiano più probabilità di incontrarsi rispetto a quelli "fisicamente" lontani (i.e. quelli vicini hanno frequentazioni assidue, quelli lontani no).  
Si può quindi pensare agli archi della griglia come quelli di relazioni forti (**strong ties**) e a quelli random come quelli di relazioni deboli (**weak ties**).
3. I triangoli sono sempre presenti a livello locale, ma sono poco probabili tra archi random (i triangoli si formano quando individui con un amico in comune si incontrano, e due amici di uno stesso individuo che vivono ai capi opposti della Terra è improbabile che possano incontrarsi e diventare amici).

Dunque il grafo generato con questo modello contiene molti triangoli, in accordo con ciò che si riscontra generalmente con le reti sociali. 

Resta però un'altra questione: è vero che coppie di nodi qualsiasi sono collegati da numerosi percorsi brevi?  
A tal proposito Watts e Strogatz hanno osservato che *se partiamo da un nodo $u$ e ci muoviamo solo lungo gli archi random, poiché gli archi random sono distribuiti u.a.r. e ind. nel grafo è molto improbabile che in questo procedimento tocchiamo due volte lo stesso nodo --> ossia molto probabilmente in $h$ passi abbiamo la possibilità di raggiungere $k^h$ nodi diversi* --> **da ogni nodo posso raggiungere tantissimi altri nodi in pochi passi**.

(NB ovviamente deterministicamente ne raggiungo $h$, ma il numero di nodi che potrei raggiungere in totale con buona probabilità è $k^h$ in quanto è improbabile che gli archi random tocchino due volte lo stesso nodo).
 
Questo ragionamento è basato su considerazioni intuitive, dimostrato successivamente da Bollobàs e Chung in modo formale, e oltre a ciò hanno anche individuato la lunghezza media degli shortest path nei grafi generati secondo il modello di Watts-Strogatz.

In effetti però il ragionamento intuitivo di Watts e Strogatz può essere ripetuto in un modello con molta meno casualità: **è sufficiente che solo da un nodo su $k$ partano archi random** e che inoltre **da tale nodo parta un solo arco random**.  
Intuitivamente di nuovo la motivazione è la seguente: raggruppiamo quadrati $k \times k$ della griglia in come fosserò città nel mondo --> ogni città ha uno e un solo arco random uscente. Ripetendo il ragionamento fatto sopra, spostandoci di $h$ passi lungo archi random, allora in quegli $h$ passi potremmo raggiungere con buona probabilità $k^h$ città diverse.

Si può quindi concludere che **poca casualità è sufficiente per avere tanti shortest path**.

Immaginiamo ora di essere un nodo $u$ in un grafo di Watts-Strogatz e di voler inviare un msg a un nodo $v$ **di cui conosciamo le coordinate** (e dalle coordinate, sappiamo che è molto distante da noi).  
Naturalmente vorremmo far arrivare il messaggio più velocemente possibile, ossia vogliamo fare in modo che venga consegnato attraverso uno shortest path (che probabilmente sarà breve) e che collega $u$ a $v$.

Ma $u$ non conosce altro nella rete se non i propri vicini --> inviamo una copia a ciascun vicino (tanto ne abbiamo pochi, 8 deterministici e al massimo $k$ random) --> **conoscere l'indirizzo della destinazione non ci aiuta direttamente**, al limite potremmo inviare il msg al vicino con coordinate più prossime a quelle di $v$, ma la casualità dei weak ties potrebbe far sì che invece un vicino al momento con coordinate peggiori abbia un arco random che lo collega direttamente a $v$.

### i problemi rispetto all'esperimento di Milgram

Dovremmo chiedere quindi ai nostri amici di ripetere il procedimento a loro volta... ma in questo modo dopo $h$ passi circoleranno nel grafo $\approx 7^h$ copie del messaggio: tecnicamente parlando cioè per spedire un messaggio da $u$ a $v$ staremmo utilizzando un protocollo di **flooding** che determina un sovraccarico della rete...

Nell'esperimento di Milgram di contro nella rete, ad ogni passo, circolavano **quanti messaggi quanti ne aveva creati Milgram** in quanto ad ogni passo non aumentavano (invece di usare flooding, i messaggi erano inoltrati euristicamente...) --> evidentemente *il modello Watts-Strogatz non riesce a descrivere qualche caratteristica della rete sociale che rende invece possibile la ricerca di Milgram*.

Un algoritmo di **ricerca decentralizzata** (o **ricerca miope**) è tale che *ciascun nodo non conosce la topologia della rete se non i propri vicini (oltre al nodo target di ricerca)* e *i nodi non comunicano in altro modo se non nell'invio dei messaggi da consegnare* (si tratta quindi di vincoli più restrittivi di un algoritmo distribuito standard).

Come detto nella logica di Milgram esisteva la possibilità di inoltrare il messaggio a un vicino che in qualche modo, probabilmente, era più vicino al destinatario --> la struttura della rete sociale reale dell'esperimento aveva qualche caratteristica per cui **"inoltrare al più vicino"** funziona bene.  
Sarebbe a dire che, in qualche modo, la struttura della rete garantisce che se ad un passo sono nel nodo $u$ e invio la lettera a $v$, allora tra gli amici $v$ c'è (*probabilmente*) qualcuno **molto più vicino di $u$ alla destinazione** e che probabilmente ci sono *tanti passi* in cui la distanza dalla destinazione **diminuisce drasticamente**

Ritorniamo ora a Watts-Strogatz: certamente ciascun nodo ha un vicino sulla griglia più vicino di sé alla destinazione, ma seguendo solo gli archi deterministici impieghiamo un sacco di passi per raggiungerlo, al più $O(\sqrt{n})$ passi --> se voglio trovare un percorso *molto breve* devo usare gli archi random!

Tuttavia non c'è garanzia che in un grafo di Watts-Strogatz, usando la regola "invia al tuo vicino più vicino alla destinazione" allora incontrerò una serie di archi random in modo che in un piccolo numero di passi giungo a destinazione.  
In effetti si può dimostrare che **nel modello di Watts-Strogatz la ricerca decentralizzata di un percorso da $s$ a $t$ individua mediamente un percorso molto più lungo di uno shortest path...** (Kleinberg, 2000).  
Questo è per via del fatto che nel modello l'estremo di un arco random uscente da un nodo è scelto u.a.r. tra tutti gli altri nodi --> gli archi random non tengono conto di quanto sono "vicini" ai nodi che congiungono, sarebbe cioè a dirsi che **gli archi random sono "troppo" random!**

### Un modello per la ricerca decentralizzata
Vogliamo allora definire un modello generativo cui corrispondono grafi *che contengano molti triangoli* e *nei quali esistano molti shortest path tra coppie di nodi*, ma anche nei quali **trovare gli shortest path mediante ricerca decentralizzata sia possibile**. 

Per soddisfare quest'ultimo punto, intuitivamente, è necessario che gli archi random siano scelti in mdoo da tener conto di quanto sono "vicini" i nodi che congiungono (nota che ancora dobbiamo definire il concetto di "vicini"). SLIDE 18
