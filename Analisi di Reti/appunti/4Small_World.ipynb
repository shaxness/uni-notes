Lo psicologo sociale Stanley Milgram condusse un esperimento nel '67: scelse una persona (*destinatario*) alla quale doveva essere recapitata una lettera della quale lo stesso Milgram era il *mittente*. Tale obiettivo risiedeva molto lontano da Milgram (es. coste opposte degli Stati Unti). 

Milgram ha poi scelto a caso un insieme di *iniziatori* e ha consegnato a ciascuno di essi una copia della lettera, fornendo anche ad ogni iniziatore una serie di informazioni sul destinatario (nome, indirizzo, occupazione, info personali come interessi, passatempi...).

L'obiettivo di ogni iniziatore era quello di far giungere la lettera al destinatario, ma con il vincolo di non poterla inviare direttamente: **ogni iniziatore doveva scrivere il suo nome sulla lettera e, con l'obiettivo di far giungere la lettera a destinazione in meno passi possibilie, consegnarla a un diretto conoscente chiedendogli di fare lo stesso.**

Ciò che ha osservato Milgram con questo esperimento è stato che:
- circa un terzo delle lettere hanno raggiunto il destinatario
- le lettere che lo hanno raggiunto hanno impiegato in media solo sei passi (i famosi 6 gradi di separazione)

In base all'esito dell'esperimento possiamo quindi trarre due conclusioni:
1. in una rete sociale è presente una moltitudine di percorsi molto brevi che connettono qualunque coppia di nodi --> fenomeno **Small World**
2. i percorsi brevi non solo esistono, ma possono essere trovati con facilità (i.e. da nodi che non conoscono la struttura della rete, ma solo i propri vicini)

Che spiegazione intuitiva si può dare al fatto che esistono così tanti shortest path tra coppie di nodi in una rete sociale? Possiamo vederla così:  
se io ho 100 conoscenze, e ciascuno di essi ha 100 conoscenze, e ciascuno di essi a loro volta 100 conoscenze e così via --> il grafo delle relazioni in questo gruppo contiene 10k percorsi di lunghezza 2 da me ad altre persone della rete, 1kk di percorsi di lunghezza 3 da me ad altre persone della rete e così via i.e. *sono collegato tramite percorsi molto brevi a una fracca di gente.  
Poiché il ragionamento può essere ripetuto per qualsiasi individuo che popola la rete, ecco la spiegazione all'esistenza di tanti percorsi brevi in una rete sociale.

Questo ragionamento è intuitivo e facile, ma nascone una pecca: non tiene conto della **chiusura triadica**, ossia del fatto che in una rete sociale esistono tanti "triangoli" di relazioni, per cui se un individuo conosce $b$ e $c$ allora è probabile che $b$ e $c$ si conoscano a loro volta.  
Quindi, fra i 100 amici dei miei amici, si troveranno diversi di quelli che sono già miei amici, e il grafo della relazione somiglierà alla figura a dx più che quella a sx:

### Il modello Watts-Strogatz
Premesso ciò, ci occupiamo di studiare un modello generativo di grafi aleatori che generi degli **Small Worlds** che **contengano molte chiusure triadiche**.  

Il modello proposto da Watts e Strogatz nel 1998 consiste di un grafo fissato deterministicamente e di un insieme di archi casuali.  
Il grafo fissato è una griglia "arricchita", che corrisponde sostanzialmente a uno *Unit Disk Graph*: considerano i nodi come punti in uno spazio metrico bidimensionale, questi *sono disposti sui punti a coordinate intere di un quadrato centrato nell'origine e ogni nodo è collegato a ciascuno dei nodi vicini in orizzontale, verticale e diagonale.* 

Formalmente: fissato $n \in \mathbb{N}$, consideriamo il grafo $G = (V, E)$ con $V = \{(i, j): 0 \le i \le n \cap 0 \le j \le n\}$ e ciascun nodo $(i, j)$ con $0 \lt u \lt n$ e $0 \lt j \lt n$ è collegato ai nodi $(i-1, j-1), (i-1, j), (i-1, j+1), (i, j-1), (i, j+1), (i+1, j-1), (i+1, j), (i+1, j+1)$  
e analogamente per i nodi $(i, j)$ con $i \in \{0, n\}$ e/o $j \in \{0, n\}$ (senza uscire dai confini del quadrato).

Una volta fissato deterministicamente il grafo come descritto sopra, si fissa un valore $k$ e **ogni nodo sceglie u.a.r. $k$ altri nodi del grafo, che diventano suoi ulteriori vicini** (da qui **griglia arricchita**).

<img src="imgs/4_2.png" alt="Watts-Strogatz" width="400"/>

Più che a una griglia su una superficie piana, dobbiamo in realtà pensare a **una griglia poggiata su una superficie sferica**, che si richiude su se stessa e che chiameremo **wrapped** (il senso è che in questo modo non ci sono nodi "al bordo" della griglia, se vado a destra oltre il confine torno a sinistra)

Osserviamo ora un grafo generato secondo Watts-Strogatz:
1. Possiamo individuare una sorta di dicotomia tra relazioni locali e relazioni a distanza rispetto agli archi deterministici e quelli random: gli archi della griglia determinisitici rappresentano infati le relazioni tra nodi "fisicamente" vicini; mentre quelli random esprimono relazioni tra quelli "fisicamente" lontani.
2. Si può quindi immaginare che i nodi "fisicamente" vicini abbiano più probabilità di incontrarsi rispetto a quelli "fisicamente" lontani (i.e. quelli vicini hanno frequentazioni assidue, quelli lontani no).  
Si può quindi pensare agli archi della griglia come quelli di relazioni forti (**strong ties**) e a quelli random come quelli di relazioni deboli (**weak ties**).
3. I triangoli sono sempre presenti a livello locale, ma sono poco probabili tra archi random (i triangoli si formano quando individui con un amico in comune si incontrano, e due amici di uno stesso individuo che vivono ai capi opposti della Terra è improbabile che possano incontrarsi e diventare amici).

Dunque il grafo generato con questo modello contiene molti triangoli, in accordo con ciò che si riscontra generalmente con le reti sociali. 

Resta però un'altra questione: è vero che coppie di nodi qualsiasi sono collegati da numerosi percorsi brevi?  
A tal proposito Watts e Strogatz hanno osservato che *se partiamo da un nodo $u$ e ci muoviamo solo lungo gli archi random, poiché gli archi random sono distribuiti u.a.r. e ind. nel grafo è molto improbabile che in questo procedimento tocchiamo due volte lo stesso nodo --> ossia molto probabilmente in $h$ passi abbiamo la possibilità di raggiungere $k^h$ nodi diversi* --> **da ogni nodo posso raggiungere tantissimi altri nodi in pochi passi**.

(NB ovviamente deterministicamente ne raggiungo $h$, ma il numero di nodi che potrei raggiungere in totale con buona probabilità è $k^h$ in quanto è improbabile che gli archi random tocchino due volte lo stesso nodo).
 
Questo ragionamento è basato su considerazioni intuitive, dimostrato successivamente da Bollobàs e Chung in modo formale, e oltre a ciò hanno anche individuato la lunghezza media degli shortest path nei grafi generati secondo il modello di Watts-Strogatz.

In effetti però il ragionamento intuitivo di Watts e Strogatz può essere ripetuto in un modello con molta meno casualità: **è sufficiente che solo da un nodo su $k$ partano archi random** e che inoltre **da tale nodo parta un solo arco random**.  
Intuitivamente di nuovo la motivazione è la seguente: raggruppiamo quadrati $k \times k$ della griglia in come fosserò città nel mondo --> ogni città ha uno e un solo arco random uscente. Ripetendo il ragionamento fatto sopra, spostandoci di $h$ passi lungo archi random, allora in quegli $h$ passi potremmo raggiungere con buona probabilità $k^h$ città diverse.

Si può quindi concludere che **poca casualità è sufficiente per avere tanti shortest path**.

Immaginiamo ora di essere un nodo $u$ in un grafo di Watts-Strogatz e di voler inviare un msg a un nodo $v$ **di cui conosciamo le coordinate** (e dalle coordinate, sappiamo che è molto distante da noi).  
Naturalmente vorremmo far arrivare il messaggio più velocemente possibile, ossia vogliamo fare in modo che venga consegnato attraverso uno shortest path (che probabilmente sarà breve) e che collega $u$ a $v$.

Ma $u$ non conosce altro nella rete se non i propri vicini --> inviamo una copia a ciascun vicino (tanto ne abbiamo pochi, 8 deterministici e al massimo $k$ random) --> **conoscere l'indirizzo della destinazione non ci aiuta direttamente**, al limite potremmo inviare il msg al vicino con coordinate più prossime a quelle di $v$, ma la casualità dei weak ties potrebbe far sì che invece un vicino al momento con coordinate peggiori abbia un arco random che lo collega direttamente a $v$.

### i problemi rispetto all'esperimento di Milgram

Dovremmo chiedere quindi ai nostri amici di ripetere il procedimento a loro volta... ma in questo modo dopo $h$ passi circoleranno nel grafo $\approx 7^h$ copie del messaggio: tecnicamente parlando cioè per spedire un messaggio da $u$ a $v$ staremmo utilizzando un protocollo di **flooding** che determina un sovraccarico della rete...

Nell'esperimento di Milgram di contro nella rete, ad ogni passo, circolavano **quanti messaggi quanti ne aveva creati Milgram** in quanto ad ogni passo non aumentavano (invece di usare flooding, i messaggi erano inoltrati euristicamente...) --> evidentemente *il modello Watts-Strogatz non riesce a descrivere qualche caratteristica della rete sociale che rende invece possibile la ricerca di Milgram*.

Un algoritmo di **ricerca decentralizzata** (o **ricerca miope**) è tale che *ciascun nodo non conosce la topologia della rete se non i propri vicini (oltre al nodo target di ricerca)* e *i nodi non comunicano in altro modo se non nell'invio dei messaggi da consegnare* (si tratta quindi di vincoli più restrittivi di un algoritmo distribuito standard).

Come detto nella logica di Milgram esisteva la possibilità di inoltrare il messaggio a un vicino che in qualche modo, probabilmente, era più vicino al destinatario --> la struttura della rete sociale reale dell'esperimento aveva qualche caratteristica per cui **"inoltrare al più vicino"** funziona bene.  
Sarebbe a dire che, in qualche modo, la struttura della rete garantisce che se ad un passo sono nel nodo $u$ e invio la lettera a $v$, allora tra gli amici $v$ c'è (*probabilmente*) qualcuno **molto più vicino di $u$ alla destinazione** e che probabilmente ci sono *tanti passi* in cui la distanza dalla destinazione **diminuisce drasticamente**

Ritorniamo ora a Watts-Strogatz: certamente ciascun nodo ha un vicino sulla griglia più vicino di sé alla destinazione, ma seguendo solo gli archi deterministici impieghiamo un sacco di passi per raggiungerlo, al più $O(\sqrt{n})$ passi --> se voglio trovare un percorso *molto breve* devo usare gli archi random!

Tuttavia non c'è garanzia che in un grafo di Watts-Strogatz, usando la regola "invia al tuo vicino più vicino alla destinazione" allora incontrerò una serie di archi random in modo che in un piccolo numero di passi giungo a destinazione.  
In effetti si può dimostrare che **nel modello di Watts-Strogatz la ricerca decentralizzata di un percorso da $s$ a $t$ individua mediamente un percorso molto più lungo di uno shortest path...** (Kleinberg, 2000).  
Questo è per via del fatto che nel modello l'estremo di un arco random uscente da un nodo è scelto u.a.r. tra tutti gli altri nodi --> gli archi random non tengono conto di quanto sono "vicini" ai nodi che congiungono, sarebbe cioè a dirsi che **gli archi random sono "troppo" random!**

### Un modello per la ricerca decentralizzata
Vogliamo allora definire un modello generativo cui corrispondono grafi *che contengano molti triangoli* e *nei quali esistano molti shortest path tra coppie di nodi*, ma anche nei quali **trovare gli shortest path mediante ricerca decentralizzata sia possibile**. 

Per soddisfare quest'ultimo punto, intuitivamente, è necessario che gli archi random siano scelti in modo da tener conto di quanto sono "vicini" i nodi che congiungono (nota che ancora dobbiamo definire il concetto di "vicini").  

Creiamo allora un modello come segue: di nuovo abbiamo un'"ossatura" deterministica (la stessa griglia arricchita di Watts-Strogatz) e da ogni nodo esce un arco random.  
Ora però, la probabilità che l'arco random uscente dal nodo $u$ sia $(u,v)$ è *inversamente proporzionale* alla **distanza sulla griglia dei nodi $u$ e $v$**, ossia:
$$\Pr((u,v) \in E) = \frac{1}{Z_u} \cdot \frac{1}{d(u,v)^q}$$
dove 
- $d(u,v)$ è la **lunghezza di uno shortest path tra $u$ e $v$ sulla griglia** (i.e. percorso più breve, ma deterministico);
- $Z_u$ è un fattore di normalizzazione
- $q$ è un parametro che prende il nome di **esponente di clustering**.  
Si ottiene un modello diverso per ogni $q$: ad es. se $q = 0$ allora gli archi random sono scelti u.a.r. tra tutti i nodi per cui si torna al modello di Watts-Strogatz; *in generale comunque gli archi random sono "più random" al decrescere di $q$*.

Per $Z_u$ in particolare; dal momento da ogni nodo deve uscire *uno e un solo arco random*, deve essere che $\sum_{v \in V-\{u\}} \Pr((u,v) \in E) = 1$ ossia $\sum_{v \in V-\{u\}} \frac{1}{Z_u} \cdot \frac{1}{d(u,v)^q} = 1$, per cui:
$$Z_u = \sum_{v \in V-\{u\}} \frac{1}{d(u,v)^q}$$
Poiché la griglia wrapped è simmetrica --> $Z_u$ è lo stesso per tutti i nodi --> **indicheremo il fattore di normalizzazione semplicemente con $Z$**.

A questo punto vogliamo mostrare che esiste un valore di $q$ *ottimale per la ricerca decentralizzata*.  
Si può dimostrare in particolare che **nel caso in cui la componente deterministica del grafo è una griglia (wrapped) bidimensionale --> l'esponente di clustering ottimale è $q = 2$** quando il numero di nodi è molto grande.

Più in generale, se la componente deterministica del grafo è una griglia wrapped di dimensione $d$ (i.e. i nodi sono disposti in uno spazio $\mathbb{R}^d$) allora l'esponente di clustering ottimale è $q = d$.

Prima di vedere perché questo è vero intuitivamente, per poi passare alla dimostrazione formale.

Ripensiamo all'esperimento di Milgram: immaginiamo che il destinatario finale viva a Garbatella, e che tra gli iniziatori ci fosse Pippo che abita a New York. Allora Pippo cercherà qualcuno che viva in Europa.  
Immaginiamo che Pippo conosca Pluto, che però abita a Mosca.  
Pluto a sua volta cerca di avvicinarsi all'Europa occidentale, e la lettera finisce così a Parigi, poi Milano, Perugia, Latina, Centocelle, fino a Garbatella.

Sarebbe a dire che il tragitto della lettera segue uno schema a "scale di risoluzione": prima continenti, poi nel continente, poi nazione, poi regione, e poi città...

Formalizziamo ora quello che abbiamo osservato per vedere a cosa ci porta nel caso della griglia bidimensionale.  
Fissiamo un nodo $u$ e partizioniamo i nodi rimanenti per blocchi, ogni blocco definito in base alla distanza da $u$ con crescita esponenziale. Sarebbe a dire che nel primo blocco ci sono i nodi a distanza da $u$ compresa tra $2^1$ e $2^2$, nel secondo blocco quelli a distanza tra $2^2$ e $2^3$, ..., $2^h$ e $2^{h+1}$ etc...

Allora il numero di nodi nel blocco $2^{h+1} - 2^h$ ($h-esimo$ blocco) è $\approx \pi (2^{h+1})^2 - \pi (2^h)^2 = \pi 4(2)^2h - \pi (2)^2h = \pi 2^{2h} (4-1) = 3\pi 2^{2h}$, ossia cresce esponenzialmente con $h$ (proporzionale a $2^{2h}$). (è una corona circolare, si calcola sottraendo le due circonferenze). 

<img src="imgs/4_4.png" alt="Blocco" width="200"/>

Scelto $v$ nel blocco $h$, la probabilità che l'arco random uscente da $u$ sia proprio $(u,v)$ è proporzionale a $\alpha \frac{1}{d(u,v)^q} = \alpha \frac{1}{2^{2h}}$ (perché $d(u,v) \approx 2^h$ e ci siamo posti nel caso $q = 2$).

Ma allora **la probabilità che l'arco random uscente da $u$ cada nel blocco $h$ è indipendente da $h$**: seguendo il ragionamento approssimato infatti:
$$
\Pr(u \rightarrow \text{blocco } h) = 3\pi 2^{2h} \cdot \alpha \frac{1}{2^{2h}} = 3\pi \alpha $$ 
e questo vale per ogni blocco --> **la probabilità di raggiungere uno qualsiasi dei blocchi è esattamente la stessa!**

Quindi la scelta di $q = 2$ è esponente di clustering ottimale dato che permette di cercare una destinazione in modo molto efficiente, in quanto **i weak ties (archi random) sono distribuiti uniformemente su tutte le scale di risoluzione!** (dove le scale di risoluzione nel caso reale erano continenti, nazioni, regioni etc.. mentre sopra sono i blocchi definiti in base alla distanza da $u$ con i nodi $[2, 4), [4, 8), [8, 16), ...$).

Sarebbe a dire che **$q=2$ è ottimale in quanto fa sì che ciascun blocco riceva la stessa probabilità complessiva di essere raggiunto da un arco random rispetto agli altri** --> anche se per un certo numero di passi può occorrere utilizzare gli archi della griglia, comunque quei pochi passi random saranno così grandi da permettermi di avvicinarmi drasticamente alla destinazione.

es. devo raggiungere $t$ a distanza 100 da me. Ad ogni passo scelgo il vicino più vicino a lui (ricorda che ho solo visione locale e le coordinate del nodo target). Allora possono succedere due cose: 
- o il mio arco random si allontana ancora di più da $t$ (es. è a distanza 130 da $t$) --> lascio il msg a un nodo della griglia; il progresso sarà piccolo ma mi affido a lui 
- oppure l'arco casuale è utile; in tal caso con buona probabilità in particolare mi porterà molto più vicino a $t$ dato che è equibrobabile finire in ciascuno dei blocchi.

#### Dimostrazione formale della $q=d$ nel caso $d$-dimensionale
Dopo l'intuizione nel caso bidimensionale, vediamo ora la dimostrazione formale dell'**ottimalità** di $q=d$ nel caso $d$-dimensionale.

*Per farlo si analizzano formalmente le prestazioni dell'algoritmo di ricerca decentralizzata applicato al modello generativo introdotto, però nel caso $d=1$.* Ciò in quanto chiaramente l'analisi risulta più semplice e la generalizzazione alle altre dimensioni è basata sostanzialmente sugli stessi argomenti.

Il caso $d=1$ corrisponde a quello in cui i nodi sono disposti su uno spazio unidimensionale wrapped, ergo **anello**, al quale però sono aggiunti archi random in accordo col modello appena descritto e con coefficiente di clustering $q = 1$.

<img src="imgs/4_5.png" alt="1D" width="200"/>

Quindi si considera un grafo $G$ tale che i nodi sono disposti su un anello, ossia $V = [n]$ e ${(i,i+1): 1 \le i \lt n} \cup {(n,1)} \subseteq E$ al quale per. sono aggiunti gli archi random; per $u,v \in V$ si ha 
$$\Pr((u,v) \in E) = \frac{1}{Z} \cdot \frac{1}{d(u,v)}$$
perché $q = 1$ e $Z = \sum_{v \in V-\{u\}} \frac{1}{d(u,v)}$.

Nel nostro caso in particolare:
$$Z = \sum_{v \in V-\{u\}} \frac{1}{d(u,v)} = 2 \sum_{h=1}^{n/2} \frac{1}{h} \le 2 \sum_{h=1}^{\infty} \frac{1}{h} = 2 \ln(n)$$
perché in un anello $u$ ha 2 nodi a distanza 1, due a distanza 2, 2 a distanza 3 etc.. --> raggruppo i nodi per distanza e ottengo la somma fino a $n/2$ (perché la distanza massima in un anello è $n/2$) e l'ultimo passaggio è dovuto alla somma armonica.

Quindi ricapitolando sia $G$ tale che i nodi sono disposti su un anello al quale sono stati aggiunti archi random: per $u,v \in V \Pr((u,v) \in E) = \frac{1}{Z} \cdot \frac{1}{d(u,v)}$ con $Z = \sum_{v \in V-\{u\}} \frac{1}{d(u,v)} \le 2 \ln(n)$.

Allora scegliamo u.a.r. due nodi $s$ e $t$ in G ed *utilizziamo l'algoritmo di ricerca decentralizzata per calcolare un percorso da $s$ a $t$*.  
Sia $X$ la v.a. che **denota la lunghezza di tale percorso**.  
Allora si può dimostrare che $\mathbb{E}[X] = O(\log^2(n))$.

> **Teorema**:
>
> $\mathbb{E}[X] = O(\log^2(n))$.
>
> **Dimostrazione**:
>
> Ricordiamo che intuitivamente l'algoritmo di ricerca decentralizzata funziona così: partendo da $s$, ad ogni passo invia il messaggio al vicino più vicino a $t$ (tra quelli della griglia e quello random) e così iterativamente fino a raggiungere $t$. 
>
> Suddividiamo allora in fasi il processo di trasmissione del msg:  
> **durante la fase j, la lettera deve essere in possesso di un nodo $u$ t.c.:**
> $$\frac{d(s,t)}{2^{j+1}} \lt d(u,t) \le \frac{d(s,t)}{2^j}$$
> in questo modo, partendo con la fase $j = 0$, ad ogni fase la distanza del nodo con il msg a $t$ si dimezza. Poiché la distanza iniziale è $d(s,t) \le n/2$ --> **il numero di fasi è $le \log_2(n)$**.
>
> Sia $X_j$ la durata della fase $j$ (i.e. numero di nodi che entrano in possesso del messaggio durante la fase $j$). Allora:
> $$X = \sum_{j=1}^{\log_2(n)} X_j \quad \text{e } \mathbb{E}[X] = \sum_{j=1}^{\log_2(n)} \mathbb{E}[X_j]$$
> Per dimostrare il teorema è quindi sufficiente **dimostrare che per ogni $j$, $\mathbb{E}[X_j] = O(\log(n))$**.
>
> Supponiamo di trovarci nel nodo $v$ durante la fase $j$: allora come detto $\frac{d(s,t)}{2^{j+1}} \lt d(v,t) \le \frac{d(s,t)}{2^j}$.  
> La fase $j$ termina sicuramente se esiste un nodo $z$ tale che $(v,z) \in E$ e $d(z,t) \le \frac{d(v,t)}{2}$ in quanto $d(z,t) \le \frac{d(v,t)}{2} \le \frac{1}{2} \frac{d(s,t)}{2^{j}}$ (perché in tal caso il nodo $z$ non è più distante al più $\frac{d(s,t)}{2^{j}}$, ma $\frac{d(s,t)}{2^{j+1}}$!)
>
> Quindi:
> $$\Pr(\text{fase j termina}) = \Pr\left(\exists z \in V: (v,z) \in E \text{ e } d(z,t) \le \frac{d(s,t)}{2^{j+1}}\right) \ge \Pr\left(\exists z \in V: (v,z) \in E \text{ e } d(z,t) \le \frac{d(v,t)}{2}\right)$$
> dove il $\ge$ deriva dal fatto che la fase potrebbe terminare anche con un progresso più piccolo: $d(v,t) = \frac{d(s,t)}{2^{j}}$. In tal caso infatti esisterebbe un arco $(v,z)$ nell'anello tale che $d(z,t) = d(v,t) -1 \lt \frac{d(s,t)}{2^j}$ e passare per quell'arco deterministico farebbe certamente terminare la fase $j$, nonostante comunque sarebbe $d(z, t) \gt \frac{d(v,t)}{2}$.
>
> Sia ora $\mathcal{I}$ l'insieme dei nodi che distano da $t$ non più della metà di quanto $v$ dista da $t$: 
> $$\mathcal{I} = \{u \in V: d(u,t) \le \frac{d(v,t)}{2}\}$$
> Allora:
>
> <img src="imgs/4_6.png" alt="I" width="600"/>
>
> Sia $z \in \mathcal{I}$, allora:
> $$d(v,z) \le d(v,t) + d(t,z) = d(v,t) + d(z,t) \le d(v,t) + \frac{d(v,t)}{2} = \frac{3}{2} d(v,t)$$
> applicando la definizione di un nodo in $\mathcal{I}$ e la disuguaglianza triangolare.
>
> Allora per ogni $z \in \mathcal{I}$:
> $$\Pr((v,z) \text{ è un arco random}) = \frac{1}{Z} \frac{1}{d(v,z)} \ge \frac{1}{Z} \cdot \frac{2}{3 d(v,t)} \ge \frac{1}{2 \ln(n)} \frac{2}{3 d(v,t)}$$
> Allora, tornando alla probabilità che la fase $j$ termini:
> $$ \Pr(\exists z \in V: (v,z) \in E \text{ e } d(z,t) \le \frac{d(v,t)}{2}) \ge \sum_{z \in \mathcal{I}} \Pr((v,z) \text{ è un arco random}) \ge \sum_{z \in \mathcal{I}} \frac{1}{3 d(v,t)\ln(n)} = |\mathcal{I}| \cdot \frac{1}{3 d(v,t)\ln(n)}$$

> Resta quindi solo da valutare $|\mathcal{I}| = |\{u \in V: d(u,t) \le \frac{d(v,t)}{2}\}|$.
>
> Siano $v_{sin}$ e $v_{des}$ i due nodi **in $\mathcal{I}$** che **hanno distanza massima da $t$**.  
> Allora $d(v_{sin}, t) = d(v_{des}, t) = \lfloor\frac{d(v,t)}{2}\rfloor$ (parte intera inf. perché $\frac{d(v,t)}{2}$ potrebbe non essere intero) e siano $v_{sin1}$ e $v_{des1}$ i due nodi **adiacenti a $t$. 
>
> <img src="imgs/4_7.png" alt="I" width="200"/>
>
> Ma allora $\mathcal{I}$ contiene i nodi $t$, poi i $\lfloor\frac{d(v,t)}{2}\rfloor$ nodi da $v_{sin}$ a $v_{sin1}$ e i $\lfloor\frac{d(v,t)}{2}\rfloor$ nodi da $v_{des}$ a $v_{des1}$. Ma allora:
> $$|\mathcal{I}| = 1 + 2 \lfloor\frac{d(v,t)}{2}\rfloor \ge 1 + 2 (\frac{d(v,t)}{2} - 1) \ge d(v,t)$$
> In conclusione quindi:
> $$\Pr(\text{fase j termina}) \ge |\mathcal{I}| \cdot \frac{1}{3 d(v,t)\ln(n)} \ge d(v,t) \cdot \frac{1}{3 d(v,t)\ln(n)} = \frac{1}{3 \ln(n)}$$
> allora:
> $$\Pr(\text{fase j NON termina}) \le 1 - \frac{1}{3 \ln(n)}$$
> e:
> $$\Pr(\text{fase j NON termina per } h \text{ passi}) \le (1 - \frac{1}{3 \ln(n)})^h$$

#### Perché $q=d$ funziona bene in $\mathbb{R}^d$
Riassumendo un po', gli "ingredienti" che permettono di dimostrare che la ricerca decentralizzata si comporta bene nell'anello con $q=d=1$ sono:
1. **fissato $d$, il numero di nodi a distanza al più $d$ dalla destinazione $t$ è all'incirca proprio $d$** (nel caso dell'anello ad es. i nodi a distanza al più $n/2$ da $t$ sono $n-1$, proporzionale a $n/2$)
2. il fattore di normalizzazione è $Z \le 2 \ln(n)$

Questi due ingredienti permettono di dimostrare che, trovandoci in un nodo $v$ a distanza $d$ da $t$, la probabilità che $v$ abbia vicino $z$ a distanza $\le d/2$ da $t$ è proporzionale a $\frac{1}{\ln(n)}$ **indipendentemente da $d$**.

Infatti, detto $\mathcal{I} = \{u \in V: d(u,t) \le \frac{d(v,t)}{2}\}$ e avendo $|\mathcal{I}| \approx d(v,t)$ (appunto numero di nodi a distanza al più $d$ è circa $d$), allora:
$$\Pr(\exists z \in V: (v,z) \in E \text{ e } d(z,t) \le \frac{d(v,t)}{2}) \ge |\mathcal{I}| \cdot \frac{1}{Z d(v,z)} \ge d(v,t) \cdot \frac{1}{\ln(n) d(v,t)} = \frac{1}{3 \ln(n)}$$
(dove $d(v,z)$ al denominatore diventa $d(v,t)$ per la disuguaglianza triangolare, vedi dim).

Considerazioni analoghe possono essere fatte **nel caso bidimensionale con $q=d=2$**. Ora:
1. $|\mathcal{I}| = \alpha d(v,t)^2$ per qualche costante $\alpha$ (perché in due dimensioni $\mathcal{I}$ è una regione attorno $t$ di raggio $\frac{d(v,t)}{2}$, quadrato di lato $\approx d(v,t)$ --> il numero di nodi cresce come la sua area $\propto \alpha d(v,t)^2$)
2. $Z = \sum_{v \in V-\{u\}} \frac{1}{d(u,v)^2}$, e si può dimostrare che $Z \le \beta \ln(n)$ per qualche costante $\beta$ 

Allora:
$$\Pr(\exists z \in V: (v,z) \in E \text{ e } d(z,t) \le \frac{d(v,t)}{2}) \ge \sum _{z \in \mathcal{I}} \Pr((v,z) \text{ è un arco random}) \ge |\mathcal{I}| \cdot \frac{1}{Z d(v,z)^2}$$
Ma poiché $Z \le \beta \ln(n)$ e $d(v,z) \le \frac{3 d(v,t)}{2}$ per la disuguaglianza triangolare ($d(v,z) \le d(v,t) + d(z,t) \le d(v,t) + \frac{d(v,t)}{2} = \frac{3 d(v,t)}{2}$), allora:
$$\Pr(\exists z \in V: (v,z) \in E \text{ e } d(z,t) \le \frac{d(v,t)}{2}) \ge |\mathcal{I}| \cdot \frac{1}{Z d(v,z)^2} \ge \frac{4}{Z 9 d(v,t)^2} |\mathcal{I}| \ge \frac{4 \alpha}{9 \beta \ln(n)}$$
Di nuovo la probabilità di trovare un vicino a distanza al più $d(v,t)/2$ da $t$ è **indipendente da $d(v,t)$**.

Allo stesso modo, considerazioni analoghe valgono anche per $d \gt 2$.

#### Perché $q \ne d$ funziona male in $\mathbb{R}^d$
Cerchiamo di capire perché il nostro modello, nel caso $d = 1$ e $q = 0$ non è prestante per la ricerca decentralizzata (i.e. compiendo una ricerca decentralizzata generando archi random in accordo al modello con $q = 0$, si trova mediamente un percorso la cui lunghezza è molto elevata).

In questo caso infatti la probabilità che l'arco random uscente da un nodo $u$ sia $(u,v)$ è $\Pr((u,v) \in E) = \frac{1}{Z} \cdot \frac{1}{d(u,v)^0} = \frac{1}{Z}$, dove poiché deve essere che $\sum_{v \in V-\{u\}} \Pr((u,v) \in E) = 1$ allora $Z = n-1$.  
Quindi $\Pr((u,v) \in E) = \frac{1}{n-1}$.

Come già osservato, nel caso "anello e $q=1$" è probabile entrare in regioni del grafo contenenti nodi sempre più vicino a $t$: gli insiemi $\mathcal{I}$ (sarebbe a dire più precisamente che la probabilità di *dimezzare* la nostra distanza da $t$ è la stessa sia quando siamo molto lontani da $t$ che quando ci siamo molto vicini).

Mostriamo ora, molto informalmente, che nel caso "anello e $q=0$" è "difficile" (improbabile) entrare nell'insieme:
$$R = \{u \in V: d(u,t) \le \sqrt{n}\}$$
Fissiamo $t$ destinazione e scegliamo sorgente $s \notin R$.  
Allora $\forall u \in R$ e $\forall v \in [n]\setminus R$ si ha
$$
\Pr((v,u)\in E) > \frac{1}{n-1} > \frac{1}{n}.
$$
(dove il primo $\gt$ deriva dal fatto che $u$ e $v$ potrebbero anche essere vicini "deterministici" lungo l'anello)

Ma allora $\forall v \in R$ si ha 
$$\Pr(\exists u \in R: (v,u) \in E) =    \sum_{u \in R} \Pr((v,u) \in E) \gt \frac{|R|}{n} = \frac{2\sqrt{n}}{n} = \frac{2}{\sqrt{n}}$$
(perché $|R| = 2 \sqrt{n}$, siamo in un anello).

Sia allora $Y$ la v.a. che rappresenta il numero di passi per raggiungere da $s$ un nodo in $R$ --> $\Pr(Y \ge h) \lt (1 - \frac{2}{\sqrt{n}})^h$ e quindi:
$$\mathbb{E}[Y] = \sum_{h=0}^{\infty} \Pr(Y \ge h) \le \sum_{h=0}^{\infty} (1 - \frac{2}{\sqrt{n}})^h = \frac{1}{1 - (1 - \frac{2}{\sqrt{n}})} = \frac{\sqrt{n}}{2}$$
(dove abbiamo usato che per una v.a. $Y$ a valori interi non negativi, $\mathbb{E}[Y] = \sum_{h=0}^{\infty} \Pr(Y \gt h)$ e la somma geometrica)

Ma questo valore è **esponenzialmente** più grande di $\mathbb{E}[X] = O(\log^2(n))$ che si ottiene nel caso $q=d=1$!

Abbiamo quindi visto che entrare in $R = \{u \in V: d(u,t) \le \sqrt{n}\}$ nel caso "anello e $q=0$" richiede in media $\frac{\sqrt{n}}{2}$ passi.


Ora vediamo in quanti passi, mediamente, entriamo nella regione $R_2 = \{u \in V: d(u,t) \le \frac{\sqrt{n}}{2}\}$.  
Facendo analogamente al caso $R$, $\forall v \in R-R_2$ si ha:
$$\Pr(\exists u \in R_2: (v,u) \in E) \ge \frac{|R_2|}{n} = \frac{\sqrt{n}}{n} = \frac{1}{\sqrt{n}}$$
e quindi $\Pr(Y_2 \ge h) \le (1 - \frac{1}{\sqrt{n}})^h$.

Indicando allora con $Y_2$ la v.a. che rappresenta il numero di passi per entrare in $R_2$:
$$\mathbb{E}[Y_2] = \sum_{h=0}^{\infty} \Pr(Y_2 \ge h) \le \sum_{h=0}^{\infty} (1 - \frac{1}{\sqrt{n}})^h = \frac{1}{1 - (1 - \frac{1}{\sqrt{n}})} = \sqrt{n}$$
Ma questo significa che, una volta entrati in $R$, per entrare in $R_2$ utilizzare gli archi random è mediamente equivalente a muoversi direttamente lungo gli archi deterministici dell'anello (i.e. *quando si raggiunge una distanza dall'obiettivo dell'ordine $\sqrt{n}$, gli archi random non sembrano più utili per avvicinarsi all'obiettivo*, e la ragione di ciò è che **gli archi random sono "troppo" random** come già osservato nel modello Watts-Strogatz).

In effetti quanto visto nel caso "anello e $q=0$" può esser generalizzato a tutti i casi "anello & $0 \lt q \lt 1$": in tutti i questi casi gli archi random sono troppo random...

Di contro nel caso "anello e $q \gt 1$" gli archi random sono troppo corti, per cui è "difficile" (improbabile) imbattersi in archi random che coprano grandi distanze e che permettano quindi di avvicinarsi alla destinazione in pochi passi.  
Quindi la ricerca decentralizzata nel caso "anello e $q \gt 1$" riesce a fare poco meglio che che utilizzare solo gli archi deterministici per arrivare a destinazione. 

In particolare, si può dimostrare quanto segue.  
**Teorema: comunque si scelga $q \ne 1$, esistono due costanti positive $a_q$ e $c_q$ tali che** (detta $X$ la v.a. che esprime la lunghezza dell'alg. di ricerca decentralizzata in un anello di $n$ nodi a cui sono aggiunti archi random in accordo al modello con coefficiente di clustering $q$) **allora $\mathbb{E}[X] \ge a_q n^{c_q}$**.
