Abbiamo introdotto due modelli di grafi aleatori: Erdős-Rényi e Barabási-Albert. Entrambi si prestano a generare grafi che corrispondono a reti reali, i cui archi sono però logici/virtuali, rappresentando la relazione tra individui/entità (es. relazioni di amicizia nelle reti sociali o collegamenti logici come hyperlink tra pagine web). 

Quando però l'obiettivo è costruire effettivamente una rete fisica (es. rete di calcolatori in un cluster o rete di sensori distribuitim per cui serve predisporre fisicamente le connessioni) allora occorre tenere in considerazione la **struttura geometrica** dello spazio nel quale i nodi sono inseriti (es. muri, ostacoli, distanze fisiche tra i nodi...) --> i modelli che abbiamo studiato sono poco significativi in questo senso.

## Grafi Geometrici
Un grafo geometrico consiste in un insieme $V$ di punti in uno spazio metrico dei quali conosciamo le coordinate, e un parametro $r \gt 0$.  

Assumiamo per fissare le idee che lo spazio metrico sia il piano cartesiano $\mathbb{R}^2$ e che la distanza sia la distanza euclidea.  
Allora ciascun punto è individuato da una coppia di coordinate $A = (x_A, y_A)$ e gli archi del grafo individuato da $V$ e $r$ sono *tutte e sole le coppie di punti la cui distanza euclidea è $\le r$*:
$$E = \{(A,B): A \in V \land B \in V \land \sqrt{(x_A - x_B)^2 + (y_A - y_B)^2} \le r\}$$
Generalmente si normalizza rispetto ad $r$, ossia si riportano i punti in scala $1:r$ (i.e. si pone pari ad $r$ l'unità degli assi cartesiani, in pratica il raggio $r$ diventa uguale a 1 affinché due nodi sono adiacenti se e solo se la loro distanza euclidea è $\le 1$).

In tal caso il grafo geometrico è detto **Unit Disk Graph**, ma noi ci occuperemo del caso non normalizzato.

Parliamo ora di **grafi geometrici aleatori**.  
Sia $n \in \mathbb{N}$ e sia $r \gt 0$ (come vedremo anche $r \lt \sqrt{2}$).  
Scegliamo $n$ punti u.a.r. in un quadrato unitario $Q = [0,1]^2$ e costruiamo il grafo geometrico corrispondente.

Poiché i punti sono scelti nel quadrato unitario e la diagonale del quadrato misura $\sqrt{2}$, è sufficiente scegliere $r \lt \sqrt{2}$ (se fosse esattamente uguale a $\sqrt{2}$ allora tutti i punti sarebbero adiacenti tra loro e il grafo sarebbe completo, e maggiore di $\sqrt{2}$ è dunque inutile).

In questo caso, l'aleatorietà del grafo $G(n,r)$ non deriva dalla scelta casuale degli archi, **ma dalla scelta casuale dei vertici nel quadrato unitario** (solo quando $r \lt \sqrt{2}$ e $r gt 0$ naturalmente, in quanto in questi due casi estremi il grafo risultante è deterministicamente completo o vuoto!)

Intuitivamente, scegliendo $r$ vicino a $\sqrt{2}$ avremo un grafo molto denso, mentre scegliendo $r$ vicino a 0 avremo un grafo molto sparso.  
Analogamente a quanto fatto con Erdős-Rényi (nel quale avevamo scelto il parametro $p$ in funzione di $n$), **poichè scegliere $r$ costante fa sì che all'aumentare di $n$ il grafo diventi sempre più denso, scegliamo $r$ in funzione di $n$: $r = r(n)$**.

(intuitivamente se $r$ è costante, la rete sarà molto probabilmente connessa quando $n$ è molto grande, e molto probabilmente non connessa quando $n$ è piccolo, ma non ci interessa questo: vogliamo capire come deve scalare $r$ in funzione di $n$ per garantire che il grafo sia connesso con alta probabilità).

Ci occupiamo del problema di scegliere il più piccolo valore di $r(n)$ tale che si ottenga un grafo connesso. *Più precisamente, essendo $G(n,r)$ un grafo aleatorio, ci chiediamo quando è connesso con alta probabilità.*  

Vediamo ora un ambito di applicazione di questo problema: le reti di sensori.

### Reti di sensori
Una *rete wireless ad hoc* è costituita da nodi che dispongono di ricetrasmettitori wireless, tramite i quali realizzano comunicazioni p2p.  
In una rete di questo tipo avvengono comunicazioni *multi-hop*, i.e. per cui i nodi non sono in grado di connettersi direttamente tra loro (es. per la distanza o per ostacoli fisici) --> devono passare attraverso altri nodi intermedi per trasmettere i dati. 

Una *rete di sensori* è una particolare classe di reti wireless ad hoc in cui ogni nodo è dotato di sensori per rilevare informazioni sull'ambiente circostante, di una **batteria** di capacità limitata, di un ricetrasmettitore wireless e di un microprocessore per elaborare i dati raccolti.

I sensori raccolgono info (temperatura, pressione, umidità...), eventualmente le elaborano (per comprimerle o aggregarle con info di nodi vicini) e le trasmettono agli altri nodi, affinché tutte le entità abbiano una visione globale della zona monitorata (che può essere quindi accessibile da un utente esterno attraverso uno o più nodi gateway).  
Applicazioni possibili abbondano: es. monitoraggio temperatura oceano per rendere possibili previsioni meteo più accurate, individuazione di incendi boschivi, rapida diffusione su info di traffico...  

Il ricetrasmettitore può essere configurato in modo da trasmettere entro un certo raggio $r$ (detto **raggio di trasmissione**), i.e. dato un dispositivo $x$ configurato con raggio $r_x$, quello può inviare direttamente messaggi solo a dispositivi a distanza $\le r_x$.  
Una rete di questo tipo può quindi essere rappresentata tramite un grafo diretto, detto **grafo di comunicazione**:

<img src=imgs/3_1.png width=400>

Chiaramente affinché le info circolino ovunque serve che anche i nodi più remoti siano informati --> è necessario che il grafo diretto sia **fortemente connesso**: per ogni coppia di nodi $a$ e $b$ della rete deve esistere una sequenza di nodi $u_1, u_2, ..., u_k$ tale che $a$ possa trasmettere direttamente (un solo hop) a $u_1$, $u_1$ a $u_2$, ..., $u_k$ a $b$, i.e. esistano gli archi diretti $(a,u_1), (u_1,u_2), ..., (u_k,b)$ nel grafo di comunicazione.

Assumeremo d'ora in avanti però che **tutti i nodi abbiano lo stesso raggio di trasmissione $r$** e che **gli $n$ nodi siano distribuiti u.a.r. in una regione limitata che, senza perdita di generalità, considereremo essere un quadrato unitario $Q = [0,1]^2$** (stessa ipotesi dei grafi geometrici aleatori, questo si verifica ad es. se i sensori sono "lanciati" da un aereo o da un drone sopra la zona da monitorare). Assumeremo infine, per gli stessi motivi descritti prima, che **$r$ sia in funzione di $n$: $r = r(n)$**.

Ma allora **la nostra rete è modellata da un grafo geometrico aleatorio**, che in particolare rappresenta **grafi non orientati** (perché se $A$ può trasmettere a $B$ allora anche $B$ può trasmettere ad $A$, in quanto hanno lo stesso raggio di trasmissione, $d(A,B) \le r \iff d(B,A) \le r$).

Sotto queste ipotesi studiamo il problema di **minimo raggio di trasmissione**: *qual è il minimo $r(n)$ tale che grafi di comunicazione generati da $n$ nodi distribuiti u.a.r. in $Q = [0,1]x[0,1]$ siano connessi con buona probabilità al crescere di $n$?*

Perché questo? Questioni di energia: più piccolo è il raggio di trasmissione, minore è la potenza necessaria per trasmettere --> maggiore durata della batteria dei nodi (indispensabile per la sopravvivenza dei sensori nel tempo, in quanto la loro dislocazione rende difficile la sostituzione delle batterie).

#### Il Teorema
Dimostreremo che, **detto $r^*(n)$ il minimo valore per $r(n)$ che garantisce, con probabilità ragionevole, che $G(n, r(n))$ sia connesso, allora $r^*(n) = \Theta\left(\sqrt{\frac{\ln n}{n}}\right)$**.

Questo risultato sarà dimostrato in particolare in due passi: se $n$ punti sono scelti u.a.r. nel quadrato $Q = [0,1]^2$ allora: 

<img src=imgs/3_2.png width=500>

Delimitazione superiore al minimo raggio di connessione perché stiamo cercando $r^*(n)$ come minimo raggio di connessione. Allora con il primo teorema dimostriamo che ci "basta" prendere $r(n) \ge \gamma_1 \sqrt{\frac{\ln n}{n}}$ per avere con alta probabilità un grafo connesso (condizione sufficiente, non necessaria, il teo. non ci dice niente su raggi inferiori a questo) --> quello è un limite superiore al minimo raggio di connessione 

Delimitazione inferiore al contrario ci dice come la connessione, scendendo troppo, non è "garantita": scegliendo $r^*(n)$ meno del valore enunciato la probabilità che il grafo non sia connesso tende, per $n \to \infty$, a un valore maggiore di 0.

##### **Teorema 1: Delimitazione superiore al minimo raggio di connessione**
>
> Esiste una costante $\gamma_1 \gt 0$ tale che, se $r(n) \ge \gamma_1 \sqrt{\frac{\ln n}{n}}$, allora $G(n,r(n))$ è connesso con alta probabilità.
>
> **Dim**:
>
> Sia $k(n) \gt 0$ un intero dipendente da $n$.  
> Allora partizioniamo il quadrato unitario $Q$ in $k(n)^2$ quadrati più piccoli di lato $\frac{1}{k(n)}$, e poniamo $r(n)$ pari alla lunghezza della diagonale di una coppia di celle adiacenti (dove due coppie sono definite adiacenti se hanno un lato in comune). 
>
> <img src=imgs/3_3.png width=300>
>
> Allora questo raggio sarà uguale, per il teorema di Pitagora, a:
> $$r(n) = \sqrt{\left(\frac{2}{k(n)}\right)^2 + \left(\frac{1}{k(n)}\right)^2} = \frac{\sqrt{5}}{k(n)}$$
> Così facendo, ogni nodo in qualsiasi cella sarà necessariamente collegato da un arco a tutti i nodi presenti (eventualmente) in celle adiacenti. 
>
> Allora se dimostriamo che *con alta probabilità ciascuna cella contiene almeno un nodo* allora avremo anche dimostrato che *$G(n,r(n))$ è connesso con alta probabilità* con questo raggio.
>
> In particolare dimostriamo che **è possibile scegliere $k(n)$ in modo tale che, w.h.p., nessuna cella è vuota**. Invece di calcolare però la prob. di questo evento sfruttiamo il complementare (più facile): **esiste almeno una cella vuota**.
>
> Sia C una cella fissata. Allora dobbiamo calcolare $\Pr(C = \emptyset)$, ma l'evento $C = \emptyset$ si verifica se e solo se *nessuno dei $n$ punti scelti u.a.r. in $Q$ cade in $C$*, i.e. coincide con l'evento:
> $$\bigcap_{i=1}^n \{i \notin C\}$$
> Poiché i punti sono posizionati in modo indipendente u.a.r. --> la probabilità di intersezione di eventi indipendenti è il prodotto delle probabilità dei singoli eventi:
> $$\Pr(C = \emptyset) = \Pr\left(\bigcap_{i=1}^n \{i \notin C\}\right) = \prod_{i=1}^n \Pr(i \notin C)$$
> Sia $i$ un nodo qualsiasi --> la probabilità che il nodo sia scelto proprio in $C$ è pari al rapporto tra l'area di $C$ e quella del quadrato unitario $Q$ (che ha ovviamente area pari a 1)
> --> $\Pr(i \in C) = \frac{\text{area}(C)}{\text{area}(Q)} = \frac{1}{k(n)^2} \implies \Pr(i \notin C) = 1 - \frac{1}{k(n)^2}$
>
> Allora:
> $$\Pr(C = \emptyset) = \prod_{i=1}^n \Pr(i \notin C) = \left(1 - \frac{1}{k(n)^2}\right)^n$$
> Ma noi abbiamo considerato una sola cella fissata, mentre il nostro obiettivo è dimostrare che esiste almeno una cella vuota w.h.p. Sfruttando lo union bound:
> $$\Pr(\exists C: C = \emptyset) = \Pr\left(\bigcup_{C \in Q} \{C = \emptyset\}\right) \le \sum_{C \in Q} \Pr(C = \emptyset) = k(n)^2 \left(1 - \frac{1}{k(n)^2}\right)^n$$
> da cui, avendo trovato $r(n) = \frac{\sqrt{5}}{k(n)}$, sostituendo $k(n) = \frac{\sqrt{5}}{r(n)}$ otteniamo:
> $$\Pr(\exists C: C = \emptyset) \le \frac{5}{r(n)^2} \left(1 - \frac{r(n)^2}{5}\right)^n$$
> Infine, ponendo $r(n) = \gamma_1 \sqrt{\frac{\ln n}{n}}$ otteniamo:
> $$\Pr(\exists C: C = \emptyset) \le \frac{5}{\gamma_1^2} \frac{n}{\ln n} \left(1 - \frac{\gamma_1^2 \ln n}{5n}\right)^n$$
> A questo punto si sfrutta la disuguaglianza (la dimostrazione sta nel primo file di appunti):
> $$\forall x \in \mathbb{R}: 1 - x \le e^{-x} \quad \text{e} \quad \text{if } x \ne 0 \text{ then } 1 - x \lt e^{-x}$$
> In virtù di questa disuguaglianza, poiché $\frac{\gamma_1^2 \ln n}{5n} \ne 0$ per ogni $n \in \mathbb{N}$, abbiamo che $(1-\frac{\gamma_1^2 \ln n}{5n})^n \lt e^{-\frac{\gamma_1^2 \ln n}{5n}}$, quindi:
>
> <img src=imgs/3_4.png width=400>
>
> Si osserva che $1 - \frac{\gamma_1^2}{5n} \lt 0$ per $\gamma_1 \gt \sqrt{5}$.
>
> Riassumendo, abbiamo dimostrato che ponendo $r(n) = \gamma_1 \sqrt{\frac{\ln n}{n}}$ allora:
> $$\Pr(\exists C: C = \emptyset) \lt \frac{5}{\gamma_1^2} n^{1-\frac{\gamma_1^2}{5}}$$
> ma come detto l'esponente di $n$ è negativo per $\gamma_1 \gt \sqrt{5}$.
>
> In conclusione quindi scegliendo $r(n) = \gamma_1 \sqrt{\frac{\ln n}{n}}$ con $\gamma_1 \gt \sqrt{5}$ e ponendo $b = \frac{5}{\gamma_1^2}$ e $c = \frac{\gamma_1^2}{5} - 1$ otteniamo:
> 
> $$\Pr(\exists C: C = \emptyset) \lt \frac{b}{n^c}$$
> con $c \gt 0$ in quanto $\gamma_1 \gt \sqrt{5}$ --> la probabilità che esista almeno una cella vuota tende a 0 al crescere di $n$.
>
> Sarebbe a dire che abbiamo dimostrato che se $r(n) \ge \gamma_1\left(\sqrt{\frac{\ln n}{n}}\right)$ allora esiste $\gamma_1 \gt 0$ (in particolare $\gamma_1 \gt \sqrt{5}$) tale che:
> $$\Pr(G(n,r(n)) \text{ è connesso}) = 1 - \Pr(\exists C: C = \emptyset) \gt 1 - \frac{b}{n^c}$$
> ossia $G(n,r(n))$ è connesso con alta probabilità.
>
> $\blacksquare$

**Teorema 2: Delimitazione inferiore al minimo raggio di connessione**

> Per ogni costante $c \gt 0$, se $r(n) = \sqrt{\frac{\ln n + c}{\pi n}}$ allora $\lim_{n \to \infty} \Pr(G(n,r(n)) \text{ non è connesso}) \gt 0$.
>
> Denoteremo nel corso della dimostrazione con G il grafo $G(n,r(n))$ e con $r$ il valore $r(n)$, altrimenti impazziamo.
>
> Osserviamo che nella prova della delimitazione superiore (i.e. $r^*(n) \le \gamma_1 \sqrt{\frac{\ln n}{n}}$ con $r^*(n)$ minimo raggio di connessione) abbiamo dovuto cercare una **maggiorazione** della probabilità che $G$ non fosse connesso, i.e. dimostrare che $\Pr(G \text{ non è connesso}) \lt \text{qualcosa}$.
>
> Ora invece, per la delimitazione inferiore $r^*(n) \ge \sqrt{\frac{\ln n + c}{\pi n}}$ dobbiamo cercare una **minorazione** della probabilità che $G$ non sia connesso, i.e. dimostrare che $\Pr(G \text{ non è connesso}) \ge \text{qualcosa}$.
>
> **Dimostrazione**:
> Dobbiamo quindi trovare una minorazione per $\Pr(G \text{ non è connesso})$.
> 
> Introduciamo per cominciare gli eventi da esprimere in funzione della probabilità che vogliamo minorare:
> - $\varepsilon_{\ge 1}$ = G contiene almeno un nodo isolato
> - $\varepsilon_{i_1, \dots, i_h}$ = i nodi $i_1, \dots, i_h$ sono tutti  isolati (con $i_1, \dots, i_h \in [n]$)
> - $\varepsilon_{i!}$ = $i$ è l'unico nodo isolato in G
>
> Se G contiene almeno un nodo isolato allora ovviamente G non è connesso  
> --> $\Pr(G \text{ non è connesso}) \ge \Pr(\varepsilon_{\ge 1})$.  
> Inoltre, se accade che $1$ è l'unico nodo isolato in G, oppure lo è 2, ..., oppure lo è $n$ allora G contiene ovviamente almeno un nodo isolato  
> --> $\Pr(\varepsilon_{\ge 1}) \ge \Pr(\bigcup_{i=1}^n \varepsilon_{i!})$  
> E poiché $\varepsilon_{1!}, \dots, \varepsilon_{n!}$ sono eventi disgiunti (non possono verificarsi contemporaneamente, solo un nodo può essere isolato alla volta) allora:
> $$\Pr(\bigcup_{i=1}^n \varepsilon_{i!}) = \sum_{i=1}^n \Pr(\varepsilon_{i!})$$
> Cioè in conclusione
> $$\Pr(G \text{ non è connesso}) \ge \sum_{i=1}^n \Pr(\varepsilon_{i!})$$
> Tuttavia non è facile calcolare direttamente $\Pr(\varepsilon_{i!})$ --> lavoriamo per minorarla con espressioni più semplici da calcolare.
>
> A questo scopo osserviamo anzitutto che $i$ è l'unico nodo isolato in G se e solo se:
> - $i$ è un nodo isolato in G
> - comunque scegliamo un altro nodo $j$, $i$ e $j$ non sono entrambi isolati in G
>
> --> $\varepsilon_{i!} = \varepsilon_i \cap \bigcap_{j \in [n] \setminus \{i\}} \varepsilon_{i,j}^c = \varepsilon_i - \bigcup_{j \in [n] \setminus \{i\}} \varepsilon_{i,j}$ (i.e. i è isolato e per ogni altro nodo $j$ non è vero che $i$ e $j$ sono entrambi isolati). Allora:
> $$\Pr(\varepsilon_{i!}) = \Pr(\varepsilon_i) - \Pr\left(\bigcup_{j \in [n] \setminus \{i\}} \varepsilon_{i,j}\right) \ge \Pr(\varepsilon_i) - \sum_{j \in [n] \setminus \{i\}} \Pr(\varepsilon_{i,j})$$
> dove l'ultima disuguaglianza deriva dallo union bound.
>
> Quindi a questo punto per trovare una minorazione di $\Pr(G \text{ non è connesso}) \ge \sum_{i=1}^n \Pr(\varepsilon_{i!})$ non ci resta che trovare:
> - **una minorazione per $\Pr(\varepsilon_i)$** (la probabilità che un nodo $i$ sia isolato)
> - **una maggiorazione per $\Pr(\varepsilon_{i,j})$** (la probabilità che due nodi $i$ e $j$ siano entrambi isolati)
>
> Prima di procedere chiariamo due ruoli distinti: $i$ come **oggetto virtuale** (nodo del grafo), e $t_i$ invece il punto in cui $i$ si trova fisicamente nel piano --> $t_i \in Q$ e $C_r(t_i)$ è il cerchio di raggio $r$ centrato in $t_i$.
>


> **Minorazione di $\Pr(\varepsilon_i)$**:
>
> Si verifica l'evento $\varepsilon_i$ se e solo se, una volta fissato $t_i$, **nessun nodo $j \neq i$ è posizionato in $C_r(t_i)$**.  
> In questo senso, **fissato $t_i$** e fissato $j \neq i$ si ha che:
> $$\Pr(t_j \notin C_r(t_i) | t_i) \ge \frac{\text{area di } Q-C_r(t_i)}{\text{area di } Q} \ge 1 - \pi r^2$$
> Il maggiore uguale deriva dal fatto che $C_r(t_i)$ potrebbe non essere interamente contenuto in $Q$ (se ad es. $t_i$ è vicino al bordo di Q) --> la probabilità che cada nel resto del quadrato è maggiore quindi non ci frega
>
> <img src=imgs/3_5.png width=300>
>
> Allora, **sempre fissato $t_i$**, $\Pr(\varepsilon_i | t_i) \ge \Pr(\forall j \neq i, t_j \notin C_r(t_i) | t_i) \ge (1 - \pi r^2)^{n-1}$.  
> 
> Ora vogliamo calcolare la probabilità complessiva, senza che $t_i$ sia fissato. Sappiamo che il punto $t_i$ nel quale posizionare $i$ sappiamo essere scelto u.a.r. in $Q$, che rappresenta un insieme continuo. La funzione di densità corrispondente alla scelta u.a.r. in un punto di $Q$ è $f(t) = \frac{1}{\text{area di }(Q)} = 1$:
> $$\Pr(\varepsilon_i) = \int_{t_i \in Q} \Pr(\varepsilon_i | t_i) f(t_i) dt_i \ge (1-\pi r^2)^{n-1} \int_{t_i \in Q} f(t_i) dt_i = (1-\pi r^2)^{n-1}$$
> (perché $f(t_i)$ è una densità e deve integrare a 1 su tutto lo spazio di probabilità $Q$).

> **Maggiorazione di $\Pr(\varepsilon_{i,j})$**:
>
> L'evento $\varepsilon_{i,j}$ si verifica se e solo se, una volta fissato $t_i$, $t_j \notin C_r(t_i)$ e nessun altro nodo $h \neq i,j$ cade in $C_r(t_i) \cup C_r(t_j)$.
>
> Questo evento può essere espresso come unione di due altri eventi mutualmente esclusivi (i.e. disgiunti):
> - $\varepsilon_{i,j}^1$ = $t_j \notin C_2r(t_i)$ e $\forall h \in [n] \setminus \{i,j\},  
> t_h \notin C_r(t_i) \cup C_r(t_j)$, ossia $t_j$ non cade nel cerchio di raggio $2r$ centrato in $t_i$ e tutti gli altri nodi non cadono nei cerchi di $i$ e $j$ (vogliamo infatti che entrambi siano isolati) --> sarebbe a dire $t_j$ cade nella regione gialla in figura
> - $\varepsilon_{i,j}^2$ = $t_j \in C_2r(t_i)$ e $\forall h \in [n] \setminus \{i,j\}, t_h \notin C_r(t_i) \cup C_r(t_j)$,  
> ossia $t_j$ cade nella corona azzurra in figura
>
> <img src=imgs/3_6.png width=300>
>
> Allora $\Pr(\varepsilon_{i,j}) = \Pr(\varepsilon_{i,j}^1 \cup \varepsilon_{i,j}^2) = \Pr(\varepsilon_{i,j}^1) + \Pr(\varepsilon_{i,j}^2)$ (eventi disgiunti)
>
> **Calcolo di $\Pr(\varepsilon_{i,j}^1)$**:
>
> Calcoliamo anzitutto $\Pr(\varepsilon_{i,j}^1)$, fissato $t_i$ fissiamo anche $t_j$ nella zona gialla e fissiamo $h \in [n] \setminus \{i,j\}$ --> la probabilità che $t_h$ cada nella regione rimanente (gialla + azzurra) è pari al rapporto dell'area della regione con quella del quadrato, ossia $1 - 2\pi r^2$.  
>
> <img src=imgs/3_7.png width=500>
>  
> Per essere precisi, questo è vero se $t_i$ e $t_j$ non sono troppo vicini ai bordi del quadrato: in tal caso infatti la regione gialla + azzurra è più grande e quindi la probabilità (che è addendo di qualcosa che vogliamo maggiorare) sarebbe maggiore di quella presa in considerazione. Sono stati tuttavia effettuati degli studi prendendo in considerazione anche i bordi, che hanno portato agli stessi identici risultati solo dopo molti più calcoli (per questo ci limitiamo a considerare il caso interno al quadrato).
>
> Questo deve valere per tutti gli $h \in [n] \setminus \{i,j\}$, quindi sempre fissati $t_i$ e $t_j$ la probabilità che nessuno di essi cada nei cerchi di $i$ e $j$ è pari a $(1 - 2\pi r^2)^{n-2}$ (perché i punti sono piazzati in modo indipendente) (e sempre **trascurando gli effetti ai bordi**).
>
> A questo punto calcoliamo la probabilità complessiva --> di nuovo integrali:
>
> <img src=imgs/3_8.png width=500>
>
> Dove stavolta portando fuori la costante più interna resta un integrale in un intervallo pari all'area della zona gialla, che è proprio $(1-4\pi r^2)$
>
> **Calcolo di $\Pr(\varepsilon_{i,j}^2)$**:
>
> Maggioriamo $\Pr(\varepsilon_{i,j}^2)$, fissato $t_i$ fissiamo anche $t_j$ nella zona azzurra e fissiamo $h \in [n] \setminus \{i,j\}$ --> la probabilità che $t_h$ cada nella regione rimanente (gialla + azzurra) è massima quando è massima l'intersezione tra i due cerchi $C_r(t_i)$ e $C_r(t_j)$, ossia quando $t_j$ è sulla circonferenza che delimita $C_r(t_i)$. 
>
> <img src=imgs/3_9.png width=300>
>
> In questo caso allora:
> $$\text{area}(C_r(t_i) \cup C_r(t_j)) = 2\pi r^2 - \text{area}(C_r(t_i) \cap C_r(t_j))$$
> Ma $\text{area}(C_r(t_i) \cap C_r(t_j)) = 2 \text{ area viola} = 2 [\text{area rosa e viola} - \text{area } t_jAB]$
>

> Ma $t_jAt_i$ è un triangolo equilatero di lato $r$ --> la sua area è $\frac{r^2 \sqrt{3}}{4}$ (base per altezza diviso 2, e l'altezza di un equilatero è $\frac{\sqrt{3}}{2} l$).
>
> Ma la regione rosa e viola è un settore circolare di $C_r(t_j)$ e il suo angolo al centro $At_jB$ è il doppio di $At_jt_i$ che misura $\frac{\pi}{3}$ (perché $t_jAt_i$ e $t_jBt_i$ sono equilateri) --> l'area della regione rosa e viola è $\frac{1}{2}r^2\frac{2\pi}{3} = r^2 \frac{\pi}{3}$.
>
> Ma allora l'area della regione viola è $r^2 \frac{\pi}{3} - \frac{r^2 \sqrt{3}}{4}$, quindi:
> $$\text{area}(C_r(t_i) \cap C_r(t_j)) = 2 r^2 (\frac{\pi}{3} - \frac{\sqrt{3}}{4})$$
> e quindi:

> (il $\ge$ deriva dal fatto che l'unione è più grande quando l'intersezione è piccola, e noi stiamo considerando il caso in cui l'intersezione invece è maggiore del minimo possibile dovendo mettere $t_j$ sul bordo del cerchio di $t_i$.  
> Il $\gt$ finale deriva invece dal fatto che $\frac{4}{3} + \frac{\sqrt{3}}{2\pi} > \frac{8}{5}$).
>
> Ossia, la probabilità di scegliere $t_h$ nella regione giallo+azzurra è $1-\text{area}(C_r(t_i) \cup C_r(t_j)) \lt 1 - \frac{8}{5} \pi r^2$
