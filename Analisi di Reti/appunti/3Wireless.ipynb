Abbiamo introdotto due modelli di grafi aleatori: Erdős-Rényi e Barabási-Albert. Entrambi si prestano a generare grafi che corrispondono a reti reali, i cui archi sono però logici/virtuali, rappresentando la relazione tra individui/entità (es. relazioni di amicizia nelle reti sociali o collegamenti logici come hyperlink tra pagine web). 

Quando però l'obiettivo è costruire effettivamente una rete fisica (es. rete di calcolatori in un cluster o rete di sensori distribuitim per cui serve predisporre fisicamente le connessioni) allora occorre tenere in considerazione la **struttura geometrica** dello spazio nel quale i nodi sono inseriti (es. muri, ostacoli, distanze fisiche tra i nodi...) --> i modelli che abbiamo studiato sono poco significativi in questo senso.

## Grafi Geometrici
Un grafo geometrico consiste in un insieme $V$ di punti in uno spazio metrico dei quali conosciamo le coordinate, e un parametro $r \gt 0$.  

Assumiamo per fissare le idee che lo spazio metrico sia il piano cartesiano $\mathbb{R}^2$ e che la distanza sia la distanza euclidea.  
Allora ciascun punto è individuato da una coppia di coordinate $A = (x_A, y_A)$ e gli archi del grafo individuato da $V$ e $r$ sono *tutte e sole le coppie di punti la cui distanza euclidea è $\le r$*:
$$E = \{(A,B): A \in V \land B \in V \land \sqrt{(x_A - x_B)^2 + (y_A - y_B)^2} \le r\}$$
Generalmente si normalizza rispetto ad $r$, ossia si riportano i punti in scala $1:r$ (i.e. si pone pari ad $r$ l'unità degli assi cartesiani, in pratica il raggio $r$ diventa uguale a 1 affinché due nodi sono adiacenti se e solo se la loro distanza euclidea è $\le 1$).

In tal caso il grafo geometrico è detto **Unit Disk Graph**, ma noi ci occuperemo del caso non normalizzato.

Parliamo ora di **grafi geometrici aleatori**.  
Sia $n \in \mathbb{N}$ e sia $r \gt 0$ (come vedremo anche $r \lt \sqrt{2}$).  
Scegliamo $n$ punti u.a.r. in un quadrato unitario $Q = [0,1]^2$ e costruiamo il grafo geometrico corrispondente.

Poiché i punti sono scelti nel quadrato unitario e la diagonale del quadrato misura $\sqrt{2}$, è sufficiente scegliere $r \lt \sqrt{2}$ (se fosse esattamente uguale a $\sqrt{2}$ allora tutti i punti sarebbero adiacenti tra loro e il grafo sarebbe completo, e maggiore di $\sqrt{2}$ è dunque inutile).

In questo caso, l'aleatorietà del grafo $G(n,r)$ non deriva dalla scelta casuale degli archi, **ma dalla scelta casuale dei vertici nel quadrato unitario** (solo quando $r \lt \sqrt{2}$ e $r gt 0$ naturalmente, in quanto in questi due casi estremi il grafo risultante è deterministicamente completo o vuoto!)

Intuitivamente, scegliendo $r$ vicino a $\sqrt{2}$ avremo un grafo molto denso, mentre scegliendo $r$ vicino a 0 avremo un grafo molto sparso.  
Analogamente a quanto fatto con Erdős-Rényi (nel quale avevamo scelto il parametro $p$ in funzione di $n$), **poichè scegliere $r$ costante fa sì che all'aumentare di $n$ il grafo diventi sempre più denso, scegliamo $r$ in funzione di $n$: $r = r(n)$**.

(intuitivamente se $r$ è costante, la rete sarà molto probabilmente connessa quando $n$ è molto grande, e molto probabilmente non connessa quando $n$ è piccolo, ma non ci interessa questo: vogliamo capire come deve scalare $r$ in funzione di $n$ per garantire che il grafo sia connesso con alta probabilità).

Ci occupiamo del problema di scegliere il più piccolo valore di $r(n)$ tale che si ottenga un grafo connesso. *Più precisamente, essendo $G(n,r)$ un grafo aleatorio, ci chiediamo quando è connesso con alta probabilità.*  

Vediamo ora un ambito di applicazione di questo problema: le reti di sensori.

### Reti di sensori
Una *rete wireless ad hoc* è costituita da nodi che dispongono di ricetrasmettitori wireless, tramite i quali realizzano comunicazioni p2p.  
In una rete di questo tipo avvengono comunicazioni *multi-hop*, i.e. per cui i nodi non sono in grado di connettersi direttamente tra loro (es. per la distanza o per ostacoli fisici) --> devono passare attraverso altri nodi intermedi per trasmettere i dati. 

Una *rete di sensori* è una particolare classe di reti wireless ad hoc in cui ogni nodo è dotato di sensori per rilevare informazioni sull'ambiente circostante, di una **batteria** di capacità limitata, di un ricetrasmettitore wireless e di un microprocessore per elaborare i dati raccolti.

I sensori raccolgono info (temperatura, pressione, umidità...), eventualmente le elaborano (per comprimerle o aggregarle con info di nodi vicini) e le trasmettono agli altri nodi, affinché tutte le entità abbiano una visione globale della zona monitorata (che può essere quindi accessibile da un utente esterno attraverso uno o più nodi gateway).  
Applicazioni possibili abbondano: es. monitoraggio temperatura oceano per rendere possibili previsioni meteo più accurate, individuazione di incendi boschivi, rapida diffusione su info di traffico...  

Il ricetrasmettitore può essere configurato in modo da trasmettere entro un certo raggio $r$ (detto **raggio di trasmissione**), i.e. dato un dispositivo $x$ configurato con raggio $r_x$, quello può inviare direttamente messaggi solo a dispositivi a distanza $\le r_x$.

Chiaramente affinché le info circolino ovunque serve che anche i nodi più remoti siano informati --> è necessario che il grafo diretto sia **fortemente connesso**: per ogni coppia di nodi $a$ e $b$ della rete deve esistere una sequenza di nodi $u_1, u_2, ..., u_k$ tale che $a$ possa trasmettere direttamente (un solo hop) a $u_1$, $u_1$ a $u_2$, ..., $u_k$ a $b$, i.e. esistano gli archi diretti $(a,u_1), (u_1,u_2), ..., (u_k,b)$ nel grafo di comunicazione.

Assumeremo d'ora in avanti però che **tutti i nodi abbiano lo stesso raggio di trasmissione $r$** e che **gli $n$ nodi siano distribuiti u.a.r. in una regione limitata che, senza perdita di generalità, considereremo essere un quadrato unitario $Q = [0,1]^2$** (stessa ipotesi dei grafi geometrici aleatori, questo si verifica ad es. se i sensori sono "lanciati" da un aereo o da un drone sopra la zona da monitorare). Assumeremo infine, per gli stessi motivi descritti prima, che **$r$ sia in funzione di $n$: $r = r(n)$**.

Ma allora **la nostra rete è modellata da un grafo geometrico aleatorio**, che in particolare rappresenta **grafi non orientati** (perché se $A$ può trasmettere a $B$ allora anche $B$ può trasmettere ad $A$, in quanto hanno lo stesso raggio di trasmissione, $d(A,B) \le r \iff d(B,A) \le r$).

Sotto queste ipotesi studiamo il problema di **minimo raggio di trasmissione**: *qual è il minimo $r(n)$ tale che grafi di comunicazione generati da $n$ nodi distribuiti u.a.r. in $Q = [0,1]x[0,1]$ siano connessi con buona probabilità al crescere di $n$?*

Perché questo? Questioni di energia: più piccolo è il raggio di trasmissione, minore è la potenza necessaria per trasmettere --> maggiore durata della batteria dei nodi (indispensabile per la sopravvivenza dei sensori nel tempo, in quanto la loro dislocazione rende difficile la sostituzione delle batterie).

#### Il Teorema
Dimostreremo che, **detto $r^*(n)$ il minimo valore per $r(n)$ che garantisce, con probabilità ragionevole, che $G(n, r(n))$ sia connesso, allora $r^*(n) = \Theta(\sqrt{\frac{\log n}{n}})$**.
