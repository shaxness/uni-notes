Ci si occuperà di Reti nell'accezione del termine più ampia possibile.  
Le analizzeremo da molti punti di vista: **prestazioni**, **struttura**, **utilizzo**, e utilizzando molte tecniche diverse.  
Sintesi tra Matematica/Informatica, Economia (relazioni come incentivo/disincentivo a comportamenti), Scienze Sociali (studio di trutture e interazioni caratteristiche di gruppi e popolazioni).

Ci occupiamo di modelli e della loro analisi formale, che esibiscano in particolare caratteristiche simili a quelle dei sistemi reali.

Una **Rete** rappresenta uno schema di interconnessione tra un insieme di entità. In base al tipo di entità si parla di reti fisiche, sociali etc...  
L'idea di Rete alla base di questo corso è quella di **un'ampia popolazione che reagisce all'azione dei singoli**. 

Chiaramente studiare e rappresentare reti di milioni di individui è difficile, ma si possono analizzarne le proprietà globali, ad esempio se la rete contiene **Componenti Giganti** o se porzioni della rete sono più densamente connesse di altre.

Studieremo i fenomeni a volte a livello di popolazione, senza considerare i singoli individui (es. rich get richer; studieremo qual è mediamente la frazione di individui con elevato grado di popolarità) e altre volte per comprendere certi fenomeni invece considereremo precisamente la topologia della rete, relazione per relazione.

Chiaramente, vista la dinamicità dei sistemi reali, ci occuperemo di studiare cose del tipo "qual è **mediamente** il diametro di una rete sociale" etc...  
Per fare ciò bisogna considerare moltissime reti, e da dove le prendiamo (i dati costano e di reti in giro non ce ne sono abbastanza per un'accurata indagine statistica)? 

L'idea è quella di **generare delle reti**, ad esempio tramite modelli aleatori, per cui alcuni di essi riproducono alcuni fenomeni osservati anche nelle reti reali --> capiremo *quali modelli utilizzare per riprodurre certi fenomeni*.

### Erdős-Rényi 
Sia $n \in \mathbb{N}$ e $p \in [0,1]$.  
A partire da $n$ e $p$ si genera un grafo nel modo seguente:
- l'insieme di nodi del grafo è $[n] = \{1,2,...,n\}$;
- per ogni coppia di nodi distinti $i,j \in [n]$ si aggiunge un arco tra $i$ e $j$ con probabilità $p$, indipendentemente da tutti gli altri archi.

Il grafo così costruito è un **evento aleatorio** che si denota con $G_{n,p}$ (in particolare è un grafo non diretto).  
Di questo grafo studieremo l'**esistenza di una componente gigante** e la **distribuzione del grado dei nodi**.

Si osserva che se $p = 0$ allora si ottiene deterministicamente un grafo senza archi, mentre se $p = 1$ si ottiene un grafo completo sugli $n$ nodi.  
In generale, intuitivamente, $G_{n,p}$ conterrà **mediamente** tanti più archi quanto più $p$ si avvicina a 1, ergo mediamente un nodo avrà tanti più vicini quanto più $p$ si avvicina a 1 e quindi tante più componenti connesse... non resta che quantificare

#### **Componente gigante** nel modello Erdős-Rényi
Una **Componente gigante** in un grafo è una componente connessa (i.e. un sottoinsieme di nodi tale che tra ogni coppia di nodi esiste un cammino) che contiene una **frazione costante del numero di nodi** (quindi non basta radice di $n$ o log, serve frazione costante perché per $n \to \infty$ cresce asintoticamente molto veloce).  
Ci interessa studiarle perché molte reti reali ne contengono, come visto nelle reti sociali con l'esperimento di Leskovec e Horvitz (2008) su Messenger. Vedremo più avanti che questo è vero anche nella rete del Web.

(ricorda componente fortemente connessa per grafi diretti, dove ogni coppia di nodi ha un cammino in entrambe le direzioni, ma qui ci occupiamo di grafi non diretti)

Vogliamo capire quindi per quali valori di $p$ il grafo $G_{n,p}$ contiene una componente gigante w.h.p.  
Ricordiamo che con w.h.p. intendiamo *con probabilità proporzionale almeno a $1 - \frac{b}{n^c}$ per qualche costante $b, c > 0$* (i.e. al crescere di $n$ la probabilità tende (velocemente) a 1).
> Sia $X$ la v.a. che conta il numero di nodi nella più grande componente connessa di $G_{n,p}$.
> 
> **Teorema:**   
> 
> **se $p \gt \frac{\ln(64)}{n}$ allora $\Pr(X \ge \frac{n}{2}) \ge 1 - 2^{-\frac{n}{8}}$**
>
> Per dimostrare il Teorema abbiamo anzitutto bisogno di un Lemma:  
> 
> **Lemma**
>
> Se $X \lt \frac{n}{2}$ allora esiste un insieme di nodi $A \subseteq [n]$ tale che $\frac{n}{4} \le |A| \lt \frac{3n}{4}$ e non esiste alcun arco tra $A$ e $[n] \setminus A$.
>
> **Dimostrazione del Lemma:**
>
> Sia $C_1, C_2, ..., C_k$ le componenti connesse di $G_{n,p}$, ordinate per cardinalità non decrescente:
> $$|C_1| \le |C_2| \le ... \le |C_k|$$
> Poiché $X \lt \frac{n}{2}$, allora $|C_i| \lt \frac{n}{2} $ per ogni $i=1, 2, ..., k$.
>
> Scegliamo quindi un indice $h$ tale che:
> $$|C_1| + |C_2| + ... + |C_{h-1}| \lt \frac{n}{4} \quad \text{e} \quad |C_1| + |C_2| + ... + |C_h| \ge \frac{n}{4}$$
> Ma $h \lt k$, altrimenti se $h = k$ allora $|C_1| + |C_2| + ... |C_{k-1}| + |C_h| \lt \frac{n}{4} + \frac{n}{2} = \frac{3n}{4}$, il che è impossibile perché la somma dei cardinalità delle componenti connesse deve essere $n$.
>
> Scegliamo quindi $A = C_1 \cup C_2 \cup ... \cup C_h$, allora $A \neq \emptyset$ e $[n] \setminus A = C_{h+1} \cup ... \cup C_k \neq \emptyset$.
>
> Ma allora $|A| \ge \frac{n}{4}$ per costruzione e, come abbiamo visto:
> $$|A| = |C_1| + |C_2| + ... + |C_{h-1}| + |C_h| \lt \frac{n}{4} + \frac{n}{2} = \frac{3n}{4}$$
>
> Infine bisogna dimostrare che non esistono archi tra $A$ e $[n] \setminus A$, ma questo è ovvio in quanto per costruzione $C_1, \dots, C_h$ sono componenti connesse di $G_{n,p}$ e quindi non esistono archi connessi con nodi al di fuori della loro unione (se ci fosse un arco tra $C_i$ in $A$ (i.e. $i \le h$) e $C_j$ in $[n] \setminus A$ (i.e. $j \gt h$) allora $C_i$ e $C_j$ sarebbero la stessa componente connessa, il che è assurdo).
>

>
> $\blacksquare$

>
> Per convenzione chiamiamo l'insieme $A$ individuato dal Lemma come "insieme buono".
>   
> Osserviamo anzitutto che il Lemma ci assicura che la probabilità che la più grande componente connessa di $G_{n,p}$ contenga meno di $\frac{n}{2}$ nodi è al più la probabilità che esista un insieme buono in $G_{n,p}$, in quanto dati due eventi $E_1$ e $E_2$ con $E_1 \Rightarrow E_2$ (i.e. $E_1 \subseteq E_2$, se avviene $E_1$ allora avviene sempre $E_2$) allora $\Pr(E_1) \le \Pr(E_2)$.

Accanto al precedente teorema, si può dimostrare (e non lo faremo) anche che:
1. se $p(1-p) \lt 1$ allora quasi sicuramente tutte le componenti connesse di $G_{n,p}$ contengono $O(\log n)$ nodi;
2. se $p(1-p) = 1$ allora quasi sicuramente esiste una componente connessa di $G_{n,p}$ con $\approx n^{2/3}$ nodi;
3. se $p(1-p) \gt 1$ allora quasi sicuramente esiste una componente connessa di $G_{n,p}$ con $\Omega(n)$ nodi, e tutte le altre componenti connesse contengono $O(\log n)$ nodi.

Con "quasi sicuramente" intendiamo che la probabilità che l'evento si verifichi tende a 1 al crescere di $n$ (un po' meno forte dell'alta probabilità). 

Il teorema rappresenta necessariamente un caso specifico del punto 3, in quanto quello è l'unico caso che copre una componente gigante. Facendo i calcoli infatti:
$$p(n-1) \gt \frac{\ln(64)}{n}(n-1) \approx \ln(64) - \frac{1}{n} \gt 4 - \frac{1}{n}$$

Ma cosa rappresenta $p(1-p)$?

### **Grado dei nodi** nel modello Erdős-Rényi
Per $i \in [n]$, sia $\delta_i$ la v.a. che conta il numero di vicini del nodo $i$ in $G_{n,p}$, ovvero il grado del nodo $i$.

Per $i \in [n]$ e $j \in [n] \setminus \{i\}$, sia $a_{i,j}$ la v.a. bernoulliana che vale 1 se esiste un arco tra $i$ e $j$ in $G_{n,p}$, e 0 altrimenti.

Allora $\delta_i = \sum_{j \in [n] \setminus \{i\}} a_{i,j}$, e quindi il valore atteso del grado del nodo $i$ è:
$$\mathbb{E}[\delta_i] = \sum_{j \in [n] \setminus \{i\}} \mathbb{E}[a_{i,j}] = \sum_{j \in [n] \setminus \{i\}} (1 p + 0 (1-p))= (n-1)p$$
sarebbe a dire che se $p$ è costante --> il grado dei nodi cresce in media linearmente con il numero di nodi.

Questo va fortemente in contrasto con le reti reali, dove il grado dei nodi cresce molto più lentamente al crescere degli utenti della rete --> al fine di modellare significativamente le reti reali di grandi dimensioni è opportuno che $p$ sia una funzione decrescente di $n$, del tipo:
$$p = p(n) = \frac{\lambda}{n} \quad \text{con } \lambda > 0$$

Fissiamo ora $k \lt n$ intero. Vogliamo calcolare **con quale probabilità un nodo di $G_{n,p}$ ha grado esattamente $k$.  
Sia $i \in [n]$ --> la probabilità che $i$ abbia grado $k$ è la probabilità che esistano esattamente $k$ archi tra $i$ e gli altri $n-1$ nodi.  
Poiché il numero di k-uple di nodi scelti nell'insieme $[n] \setminus \{i\}$ è $\binom{n-1}{k}$ allora:
$$\Pr(\delta_i = k) = \binom{n-1}{k} p^k (1-p)^{n-1-k}$$

e poiché per l'approssimazione di Stirling $k! \approx \sqrt{2 \pi k} \left(\frac{k}{e}\right)^k$ allora:
$$\Pr(\delta_i = k) \lt \frac{(\lambda e)^k}{\sqrt{2 \pi k} k^k} e^{-\lambda} = \frac{e^{-\lambda}}{\sqrt{2 \pi k}} \left(\frac{\lambda e}{k}\right)^k$$
ossia possiamo concludere che **la probabilità che un generico nodo abbia grado $k$ decresce esponenzialmente al crescere di $k$.** (decresce come $k^{-k}$)

A questo punto ci chiediamo quale sia la frazione del numero di nodi con grado $k$ in $G_{n,p}$.  
Sia allora $F_k$ la v.a. che esprime tale frazione, e sia $\delta_{i,k}$ la v.a. che vale 1 se il nodo $i$ ha grado $k$ e 0 altrimenti:
$$\delta_{i,k} = \begin{cases}
1 & \text{se } \delta_i = k \\ 0 & \text{altrimenti}
\end{cases}$$
Allora $F_k = \frac{1}{n} \sum_{i=1}^{n} \delta_{i,k}$.
