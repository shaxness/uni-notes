Abbiamo visto come, in un grafo generato secondo il modello di grafi aleatori Erdős-Rényi, il numero di nodi con grado $k$ decresce esponenzialmente in $k$.  

Cerchiamo di capire perché la decrescita è così veloce.  
Il **Teorema del Limite Centrale** (intuitivamente e informalmente) afferma come *la somma di molti valori aleatori indipendenti si distribuisce approssimativamente come una distribuzione normale attorno al valore atteso della somma*. Il valore della somma quindi decresce molto velocemente quando ci si allontana dal valore atteso!

<img src="imgs/2_1.png" alt="Distribuzione normale" width="400"/>

Il TLC si lega al nostro caso in quanto nel grafo $G_{n,p}$ gli archi vengono aggiunti come una sequenza di oggetti, indipendentemente gli uni dagli altri. Poiché il grado di un nodo è quindi somma di eventi indipendenti:
$$\delta_i = \sum_{j \in [n] - i} e_{ij}$$
dove $e_{ij}$ è una variabile aleatoria che vale 1 se esiste l'arco $(i,j)$, allora ci troviamo proprio nel contesto del TLC.

Il punto cruciale è che **la porzione di grafo già costruita (i.e. gli archi che sono già stati inseriti in $G_{n,p}$) non influenza in alcun modo la probabilità di inserire un nuovo arco**.

Consideriamo però un caso realistico: **il grafo orientato del web**, con nodi che rappresentano le pagine web e se la pagina $a$ contiene un hyperlink alla pagina $b$ --> $(a,b)$ è un arco (diretto) del grafo.  

Nel 2000 uno studio di Broder et al., al fine di analizzare la distribuzione dei gradi dei nodi (archi entranti) di questo grafo, ha osservato come differisca **sostanzialmente** da quanto indicato dal TLC: **la frazione di pagine Web con grado entrate $k$ è proporzionale a $1/k^c$ (power law) per qualche costante $c > 1$, invece che a $1/k^k$** --> sarebbe a dire che il numero di nodi del grafo del Web con grado entrante elevato **è molto maggiore di quanto si aspetterebbe, assumendo che gli archi si formino indipendentemente gli uni dagli altri**.

(una funzione che decresce come l'inverso di un polinomio è detta **Power Law**).

Ma come hanno fatto i ricercatori ad accorgersi del fenomeno?  
Nel senso, partendo da un grafico (discreto) di un grafo che per ogni intero $k$ in ascissa riporta in ordinata il numero di nodi del grafo con grado $k$, come può uno accorgersi che la funzione decresce come l'inverso di un polinomio invece che come l'inverso di un esponenziale?

<img src="imgs/2_2.png" alt="Distribuzione dei gradi del grafo del Web" width="400"/>

Per farlo si cambia semplicemente la scala del grafico a quella logaritmica, considerando quindi grafico log-log (asse $x$ e asse $y$ rappresentano rispettivamente $\ln(x)$ e $\ln(y)$) --> invece di rappresentare $y = f(x)$ si rappresenta $\ln(y) = \ln(f(x))$, in questo modo (avendo $f(k) = 1/k^c$) otteniamo:
$$\ln(f(k)) = \ln(1/k^c) = -c \ln(k)$$
ossia grossomodo una retta con pendenza $-c$.

<img src="imgs/2_3.png" alt="Distribuzione dei gradi del grafo del Web in scala log-log" width="300"/>

### Rich and get Richer: Barabási-Albert Model
Cerchiamo quindi di capire il perché si verifica il fenomeno della Power Law.

Supponiamo di costruire una pagina Web, come una in cui descriviamo come funziona Erdős-Rényi --> sarebbe opportuno inserire un hyperlink a qualche pagina che descrive qualche altro modello generativo... ma come scegliamo quale pagina?  
Ha senso, tra le centinaia di pagine esistenti di questo tipo, scegliere quella che ci sembra più autorevole, ma come capire l'autorevolezza di una pagina? Magari se la pagina appare tra i primi risultati su un motore di ricerca (e questo vedremo accadere proprio quando una pagina ha tanti link entranti) --> **aumentiamo il numero di pagine che hanno già un grado elevato, scegliendo quindi quale arco aggiungere alla rete Web sulla base di archi già esistenti**.

Qui sta la differenza che non permette di applicare direttamente il TLC: l'aggiunta di un nuovo arco non è più evento indipendente dagli archi già aggiunti precedentemente nella rete.

Lo stesso fenomeno avviene anche in altri contesti, come le reti sociali: la rete mostra i post dei nostri contatti, che se commentano post di pagine che loro seguono (tipicamente già famose), se piacciono anche a noi potremmo seguirle a sua volta contribuendo a farle diventare ancora più popolari. Si può dire lo stesso su personaggi famosi come attori, politici etc...

A questo punto vogliamo definire un modello generativo di grafi per il quale, a differenza di Erdős-Rényi, l'aggiunta di un nuovo arco dipenda dagli archi già presenti nel grafo e che tale modello *esibisca una Power Law*.

Per il nostro modello, assumeremo che **gli individui tendano a copiare il comportamento degli altri** (abbiamo visto infatti come questo valga già nel processo di creazione di pagine web e nella rete sociale; lo vedremo ancora meglio quando parleremo di Herding).  
In particolare le pagine vengono create una alla volta, **in sequenza**, e quando viene creata una pagina si decide a quali essa debba puntare, collegando quindi con un arco diretto.

Sia $p \in [0,1]$, i nodi vengono aggiunti al grafo in time-step discreti:
- al passo 1 viene creato il nodo 1
- al passo 2 vengono creati il nodo 2 e l'arco (diretto) $(2,1)$
- finita la fase di inizializzazione, i nodi e gli archi si creano secondo la regola seguente:  
*al passo $i$ vengono creati il nodo $i$ e un arco uscente da $i$, scegliendo u.a.r. un nodo $a \lt i$ e:*
    - *w.p. $p$ si crea l'arco $(i,a)$*
    - *w.p. $1-p$ viene creato l'arco $(i,b)$, dove $(a,b)$ è (l'unico) arco uscente da $a$*

Così, intuitivamente, man mano che il numero di archi entranti in un nodo aumenta cresce anche la probabilità che quel nodo venga selezionato come estremo di un arco uscente da un nodo di nuova creazione --> **abbiamo costruito un modello che descrive il fenomeno Rich Get Richer.**  
Questo modello è stato proposto nel 1999 ed è chiamato **modello di Barabási-Albert**.

(si osservi come ogni nodo $i \gt 1$ abbia esattamente un arco uscente, mentre il nodo 1 ne ha zero)

<img src="imgs/2_4.png" alt="Esempio di grafo generato secondo il modello di Barabási-Albert" width="500"/>

Un modo equivalente per descrivere il passo iterativo post inizializzazione è il seguente, utile per la formalizzazione del modello:
- *al passo $i$ vengono creati il nodo $i$ e un arco uscente da $i$, in particolare*:
    - *w.p. $p$ si sceglie u.a.r. un nodo $a \lt i$ e si crea l'arco $(i,a)$*
    - *w.p. $1-p$ si sceglie u.a.r. un nodo $a \lt i$ e, detto $(a,b)$ l'arco uscente da $a$, si crea l'arco $(i,b)$*

#### Barabási-Albert esibisce una Power Law
Resta da verificare se i grafi generati in accordo con questo modello esibiscono una Power Law (i.e. la funzione che descrive il numero atteso di nodi di grado entrante $k$ si comporta come l'inverso di un polinomio).

Per farlo formalizziamo: per ogni coppia di interi $i, j$ t.c. $i \gt j$ e $i \ge 2$ introduciamo la v.a.:
$$d_{ij} = \begin{cases}
1 & \text{se (i,j)} \in E \\ 
0 & \text{altrimenti}
\end{cases}$$
Allora, in accordo con le regole che abbiamo descritto:
$$\Pr(d_{ij} = 1) = p \cdot \text{prob. che venga scelto il nodo $j$} + (1-p) \cdot \text{prob. che venga scelto un nodo $h$ che abbia arco uscente proprio verso $j$, i.e. t.c. } d_{hj} = 1$$
sviluppando i calcoli:

<img src="imgs/2_5.png" alt="Calcolo della probabilità di avere un arco (i,j)" width="500"/>

dove si ottiene per il secondo addendo prob. dell'unione di tutti gli eventi per cui, per tutti i possibili $h \lt i$ t.c. esiste l'arco $(h,j)$, scelgo proprio $h$ come nodo.  
Questi eventi sono disgiunti in quanto nella scelta viene effettivamente scelto un solo nodo $h$ (non possono verificarsi diversi eventi contemporaneamente) --> la probabilità dell'unione è la somma delle probabilità ($\Pr(A \cup B) = \Pr(A) + \Pr(B) - \Pr(A \cap B)$ ma l'intersezione è nulla, eventi disgiunti).

Nella parte finale dei passaggi invece il fatto che il valore sia 1 se esiste l'arco $(h,j)$ equivale a prendere come elemento della somma la bernouilli $d_{hj}$.

Abbiamo così ottenuto:
$$\Pr((i,j) \in E) = \Pr(d_{ij} = 1) = \frac{p}{i-1} + \frac{1-p}{i-1} \sum_{h=1}^{i-1} d_{hj}$$
**Oss. 1**: ha senso calcolare $\Pr((i,j) \in E)$ per $i \gt 1$ (in quanto il nodo 1 non ha archi uscenti)  

**Oss. 2**: ogni nodo $i \gt 1$ ha esattamente un arco uscente --> deve essere $\Pr(\exists j \lt i : (i,j) \in E) = 1$.  
Nonostante sia abbastanza chiaro a livello intuitivo per via del funzionamento del modello, lo si può dimostrare per induzione su $i$.  
Sappiamo che $$\Pr(\exists j \lt i : (i,j) \in E) = \sum_{j=1}^{i-1} \Pr((i,j) \in E)$$
- **Base**: per $i=2$ è vero in quanto l'unico arco uscente da 2 è $(2,1)$  --> $\sum_{j=1}^{1} \Pr((2,j) \in E) = \Pr((2,1) \in E) = 1$
- **Passo induttivo**: supponiamo che $\sum_{j=1}^{a-1} \Pr((a,j) \in E) = 1$ sia vero per ogni $a \le i-1$ e mostriamo che vale anche per $i$:

    <img src="imgs/2_6.png" alt="Dimostrazione per induzione" width="500"/>

    (posso scambiare l'ordine delle sommatorie in quanto entrambe somme finite)

    L'osservazione in rosa sfrutta l'ipotesi induttiva: fissato $h \lt i$ questo ha esattamente un solo arco uscente e quindi $\sum_{j=1}^{i-1} d_{hj} = 1$ ($d_{hj}$ è la bernouilli che vale 1 se esiste l'arco $(h,j)$, quindi per ogni $h$ esiste un solo $j$ t.c. $d_{hj} = 1$).

**Oss. 3**: Abbiamo la formula
$$\Pr((i,j) \in E) = \Pr(d_{ij} = 1) = \frac{p}{i-1} + \frac{1-p}{i-1} \sum_{h=1}^{i-1} d_{hj}$$
e abbiamo osservato come ogni nodo $i \gt 1$ abbia esattamente un arco uscente --> $\sum_{h=1}^{i-1} d_{hj}$ rappresenta null'altro che il **grado entrante di $j$ al passo $i$**.

La regola per costruire il grafo può quindi essere interpretata come segue:  
**il nodo $j$ a cui connettere $i$ è scelto u.a.r. w.p. $p$, oppure con probabilità proporzionale al grado entrante di $j$ w.p. $1-p$** (in quanto $\sum_{h=1}^{i-1} d_{hj}$ è proprio il grado entrante di $j$ al passo $i$).

Questa terza osservazione rappresenta quindi un terzo modo valido per descrivere il modello di Barabási-Albert e il fenomeno Rich Get Richer.

#### Dimostrazione della Power Law
Siamo pronti a dimostrare che il modello appena definito esibisce una Power Law.  
La dimostrazione procede attraverso i seguenti passi intermedi:
1. definizione della legge aleatoria che esprime la varizione del grado di un nodo nel tempo
2. approssimazione deterministica e continua della legge aleatoria al punto 1., che porterà ad un'equazione differenziale
3. risoluzione dell'equazione differenziale: la soluzione costituirà un'approssimazione della funzione che esprime il grado di un nodo nel tempo
4. individuazione della Power Law

**Legge aleatoria per la variazione del grado**: 

Sia $D_j(t)$ la v.a. che esprime il numero di archi entranti al nodo $j$ al passo $t$ di generazione del grafo. Questa variabile è chiaramente definita per $t \ge j$ e per ogni $j \ge 1$. 

Al passo $t = j$ in particolare, il grado entrante di $j$ è 0, in quanto il nodo $j$ è appena stato creato e non ha ancora archi entranti.

Esprimiamo ora la variazione attesa nel tempo della variabile $D_j(t)$: al passo $t+1$, il grado di $j$ può essere invariato rispetto al passo $t$ oppure aumentato di una sola unità. In particolare:
$$D_j(t+1) = D_j(t) + 1 \iff \text{è stato creato l'arco (t+1, j)}$$
Quindi:
$$\Pr(D_j(t+1) - D_j(t) = 1) = \Pr(d_{t+1,j} = 1) = \frac{p}{t} + \frac{1-p}{t} \sum_{h=1}^{t} d_{hj} = \frac{p}{t} + \frac{1-p}{t} D_j(t)$$
Ossia:
$$\Pr(D_j(t+1) - D_j(t) = 1) = \frac{p}{t} + \frac{1-p}{t} D_j(t)$$

**Approssimazione deterministica e continua**:

Per ogni $j \ge 1$, definiamo anzitutto una funzione **discreta deterministica** $X_j(t)$ che "assomiglia" a $D_j(t)$. Questa funzione è definita per ogni $t \ge j$ e ha le seguenti proprietà:
- $X_j(j) = 0$ (come $D_j(j)$)
- $X_j(t+1) - X_j(t) = \frac{p}{t} + \frac{1-p}{t} X_j(t)$

A questo punto approssimiamo il comportamento di $X_j(t)$ con una funzione $x_j(t)$ definita invece su dominio continuo, e ancora definita per $t \ge j$ e con proprietà:
- $x_j(j) = 0$ 
- $\frac{d}{dt} x_j(t) = \frac{p}{t} + \frac{1-p}{t} x_j(t)$ (equazione differenziale)

Non è detto che l'andamento della funzione $x_j(t)$ sia vicino al comportamento della v.a. $D_j(t)$, ma le analogie tra le due sono forti e vedremo più avanti quanto effettivamente le due si assomigliano.

**Risoluzione dell'equazione differenziale**: 

<img src="imgs/2_7.png" alt="Risoluzione dell'equazione differenziale" width="500"/>

In pratica si integrano entrambi i membri rispetto a $t$, e a sx dt e dt si semplificano. A questo punto sempre a sx si fa un cambio di variabile: il denominatore è $p + (1-p) x_j(t)$ la cui derivata w.r.t. $x_j(t)$ è $1-p$ --> per avere al numeratore la derivata del denominatore si moltiplicano entrambi i membri per $(1-p)$.  
Ponendo $y = p + (1-p) x_j(t)$ --> $dy = (1-p) dx_j(t)$ e quindi:
$$\int \frac{dy}{y} = (1-p) \int \frac{1}{t} dt$$

<img src="imgs/2_8.png" alt="Risoluzione dell'equazione differenziale" width="500"/>

A questo punto usiamo la formula $\int \frac{du}{u} = \ln|u| + c$ e otteniamo quanto si vede nella seconda riga (compare una sola $c$ per entrambi gli integrali perché le due costanti di integrazione si raccolgono in una sola).

Si elimina poi il log esponenziando. Poiché $e$ e $c$ sono costanti, chiamiamo $e^c = H$. Per trovare $H$ sfruttiamo la condizione iniziale $x_j(j) = 0$ e otteniamo $H = \frac{p}{j^{1-p}}$ e sostituendo w.r.t. $x_j(t)$ otteniamo:

**Individuazione della Power Law**:

Calcoliamo infine, nel modello deteministico e continuo costruito, dati $k$ e $t$, quale frazione dei nodi al passo $t$ ha "grado" $k$ (virgolette perché stiamo usando una misura derivata secondo l'approssimazione).

Poiché l'insieme dei nodi al passo $t$ è $[t] = {1, 2, \dots, t}$, definiamo $A_t(k)$ come l'insieme $A_t(k) = \{j \le t: x_j(t) \ge k\}$, ossia l'insieme dei nodi al passo $t$ che hanno grado almeno $k$. Vogliamo calcolare:
$$\frac{1}{t} |A_t(k) - A_t(k+1)| = \frac{1}{t} |A_t(k)| - \frac{1}{t} |A_t(k+1)|$$
(in quanto $A_t(k+1) \subseteq A_t(k)$, quindi $|A_t(k) - A_t(k+1)| = |A_t(k)| - |A_t(k+1)|$ poiché togliere dal primo un insieme fatto solo di elementi in esso contenuti significa sottrarre le loro cardinalità).

Per definizione, $j \in A_t(k) \iff j \le t$ e $x_j(t) \ge k$.  
Ma $x_j(t) = \frac{p}{1-p} \left[ \left( \frac{t}{j} \right)^{1-p} - 1 \right]$, quindi $x_j(t) \ge k \iff \frac{p}{1-p} \left[ \left( \frac{t}{j} \right)^{1-p} - 1 \right] \ge k$. Svolgendo i calcoli:

<img src="imgs/2_10.png" alt="Individuazione della Power Law" width="400"/>

Manca l'altra parte: come detto $j \in A-t(k) \iff j \le t$ e poiché $\frac{1-p}{p} k + 1 \gt 1$ (in quanto il grado $k \ge 0$) e $-\frac{1}{1-p} \lt 0$ allora:
$$\left[ \frac{1-p}{p} k + 1 \right]^{-\frac{1}{1-p}} \lt 1$$
ossia effettivamente:
$$t \left[ \frac{1-p}{p} k + 1 \right]^{-\frac{1}{1-p}} \le t$$
Per cui, in conclusione:
$$j \in A_t(k) \iff j \le t \left[ \frac{1-p}{p} k + 1 \right]^{-\frac{1}{1-p}}$$

Ma allora possiamo riscrivere l'insieme come:
$$A_t(k) = \left\{j: j \le t \left[ \frac{1-p}{p} k + 1 \right]^{-\frac{1}{1-p}}\right\}$$
e quindi:
$$|A_t(k)| = t \left[ \frac{1-p}{p} k + 1 \right]^{-\frac{1}{1-p}} $$
Tornando a noi, volevamo calcolare $\frac{1}{t} \left[ |A_t(k)| - |A_t(k+1)| \right]$ e quindi:

(in pratica definiamo $F(k)$ per comodità, dato che la $t$ si semplifica, poi approssimiamo la differenza sfruttando Taylor con $-\frac{dF(k)}{dk}$ e derivando otteniamo finalmente una Power Law, in quanto funzione inversa di un polinomio in $k$, in particolare è una Power Law con esponente $1 + \frac{1}{1-p}$).

##### Conclusioni sull'analisi del modello di Barabási-Albert
Per analizzare il modello Rich get Richer, abbiamo approssimato mediante una funzione deterministica e continua la legge probabilistica che esprime al variazione del grado di un nodo tra due passi consecutivi, ma *non abbiamo idea di quanto tale approssimazione si avvicini effettivamente al modello discreto e aleatorio originale!*

Tuttavia nel 2005 è stato dimostrato da Bollobas et Riodan che **nel modello Rich get Richer che abbiamo proposto, w.h.p., la frazione del numero di nodi di grado $k$ è proporzionale a $k^{-\frac{1}{1-p} - 1}$**, esattamente come nell'approssimazione che abbiamo discusso!!!
