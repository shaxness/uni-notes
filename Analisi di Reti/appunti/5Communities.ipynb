Finora abbiamo studiato grafi che modellano reti a lvl statistico, tramite modelli generativi, disinteressandoci delle peculiarità dei singoli nodi e focalizzandoci invece sulle proprietà che valevano *mediamente* per tutti i nodi. 

Da questo punto del programma invece cambiamo punto di vista: analizzeremo la posizione dei nodi nella rete per studiare:
- se è possibile evidenziare particolari strutture nella rete
- se tutti i nodi si trovano, grosso modo, nella stessa situazione nella rete *oppure* se vi sono differenze osservabili tra nodo e nodo
- se i nodi possono trarre vantaggio (e di quale tipo) da queste differenze

### Granovetter: weak ties, strong ties e local bridges

Negli anni '60, il sociologo statunitense Mark Granovetter nel preparare la tesi di dottorato intervistò un gruppo di persone che avevano cambiato lavoro di recente.  
Granovetter fece loro una serie di domande per capire in che modo erano venuti a conoscenza della possibilità di ottenere l'impiego che avevano poi accettato.

Il risultato fu che molti degli intervistati avevano avuto le informazioni necessarie attraverso una comunicazione da parte di qualcuno che conoscevano - **ma la cosa interessante è che spesso questi non erano amici stretti, ma più conoscenti superficiali**!

Strano perché ci aspetteremmo che siano gli amici stretti a darci una mano per cercare un lavoro... l'esperimento suggerisce che invece *non è tanto la forza del legame ad aver aiutato a trovare lavoro*, quanto **il tipo di informazione che un legame riesce a veicolare**.

Il punto intuitivamente è che gli amici stretti frequentano grossomodo gli stessi posti che frequentiamo noi --> hanno accesso alle stesse informazioni che abbiamo già noi; invece *i conoscenti alla lontana provengono da "zone" diverse della rete alle quali non avremmo accesso senza il loro aiuto*, ampliano il nostro "raggio d'azione"...

Se consideriamo il grafo $G = (V, E)$ che modella la rete, gli archi che "aumentano" le nostre informazioni sono quelli che ci collegano a regioni della rete *inaccessibili* ai nostri amici  
--> definiamo **bridge** un arco la cui rimozione disconnette la rete (segue dall'esperimento di Granovetter che sono proprio i bridge gli archi con maggiore valore informativo!)

Tuttavia sappiamo bene che la rete contiene con buona probabilità **componenti giganti densamente connesse** --> la presenza di bridge è poco probabile...  
--> rilassiamo la definizione: una arco è **local bridge** se i nodi a lui incidenti **non hanno vicini in comune**: $(u, v)$ è un local bridge se $N(u) \cap N(v) = \emptyset$ dove $N(u)$ è l'insieme dei nodi adiacenti a $u$ ($u$ non compreso).

Ma l'esperimento di Granovetter fornisce anche un altro tipo di indicazione oltre al potere informativo di bridge e local bridge:  
**spesso l'informazione era arrivata da qualcuno conosciuto solo superficialmente**!  

Allora all'interno della rete sociale possiamo riconoscere relazioni forti e relazioni deboli --> possiamo modellare una rete di questo tipo con un grafo $G$ in cui gli archi sono partizionati in due sottinsiemi: **strong ties** e **weak ties**, i.e.  
$$G = (V, S \cup W) \quad \text{con } S \cap W = \emptyset$$
Come ha mostrato Granovetter, sono i weak ties a veicolare le info a cui non avevamo accesso tramite gli archi forti: questa è la **forza degli archi deboli**.

Recuperiamo ora il concetto di **chiusura triadica** nel grafo dei weak e strong ties: due nostri cari amici è ben probabile che entrino in contatto tra loro -->  
**se $(a, b)$ e $(b, c)$ sono archi forti $\Rightarrow$ è probabile che prima o poi si creerà $(a, c)$ (forte o debole che sia)**, ossia si genera una chiusura **triadica**.

Per descrivere queste reti quindi ci servono **grafi dinamici** (i.e. che evolvono nel tempo).  
Lasciamo ora che la rete evolva, e che si formino via via triangoli (i.e. chiusure triadiche) fino a raggiungere configurazione stabile in cui la rete non cambia più nel tempo.  
Allora potremo dire di $G = (V, S \cup W)$ che, una volta raggiunta configurazione stabile, tutti i triangoli possibili si sono formati i.e. *tutte le coppie di nodi collegati da un arco forte a un altro nodo in comune sono ora collegati tra loro!*

Formalizziamo il concetto:  
Sia $G = (V, S \cup W)$: allora un nodo $u$ soddisfa la **proprietà di chiusura triadica forte** (in breve **strong triadic closure property STCP**) se per ogni coppia di archi forti incidenti su $u$, i loro estremi sono collegati da un arco (forte o debole che sia): 
$$u \in V \text{ soddisfa STCP } \iff \forall (u,v) \in S \forall (v, z) \in S [(v,z) \in S \cup W]$$
Diremo infine che $G$ soddisfa la STCP se *tutti i suoi nodi la soddisfano*.

> Teorema: 
>
> Sia $G = (V, S \cup W)$: se un nodo $u \in V$ soddisfa la STCP e se esistono due nodi distinti $x, z$ t.c. $(u, x) \in S$ e $(u, z)$ è local bridge --> $(u, z) \in W$
>
> Dim:
>
> Procediamo per assurdo. Poiché $(u, z)$ è local bridge $\Rightarrow N(u) \cap N(z) = \emptyset$.  
> Ma allora se fosse $(u, z) \in S$ allora poiché $(u, x) \in S$ e $u$ soddisfa la STCP, dovrebbe essere che $(x, z) \in S \cup W$ ossia $x$ sarebbe sia vicino di $z$ che di $u$ $x \in N(u) \cap N(z) \ne \emptyset$ --> assurdo!
>
> <img src="imgs/5_1.png" width="400" />
>
> $\blacksquare$

Questo teorema mostra come *una **proprietà globale (essere local bridge)** si rifletta su una **proprietà locale (essere weak tie)***.  
(weak tie locale perché un nodo sa da solo se un suo vicino è amico stretto o conoscente, invece local bridge globale perché un nodo deve chiedere a tutti i vicini se qualcuno di loro conosce un suo conoscente per vedere se l'arco è local bridge o meno).  
Questo si riflette anche sull'esperimento di Granovetter: i weak ties, in quanto portatori di nuova informazione, erano anche local bridges!

### Clusters
Se $(a,b)$ e $(a,c)$ sono **strong ties**, abbiamo detto che è probabile si formi nel tempo anche l'arco $(b,c)$, ossia che si una chiusura triadica.  
Naturalmente, una chiusura triadica dopo l'altra, si formerà nella rete **un gruppo di nodi fortemente coeso**, con elevato grado di interconnessione solo tra i nodi che lo compongono: si parla di **cluster**/**comunità**.

*Ma quando possiamo dire che un gruppo di nodi è abbastanza coeso da poter essere definito comunità?*

Intanto, per misurare il grado di coesione di un nodo $u$ in un gruppo di nodi è stato definito il **coefficiente di clustering c(u)**:
$$c(u) = \frac{|{(x,y) \in E : x,y \in N(u)}|}{\binom{|N(u)|}{2}}$$
dove:
- il numeratore conta il numero di archi tra i vicini di $u$ (ossia quanti triangoli sono formati da $u$ e due suoi vicini)
- il denominatore conta il numero massimo di archi che potrebbero esserci tra i vicini di $u$ $\left(\binom{|N(u)|}{2} = \frac{|N(u)|(|N(u)|-1)}{2}\right)$

Intuitivamente cioè il coefficiente di clustering misura **quanto un nodo è "ben inserito" nella rete costituita dai vicini**: un nodo con coefficiente basso è mal inserito nel gruppo dei suoi amici (si trova in una posizione periferica in quel gruppo), mentre un nodo con coefficiente alto è ben inserito (è come un elemento centrale nel gruppo dei suoi amici)  
--> non a caso il coefficiente di clustering è **indice di centralità** di un nodo.

A questo punto, dato $C$ sottinsieme di nodi, possiamo definire per ogni nodo $u \in C$ il **coefficiente di clustering di $u$ relativo però al sottinsieme $C$** come:  
$$c_C(u) = \frac{|{(x,y) \in E : x,y \in N(u) \cap C}|}{\binom{|N(u) \cap C|}{2}}$$
ossia lo stesso coefficiente di clustering, ma calcolato solo considerando i vicini di $u$ che appartengono al sottinsieme $C$.

Se tutti i nodi contenuti in $C$ hanno coefficiente di clustering relativo a $C$ **elevato** --> **possiamo ben pensare che $C$ sia una comunità**.  

Ma **quanto elevato** deve esseere il coeff. di clustering relativo a $C$ dei nodi in $C$ per poter definire $C$ una comunità?  
E poi sì, ok, il coeff di clustering sembra promettente per definire una comunità, ma non possiamo definire una comunità anche sulla base di altri criteri?

La risposta alla seconda domanda è sì: esistono tanti concetti diversi di comunità che sono stati proposti in letteratura. In queste lezioni parleremo in particolare di:
- cut-communities
- web-communities
- metodi agglomerativi e partitivi per individuare comunità
- descriveremo in particolare un metodo partitivo che porterà ad individuare comunità basate su un altro concetto di centralità dei nodi: la **betweenness**

#### Cut-communities
Una **cut-community** per un grafo $G = (V, E)$ è un sottinsieme proprio e non vuoto $C$ dei nodi di $G$ che minimizza gli **archi del taglio** (i.e. gli archi che collegano nodi in $C$ a nodi in $V \setminus C$).

Formalmente, dato $G = (V, E)$ diremo che $C \subset V$ è una **cut-community** se $C \ne \emptyset$ e:
$$|\{(u,v) \in E: u \in C, v \in V \setminus C\}| = \min_{C' \subset V, C' \ne \emptyset} |\{(u,v): u \in C', v \in V \setminus C'\}|$$
ossia il numero di archi che collegano $C$ al resto della rete è minimo tra tutte le possibili bipartizioni dei nodi del grafo.

Diciamo che dato $G = (V, E)$ e una coppia di nodi $s, t \in V$, un *taglio minimo w.r.t. la coppia $(s, t)$* è un sottinsieme proprio e non vuoto $C \in V$ dei nodi di $G$ che contiene $s$ ma non $t$ e che minimizza il numero di archi del taglio.  
Sappiamo bene che **calcolare il taglio minimo w.r.t. una coppia di nodi $(s, t)$ è facile**, esistono molti algoritmi che lo fanno in tempo polinomiale (es. Ford-Fulkerson, Edmonds-Karp etc..)

<img src="imgs/5_2.png" width="500" />

Per calcolare una cut-community di un grafo possiamo quindi procedere così:
- $\forall$ coppia di nodi distinti $s, t \in V$ calcoliamo l'insieme $C_{s,t}$ tale che $s \in C_{s,t}$, $t \in V \setminus C_{s,t}$ e $C_{s,t}$ minimizza il numero di archi del taglio
- La cut-community cercata è quella che minimizza il numero di archi del taglio tra tutte le cut-communities calcolate:
    $$|{(u,v): u \in C_{x,y} \cap v \in V \setminus C_{x,y}}| = \min_{s,t \in V: s \ne t} |{(u,v): u \in C_{s,t} \cap v \in V \setminus C_{s,t}}|$$

Sembra tutto bello, ma c'è l'inganno... questi algoritmi non permettono di controllare la dimensione della cut-community trovata: è possibile che il taglio minimo w.r.t. una coppia di nodi $(s, t)$ sia costituito da un solo nodo (es. $C_{s,t} = \{s\}$) e quindi la cut-community trovata sia composta da un solo nodo --> non è una comunità!

<img src="imgs/5_3.png" width="500" />

nell'esempio ogni nodo ha grado 7, e i tagli minimi sono tutti e soli quelli che isolano singoli nodi (se ne prendo di più vado sempre oltre 7 archi del taglio) --> una comunità con un solo nodo non è ragionevole!

#### Web-communities
Una **web-community** è un sottinsieme di nodi di un grafo ciascuno dei quali **ha più vicini all'interno della comunità che all'esterno**.

Formalmente, dato $G = (V, E)$ diremo che $C \subset V$ è una **(strong) web-community** se:
$$C \ne \emptyset \text{ e } \forall u \in C: |N(u) \cap C| > |N(u) - C| = |N(u) \cap (V \setminus C)|$$
o equivalentemente poiché $|N(u) \cap C| + |N(u) - C| = |N(u)|$ possiamo riscrivere la condizione come: 
$$\frac{|N(u) \cap C|}{|N(u)|} > \frac{1}{2}$$
(più della metà dei vicini di ogni nodo in $C$ appartiene a $C$ stesso)

Analogamente diremo che dato $G = (V, E)$, $C \subset V$ è una **(weak) web-community** se:
$$C \ne \emptyset \text{ e } \forall u \in C: |N(u) \cap C| \ge |N(u) - C| = |N(u) \cap (V \setminus C)|$$
o equivalentemente:
$$\frac{|N(u) \cap C|}{|N(u)|} \ge \frac{1}{2}$$
Questa definizione, proprio come la cut-community, è semplice ed intuitivamente ragionevole, nonché generalizzabile richiedendo che $\frac{|N(u) \cap C|}{|N(u)|} \gt \alpha$ (o $\ge \alpha$) per un opportuno $\alpha \in (0,1)$.

Le definizioni di cut-community e web-community non sono del tutto scorrelate, come si vede dal seguente teorema.

> **Teorema**
>
> sia $G = (V, E)$ un grafo. Se $C \subset V$ è una cut-community di $G$ tale che $|C| \ge 2$, allora $C$ è anche una weak web-community di $G$.
>
> **Dim**:
>
> Sia $t(C)$ il numero di archi del taglio $C$.  
> Supponiamo per assurdo che $C$ non sia una weak web-community.  
> Ma allora esiste un nodo $u \in C$ tale che $|N(u) \cap C| < |N(u) - C|$.  
> Ma dato che $C \gt 1$ --> esiste in $C$ un nodo $v \ne u$ i.e. $C - \{u\} \ne \emptyset$ --> possiamo considerare il taglio $C' = C - \{u\}$.
>
> Dimostriamo che il taglio in questione $C' = C - \{u\}$ ha meno archi del taglio $C$, arrivando così all'assurdo.  
> $$t(C') = |{(x,y): x \in C \setminus \{u\}, y \in (V \setminus C) \cup \{u\}}|$$
> spostando $u$ in questo modo cambiano ruolo solo gli archi incidenti ad $u$: quelli tra $u$ e i nodi in $V \setminus C$ prima erano archi del taglio e ora sono esterni a $C'$, mentre quelli tra $u$ e i nodi in $C$ prima erano interni a $C$ e ora sono archi del taglio.
>
> Quindi partendo dal numero di archi del vecchio taglio **sottraiamo** quelli tra $u$ e l'esterno (che sono $|N(u) - C|$) e **aggiungiamo** quelli tra $u$ e l'interno (che sono $|N(u) \cap C|$):
> $$t(C') = t(C) - |N(u) - C| + |N(u) \cap C|$$
> Ma come detto all'inizio poiché $C$ non è weak web-community --> $u$ è tale che $|N(u) \cap C| < |N(u) - C|$ --> $t(C') < t(C)$, assurdo perché $C$ era cut-community!
>
> Quindi per forza se $C$ è cut-community con $|C| \ge 2$ allora $C$ è anche weak web-community.
>
> $\blacksquare$

Quindi una cut-community con almeno due nodi è anche una weak web-community.  
Abbiamo già osservato però che nel calcolare una cut-community è difficile controllarne la cardinalità in quanto gli algoritmi efficienti che calcolano cut communities possono restituire comunità contenenti un solo nodo e quindi poco significative. 

Inoltre non è detto che una cut-community di un solo nodo sia anche una web community: in figura si vede come $C = {u_0}$ sia cut-community, ma non una web-community perché $u_0$ ha ovviamente più vicini in $V \setminus C$ che in se stesso!

<img src="imgs/5_4.png" width="200"/>

In effetti, più che individuare una singola comunità in un grafo, quello che ci interessa è partizionarlo in comunità: le motivazioni sono diverse:
- conoscere le comunità potrebbe aiutarci a capire come fluisce l'informazione nella rete (in linea con Granovetter)
- oppure capire come si diffondono idee, innovazioni o epidemie in quella rete
- ma anche studiare una rete di grandissime dimensioni, riducendone la granularità (considerando cioè le comunità come una sorta di macro-nodi e studiando il grafo in questione)

Si osserva ovviamente che **se $C$ è cut community allora anche $V \setminus C$ lo è** (infatti $V-C \ne \emptyset$ e $t(V-C) = t(C)$) --> un algoritmo che calcola un taglio minimo individua una bipartizione del grafo in due comunità.  
Inoltre per il teo. visto prima se $C$ è cut-community con $|C| \ge 2$ e $|V - C| \ge 2$ allora $<C, V - C>$ è una bipartizione del grafo in due weak web-communities.

Ma questo significa che si può calcolare una partizione di un grafo $G$ in due cut-communities in tempo polinomiale in $|G|$, **ma non possiamo garantire che le due comunità trovate siano entrambe di dimensione maggiore di 1**.  

In effetti calcolare una partizione di un grafo in due web-communities è un compito molto più complesso: *mentre esiste sempre una partizione di un grafo in due cut-communities (dato che un taglio minimo esiste sempre)* **non è detto invece che sia sempre possibile partizionare un grafo in due web-communities**.  
In realtà, **decidere se un grafo è partizionabile in due (strong) web-communities è un problema NP-completo!**

##### Problema Strong Web-communities Partitioning (SWCP)
Il problema SWCP è il seguente: dato un grafo $G = (V, E)$, dobbiamo decidere se esiste un sottinsieme (proprio e non vuoto) $C \subset V$ tale che sia $C$ che $V \setminus C$ siano **strong web-communities**.

> Prima di dimostrare il teorema di NP-completezza di SWCP, ci è utile il seguente lemma.
>
> **Lemma**:
> 
> Se $G = (V, E)$ è un grafo partizionabile in due strong web-communities ed esistono $x, y, z \in V$ tali che $N(x) = \{y, z\}$  
> $\Rightarrow$ per ogni $C \subset V$ tale che $C$ e $V \setminus C$ siano strong web-communities, $x, y, z \in C$ **oppure** $x, y, z \in V \setminus C$.
>
> **Dim**:
>
> Sia $C \subset V$ tale che $C$ e $V \setminus C$ siano strong web-communities. Senza perdita di generalità supponiamo che $x \in C$.
>
> - se fosse $y \in V \setminus C$ e $z \in V \setminus C$ allora $|N(x) \cap C| = 0 < |N(x) \cap V \setminus C| = 2$ --> assurdo perché $C$ è strong web-community.
> - se fosse $y \in C$ e $z \in V \setminus C$ allora $|N(x) \cap C| = 1 = |N(x) \cap V \setminus C|$ --> assurdo perché $C$ è strong web-community (e analogamente se $y \in V \setminus C$ e $z \in C$).
>
> Quindi per forza $y, z \in C$.
>
> $\blacksquare$


> **Teorema** 
> 
> SWCP è NP-completo.
>
> **Dimostrazione**:
>
> SWCP chiede informalmente: dato un grafo $G = (V, E)$, possiamo dividerne i nodi in due insiemi non vuoti $C$ e $V \setminus C$ in modo che ogni nodo abbia **più vicini dalla propria parte che dall'altra?**  
> La disuguaglianza deve essere stretta perché stiamo parlando di strong web-communities.
>
> Per dimostrare che SWCP è NP-completo bisogna mostrare due cose:
> 1. **SWCP è in NP**: riusciamo a verificare una soluzione proposta in tempo polinomiale?
> 2. **SWCP è NP-hard**: possiamo ridurre un problema NP-completo noto a SWCP?
>
> **SWCP è in NP**: un certificato per SWCP è un sottinsieme $C \subset V$ e per verificare che $C$ e $V \setminus C$ siano strong web-communities basta controllare anzitutto che $C$ e $V \setminus C$ siano non vuoti, e poi per ogni nodo contare i vicini nella propria partizione e quelli nell'altra, verificando che il primo numero sia maggiore del secondo. Se questo vale per ogni nodo, la soluzione è corretta. 
>
> Questo controllo richiede tempo polinomiale, precisamente $O(|V| + |E|)$, quindi SWCP è in NP.
>
> **SWCP è NP-hard**: si sceglie 3-SAT come problema NP-completo noto da ridurre a SWCP.
>
> Sia $X = \{x_1, x_2, ..., x_n\}$ l'insieme delle variabili booleane dell'istanza di 3-SAT. 
> Un'istanza di 3-SAT è una formula del tipo
> $$f = c_1 \wedge c_2 \wedge ... \wedge c_m$$
> dove ogni clausola $c_i$ è OR di tre letterali, es. $c_j = l_{j1} \vee l_{j2} \vee l_{j3}$, dove un letterale è una variabile booleana o la sua negazione.  
> Sappiamo che soddisfare $f$ significa **rendere vero almeno un letterale in ogni clausola**.
>
> Per la riduzione si costruisce un grafo $G_f$ tale che:
> $$f \text{ è soddisfacibile } \iff G_f \text{ è partizionabile in due strong web-communities}$$
> Questo grafo contiene:
> - due nodi speciali $T$ ed $F$;
> - un piccolo sottografo (**gadget**), per ogni variabile $x_i$
> - un altro **gadget** per ogni clausola $c_j$.
>
> La comunità di $T$ rappresenterà il vero, mentre quella di $F$ il falso. Dovremo dimostrare che $T$ ed $F$ sono obbligati a stare separati in due partizioni (comunità) diverse.
>
> Abbiamo dimostrato con il Lemma iniziale che, assumendo esista bipartizione di strong web-communities nel grafo, **ciascun nodo con esattamente due vicini deve stare nella stessa comunità dei suoi vicini**.  
> Dal Lemma segue direttamente che anche tutti i nodi di **grado esattamente 1** (nodi foglia) devono stare nella stessa comunità del loro unico vicino (altrimenti avrebbe zero vicini interni e uno esterno...).
> 
> ---
> **Gadget per le variabili**:
>
> Per ogni variabile introduciamo i nodi $x_i, w_i, y_i, z_i, t_i, f_i$ dove il nodo $x_i$ rappresenta il letterale positivo mentre il nodo $w_i$ rappresenta il letterale negativo $\lnot x_i$.  
> Il gadget per la variabile $x_i$ è il seguente:
>
> <img src="imgs/5_5.png" width="200" />
>
> dove a $x_i$ e $w_i$ sono aggiunte anche le foglie (nodi senza nome) e aggiungeremo collegamenti alle clausole con il prossimo gadget.
>
> Supponiamo che T ed F siano separati: $T \in C$ e $F \in V \setminus C$.  
> Allora poiché $t_i$ e $f_i$ hanno grado 2, il lemma impone che:
> $$T, t_i, y_i \in C \quad \text{e} \quad F, f_i, z_i \in V \setminus C$$
> 
> Guardando ora $y_i$, questo ha tre vicini: $t_i, x_i, w_i$.  
> $t_i$ già sta dalla sua parte per il Lemma. Allora per avere maggioranza stretta tra i vicini è necessario che **almeno uno tra $x_i$ e $w_i$ stia con $T$**. 
>
> Lo stesso identico ragionamento si può fare per $z_i$: $f_i$ già sta dalla sua parte per il Lemma --> per avere maggioranza stretta tra i vicini è necessario che **almeno uno tra $x_i$ e $w_i$ stia con $F$**.
>
> Poiché i nodi sono soltanto due, è quindi **necessario che $x_i$ e $w_i$ debbano stare in comunità diverse**. 
> 
> Qui troviamo il "parallelismo" con la logica booleana: le due possibilità codificano esattamente i due valori della variabile:
> - se $x_i$ sta con $T$ in $C$ allora la rispettiva assegnazione di verità è $x_i = \text{true}$ --> $w_i$ sta con $F$ in $V \setminus C$ e quindi $w_i = \text{false}$ (e torna perché $w_i$ rappresenta $\lnot x_i$)
> - se $w_i$ sta con $T$ in $C$ allora la rispettiva assegnazione di verità è $x_i = \text{false}$ ($w_i = \lnot x_i = \text{true}$)
> ---
> **Gadget per le clausole**:
>
> Per ogni clausola $c_j$ si introducono 4 nodi: $c_j, l_{j1}, l_{j2}, l_{j3}$ dove $l_{jk}$ rappresenta il k-esimo letterale della clausola $c_j$ (NB sono nodi ausiliari, non rappresentano il valore effettivo del letterale, a quello pensano $x_i$ e $w_i$).  
> Ogni nodo $l_{jk}$ ha esattamente due vicini: $T$ e $c_j$. Inoltre si collega $c_j$ ai tre nodi che rappresentano i tre letterali della clausola.
>

> Per il Lemma, dato che tutti i letterali hanno grado 2 e sono collegati a $T$ (che per ipotesi è in $C$) --> i nodi $l_{j1}, l_{j2}, l_{j3}, c_j$ devono stare tutti in $C$.
>
> Avendo collegato $c_j$ poi ai nodi che rappresentano i letterali della clausola, $c_j$ ha in totale 6 vicini:
> - 3 ausiliari $l_{j1}, l_{j2}, l_{j3}$ che stanno in $C$
> - 3 nodi letterali la cui posizione dipende dall'assegnazione di verità
>
> Ma se tutti i 3 nodi letterali fossero dalla parte di $F$ --> $c_j$ avrebbe 3 vicini in $C$ e 3 in $V \setminus C$ --> non sarebbe strong web-community --> è necessario che almeno un nodo letterale stia con $T$. Dunque:
> $$c_j \text{ rispetta la condizione strong} \iff \text{almeno un letterale della clausola è vero}$$
> che è proprio la condizione di soddisfacibilità della clausola $c_j$.
>
> ---
> **A cosa servono i nodi senza nome?**  
> 
> Si tratta di nodi **foglia**, ciascuno collegato al proprio nodo letterale.  
> Quello che facciamo è quanto segue: **se il letterale $x_i$ compare in $k$ clausole --> aggiungiamo a $x_i$ un totale di $k+1$ foglie.**  
> Facciamo lo stesso anche per $w_i$ usando il numero di clausole che contengono $\lnot x_i$.
>
> Sappiamo che ogni foglia per quanto detto prima è obbligata a stare nella stessa comunità del proprio vicino. Lo scopo di questi nodi è sostanzialmente permettere ad un nodo letterale di stare con F (i.e. avere assegnazione di verità false) anche se è presente in molte clausole (i.e. è collegato a molti nodi clausola, che sappiamo stare per definizione di $G_f$ con T).
>
> Supponiamo ad es. che $x_i$ sia dalla parte di F. Allora i suoi vicini sono:
> - $k$ nodi clausola che stanno con T
> - $y_i$, che sta con T
> - $z_i$, che sta con F
> - $k+1$ foglie, che stanno con F perché $x_i$ sta con F
>
> Allora il numero di vicini interni è $1 + (k+1) = k+2$ mentre il numero di vicini esterni è $k + 1$ --> $k + 2 > k + 1$ quindi le foglie garantiscono che effettivamente $x_i$ possa starsene in F.

> **Finora abbiamo solo supposto che T ed F debbano trovarsi in comunità diverse, ora dimostriamolo**. 
>
> Supponiamo per assurdo che entrambi siano nella stessa comunità $C$.  
> Allora per il Lemma sui nodi di grado 2:
> - tutti i nodi $t_i, f_i, y_i, z_i$ devono stare in $C$ 
> - tutti i nodi ausiliari delle clausole e tutti i nodi $c_j$ devono stare in $C$ (sono collegati a $T$, e $T$ sta in $C$).
>
> Rimangono da sistemare solo i nodi letterali $x_i$ e $w_i$ e le loro foglie.  
> Prendiamo un generico $x_i$ e supponiamo che stia in $k$ clausole. Allora ha già $k+2$ vicini in $C$ (i $k$ nodi clausola + $y_i$ + $z_i$). Se provassimo a mettere $x_i$ in $V \setminus C$ allora potrebbe al massimo avere le sue $k+1$ foglie come vicini interni, mentre avrebbe $k+2$ vicini esterni --> non sarebbe strong web-community --> $x_i$ deve stare in $C$.  
> Lo stesso ragionamento vale per ogni $w_i$.
>
> Da ciò segue che **tutti i nodi del grafo devono stare in $C$** --> ma ciò porterebbe a una soluzione non ammissibile per SWCP perché $V \setminus C$ deve essere non vuoto per definizione di partizione in due comunità.  
> Quindi **T ed F devono stare in comunità diverse**.
>
> ---
> Ora **dobbiamo dimostrare che 3-SAT è riducibile polinomialmente a SWCP**, i.e. devono valere le due condizioni:
> - ogni formula $f$ può essere trasformata in un grafo $G_f$ in tempo polinomiale
> - si conserva la risposta al problema: $f$ è soddisfacibile $\iff G_f$ è partizionabile in due strong web-communities
>
> La prima condizione è chiaramente soddisfatta: il grafo $G_f$ è costruito in tempo polinomiale in $n + m$ (numero di variabili e clausole). Concentriamoci sulla seconda condizione.
>
> $\Rightarrow$ **Se $f$ è soddisfacibile, allora esiste una partizione di $G_f$ in due strong web-communities.**
>
> Supponiamo $f$ soddisfacibile e prendiamo un'assegnazione che la soddisfa.  
> Allora costruiamo la partizione come segue: 
>
> <img src="imgs/5_8.png" width="500" />
>
> Ora contiamo i vicini interni ed esterni per ogni nodo, per verificare che la partizione sia effettivamente in due strong web-communities:
> - T ed F hanno vicini solo dalla propria comunità (per il Lemma sui nodi di grado 2)
> - $t_i, f_i$ e ausiliari delle clausole hanno vicini solo dalla propria comunità
> - $y_i$ e $z_i$ hanno ciascuno 2 vicini interni e uno solo esterno (perché come visto prima $x_i$ e $w_i$ devono stare in comunità diverse)
> - i nodi senza nome sono foglie e quindi per forza il loro vicino è dalla loro parte
> - nodi clausola $c_j$: hanno tre ausiliari e almeno un letterale dalla loro parte (i.e. dalla parte di T) e al massimo due letterali esterni, quindi apposto
> - nodi letterali $x_i$ e $w_i$: li verifichiamo come fatto prima con il conteggio: 
>   - se il nodo sta con $T$ stiamo appostissimo perché ipotizzando stia in $k$ clausole allora ha come vicini interni $k+1$ nodi foglia, le $k$ clausole true, $y_i$ che è sempre true per il lemma sui nodi di grado 2 --> $k + 1 + k + 1 = 2k + 2$ vicini interni, mentre ha come vicini esterni solo $z_i$
>   - se il nodo sta con $F$ allora ha come vicini interni $k+1$ nodi foglia, $z_i$ (che sta sempre in $F$ per il lemma sui nodi di grado 2) --> $k + 1 + 1 = k + 2$ vicini interni, mentre ha come vicini esterni le $k$ clausole in cui compare e $y_i$ --> $k + 1$ vicini esterni --> $k + 2 > k + 1$ quindi apposto anche in questo caso.
>
> Entrambe le parti sono non vuote perché contengono rispettivamente T ed F --> abbiamo costruito una partizione valida in due strong web-communities.
> 
> ---
> $\Leftarrow$ **Se $G_f$ è partizionabile in due strong web-communities, allora esiste un'assegnazione di verità delle variabili che rende vera ogni clausola.**
>
> Supponiamo che il grafo costruito ammetta due strong web-community.  
> Sappiamo che T ed F devono essere in comunità diverse, sia $C$ la comunità di T. 
>
> Abbiamo visto che per ogni variabile il gadget impone che $x_i$ e $w_i$ siano in comunità diverse. Definiamo:
> $$a(x_i) = \begin{cases} \text{true} & \text{se } x_i \in C \\ \text{false} & \text{se } x_i \in V \setminus C \end{cases}$$
> di modo che un nodo letterale sta in $T$ esattamente quando quel letterale è vero.
>
> Inoltre, per il discorso fatto sui nodi clausola, ogni nodo clausola $c_j$ deve avere almeno un nodo letterale dalla parte di T per rispettare la maggioranza stretta --> ogni clausola ha almeno un letterale vero --> tutte le clausole sono soddisfatte --> $f$ è soddisfacibile.
>
> Abbiamo quindi finalmente dimostrato che $3SAT \le_p SWCP$ e quindi che SWCP è NP-hard. Poiché abbiamo già dimostrato che SWCP è in NP, concludiamo che SWCP è NP-completo.
>
> $\blacksquare$

(ricorda:
- NP-HARD == tutti i problemi in NP si riducono a questo problema in tempo polinomiale
- NP-COMPLETE == il problema è NP-HARD e appartiene a NP, i.e. si può verificare una soluzione in tempo polinomiale
  
Se dimostrassi solo che SWCP è NP-HARD non avrei dimostrato che è NP-COMPLETE, perché non avrei dimostrato che appartiene a NP)
