La presenza di una rete influenza il comportamento degli individui che la popolano, talvolta modificandolo (come visto con Granovetter e l'effetto dei weak ties).  

Molte delle nostre interazioni avvengono a livello locale --> ci interessa il comportamento degli individui con i quali siamo in relazione, piuttosto che quello dell'intera popolazione.  
In particolare **sono proprio le persone che possono influenzare il nostro comportamento quelle con le quali siamo in relazione**, in questa lezione cercheremo di entrare nel merito di questa "influenza di rete".

Abbiamo incontrato il fenomeno della chiusura triadica (i.e. tendenza che si stabiliscano relazioni tra individui che hanno relazioni forti in comune), questo fenomeno è strettamente connesso al concetto di **omofilia** che si esplica in due modi:
- in un senso, *è la tendenza a connetterci con gli individui che ci assomigliano* --> ciò ha senso perché tendenzialmente si sta meglio con chi ci somiglia (*selezione*) e *opportunità* (le persone simili tendono a trovarsi negli stessi contesti, es. mare o montagna)
- nell'altro senso, è la tendenza a *diventare simili* agli individui con i quali siamo in relazione --> ciò è dovuto all'esigenza di ridurre la tensione sociale (non si può sempre litigare se vedere thriller o commedia.. prima o poi uno dei due amici si farà piacere il genere che piace all'altro) ma anche motivazione del tutto *razionale* (es. se tutti i miei amici hanno linux, probabilmente anche io lo userò perché avrò più supporto e risorse disponibili).

Prendiamo proprio l'esempio del SO. Se io cambio a linux, magari anche i miei amici lo faranno, e forse anche i loro amici etc.. si parla di **processo di diffusione**.  

Il *processo di diffusione delle innovazioni* è stato studiato in sociologia da quasi un secolo: Ryan e Gross (1943) hanno osservato il processo di adozione di nuovi sistemi ibridi di mais da parte di un gruppo di agricoltori in Iowa. Sebbene la maggior parte di agricoltori veniva a conoscenza di questi tramite i venditori, la maggior parte di essi li iniziava a usare solo dopo aver osservato che un certo numero di vicini li stava utilizzando.

Per modellare il processo di diffusione in una rete, dobbiamo stabilire le regole in base alle quali un nodo decide di cambiare (comportamento, opinione, prodotto...). Per farlo definiamo anzitutto un modello di **decisioni individuali** (i.e. non c'è coalizzazione tra nodi per prendere collettivamente la stessa decisione) nel quale le scelte dei nodi sono guidate da *motivazioni di puro interesse personale*.

Assumiamo che nella rete si sia stabilizzato un certo stato delle cose $B$ (es. tutti gli individui nella rete usano lo stesso sistema operativo) e che a un certo istante, alcuni individui cambino il loro stato in $A$ (nuovo SO) --> ci chiediamo in quali casi un individuo decide di passare ad $A$, **assumendo che chi è nello stato $A$ non torni mai più in $B$**.

## Formalizzazione del modello per il processo di diffusione
Modelliamo il processo di diffusione, **individuale** e **basato sul vantaggio personale**. 

Per farlo usiamo un *Network Coordination Game*: sia $(u,v)$ un arco della rete. Assumiamo che il beneficio reciproco di adottare $A$ o $B$ sia quello illustrato in tabella:

<img src="imgs/6_1.png" alt="Tabella dei benefici" width="300"/>

i.e. se $u$ e $v$ adottano entrambi $A$ hanno entrambi beneficio $a$, se $B$ allora $b$ e se diversi allora nada, nessun beneficio dalla reciproca relazione.  

Un nodo della rete però ha in genere più di un vicino --> cosa accade quando qualcuno dei vicini di un nodo $u$ è nello stato $A$ e qualcun altro in $B$?  

Semplicemente, detti $n_A$ il numero di vicini di $u$ nello stato $A$ e $n_B$ quelli nello stato $B$, se $u$ resta in $B$ ha beneficio pari a $n_B  b$ mentre se passa ad $A$ ha beneficio pari a $n_A  a$.  
Quindi, per la questione egoistica, $u$ resta in $B$ se $n_B  b > n_A  a$ mentre passa ad $A$ se $n_A  a \ge n_B  b$.  
Si osservi come a parità di beneficio $u$ passi ad $A$ (l'innovazione è preferibile al vecchio stato).

Riformulando, poiché $n_B = |N(u)| - n_A$, **$u$ passa ad $A$ se:** $n_A a \ge (|N(u)| - n_A) b$, ossia se:
$$\frac{n_A}{|N(u)|} a \ge \frac{|N(u)| - n_A}{|N(u)|} b$$
Ossia, detto $p_A = \frac{n_A}{|N(u)|}$ la frazione di vicini di $u$ nello stato $A$, $u$ passa ad $A$ se $p_A a \ge (1-p_A) b$ ossia **$u$ passa ad $A$ se, detta $p_A$ la frazione di vicini di $u$ nello stato $A$, vale che**:
$$p_A \ge \frac{b}{a+b}$$


Chiamiamo $q = \frac{b}{a+b}$ la **soglia di adozione di $A$**: 
- quando $q$ è molto piccolo --> $a$ è più grande di $b$, ergo occorrono pochi vicini nello stato $A$ per indurre un nodo a cambiare stato
- quando $a = b$ --> $q = 1/2$, ossia occorre che almeno metà dei vicini di un nodo siano nello stato $A$ per indurlo a cambiare
- quando $q$ è molto grande --> $b$ è più grande di $a$, ergo occorrono molti vicini nello stato $A$ per indurre un nodo a cambiare stato (è faticoso/rischioso/costoso adottare $A$).

### Configurazioni di equilibrio

Cerchiamo ora di capire se e quali **configurazioni di equilibrio** ha il network coordination game appena introdotto, i.e. configurazioni dove nessun nodo cambia più il proprio stato da $B$ ad $A$ (sapendo che chi entra in $A$ non torna mai più in $B$).

Si osserva che esistono almeno due configurazioni di equilibrio, quelle banali:
- se $A$ non è stato mai introdotto nella rete --> tutti restano in $B$
- se tutti i nodi passano allo stato $A$ 

Del resto una volta che $A$ viene introdotto nella rete, il processo inizia a diffondersi, ci si chiede quindi quando termina il processo, se riesce sempre a raggiungere tutti i nodi oppure se talvolta si blocca in configurazioni di equilibrio intermedie, e in questo caso perché...

Vediamo qualche esempio:

qui lo stato $A$ è inizialmente forzato sui nodi $w$ e $v$. Impostando $a = 3$ e $b = 2$ --> $q = \frac{2}{5}$ i.e. per adottare $A$ un nodo deve avere i $\frac{2}{5}$ dei vicini nello stato $A$.

Allora prima di tutto $r$ e $t$ hanno $\frac{2}{3} \gt \frac{2}{5}$ dei vicini nello stato $A$ --> quindi passano ad $A$, e così a cascata $s$ ed $u$.

Altro esempio:

<img src="imgs/6_3.png" alt="Esempio di rete" width="400"/>

Se $a = 3$ e $b = 2$ --> $q = \frac{2}{5}$ ed è il caso illustrato in figura: se gli iniziatori sono 7 e 8 allora di seguito si adotteranno i nodi nell'esagono ma poi ci si ferma perché i nodi esterni non hanno abbastanza vicini nello stato $A$ per adottarlo.

Se invece avessimo avuto un valore più forte per $A$, come $a=4$ --> $q = \frac{2}{6} = \frac{1}{3}$ e allora si sarebbe propagato il processo fino a **cascata completa**.

Dai due esempi capiamo come non sempre tutti i nodi si adeguano all'innovazione, e che per incentivare una cascata completa possiamo aumentare il beneficio derivante dall'adozione di $A$ (es. abbassando il costo di adozione $a$, aumentando così l'attrattiva!). 

Però un altro elemento fondamentale che influenza il processo di diffusioen **è la specifica scelta di nodi iniziatori**: il loro numero e la loro posizione nella rete. 

Es. se forziamo $A$ su 7,8,12 e 2 --> cascata completa  
se forziamo $A$ su 7,8,2 e 14 no...

---

Studiamo il fenomeno in modo rigoroso, ci serve qualche altra notazione:  
sia $V_0$ l'insieme di nodi **iniziatori** a cui è forzato inizialmente lo stato $A$.

Abbiamo poi visto che una volta introdotto $A$ nella rete esso inizia a diffondersi in una sequenza di passi discreti: al passo $1$ un insieme $V_1$ di vicini di nodi in $V_0$ adotta $A$, al passo $2$ un insieme $V_2$ di vicini di nodi in $V_0 \cup V_1$ adotta $A$, ..., al passo $t$ un insieme $V_t$ di vicini di nodi in $V_0 \cup V_1 \cup ... \cup V_{t-1}$ adotta $A$...  
Ossia indichiamo con $V_i$** l'insieme di nodi che adottano $A$ al passo $i$ del processo di diffusione**.

Diremo allora che viene generata una **cascata completa** se ad un certo passo $t$ tutti i nodi hanno adottato $A$, i.e.  
**esiste passo $t \ge 0$ t.c. $\bigcup_{i=0}^{t} V_i = V$**.

Ripensando all'esempio, sembra che l'innovazione abbia difficoltà ad uscire dall'"esagono centrale", che non a caso appare come un gruppo coeso di nodi... una **comunità**!  
Definiamo un **cluster di densità $p$** come un sottinsieme di nodi $V' \subseteq V$ tale che la frazione di vicini che ogni suo nodo ha in $V'$ è almeno $p$:
$$\forall u \in V' : \frac{|N(u) \cap V'|}{|N(u)|} \ge p$$
In particolare, il sottinsieme di nodi dell'esagono ${4, 5, 6, 7, 8, 9, 10}$ è un cluster di densità $p = \frac{2}{3}$ a causa del nodo $6$.

Chiaramente, l'unione di due cluster di densità $p$ è ancora un cluster di densità $p$ --> da dimostrare come esercizio se decidi di portare questo argomento all'esame.

Ultima notazione prima del teorema: dato $G = (V, E)$ e un insieme di nodi $V' \subseteq V$, indichiamo con $G - V'$ il grafo ottenuto da $G$ rimuovendo i nodi in $V'$ e tutti gli archi incidenti a questi nodi.

> **Teorema**
>
> Sia $G = (V, E)$ un grafo e siano $V_0 \subseteq V$ l'insieme di iniziatori e $q$ la soglia di adozione di $A$ --> $V_0$ **non genera cascata completa** $\iff$ esiste un cluster di densità $\gt 1-q$ in $G - V_0$.
>
> (il senso intuitivo è che se esiste un cluster di densità $\gt 1-q$ allora i nodi in quel cluster non hanno abbastanza vicini nello stato $A$ per adottarlo, quindi il processo si ferma lì.. vedi l'esempio di prima con il triangolo in alto a sx... se la soglia di adozione fosse $\frac{2}{3}$ e gli iniziatori fuori dal triangolo, allora poiché il triangolo è un cluster di densità $\frac{2}{3}$ (e $\frac{2}{3} > 1 - \frac{2}{3}$) i nodi nel triangolo non potranno mai adottare $A$!).
>
> **Dimostrazione**
>
> $\Rightarrow$ **Se $V_0$ non genera cascata completa** --> esistono dei nodi che non adottano $A$. Avendo definito $V_i$ l'insieme di nodi che adottano $A$ al passo $i$, allora sia $t$ il passo t.c. $V_t \neq \emptyset$ e $V_{t+1} = \emptyset$ (i.e. il passo $t+1$ è il primo in cui $A$ non si diffonde più).
>
> Sia $V_A = \bigcup_{i=0}^{t} V_i$, i.e. $V_A$ contiene tutti e soli i nodi che adottano $A$, e poiché esistono dei nodi che non adottano $A$ allora $V-V_A \neq \emptyset$. 
>
> Poiché i nodi $V - V_A$ non adottano $A$ --> $\forall v \in V - V_A$ vale che $\frac{|N(v) \cap V_A|}{|N(v)|} < q$ e quindi, poiché:
> $$\frac{|N(v) \cap V_A|}{|N(v)|} + \frac{|N(v) \cap (V - V_A)|}{|N(v)|} = 1$$
> allora $\frac{|N(v) \cap (V - V_A)|}{|N(v)|} > 1 - q$ ossia $V-V_A$ è cluster di densità maggiore di $1-q$ e $V-V_A \subseteq G - V_0$.
>
> $\Leftarrow$ **Esiste un cluster di densità $\gt 1-q$ in $G - V_0$**, cioè esiste $C \subseteq V - V_0$ t.c. $\forall v \in C : \frac{|N(v) \cap C|}{|N(v)|} > 1 - q$.  
>
> Supponiamo per assurdo che si generi una cascata completa --> i nodi in $C$ adotteranno $A$ prima o poi.  
> Sia quindi $t$ il primo passo t.c. $V_t \cap C \neq \emptyset$, ossia per $i \lt t$ si ha $V_i \cap C = \emptyset$ ed esiste $u \in V_t \cap C$.
>
> Poniamo $V' = \bigcup_{i=0}^{t-1} V_i$ --> $V'$ rappresenta l'insieme di nodi che hanno adottato $A$ prima del passo $t$, e quindi è disgiunto da $C$. Allora:
> 1. poiché $C$ è un cluster di densità $\gt 1-q$ e $u \in C$ allora $\frac{|N(u) \cap C|}{|N(u)|} \gt 1 - q$
> 2. poiché $u \in V_t$ allora $\frac{|N(u) \cap V'|}{|N(u)|} \ge q$ 
>
> Ossia poiché $V' \cap C = \emptyset$ (si usa nella terza uguaglianza) allora:
> $$1 = \frac{|N(u)|}{|N(u)|} \ge \frac{|N(u) \cap (C \cup V')|}{|N(u)|} = \frac{|N(u) \cap C|}{|N(u)|} + \frac{|N(u) \cap V'|}{|N(u)|} \gt 1 - q + q = 1$$
> assurdo!
>
> $\blacksquare$

Il teorema appena dimostrato mette in luce un nuovo aspetto della dicotomia strong ties / weak ties: **le innovazioni si diffondono con relativa facilità nei cluster** (i.e. quando viaggia lungo strong ties, l'innovazione ha impatto significativo sui nodi che raggiunge) e invece **incontrano difficoltà ad uscire dai cluster** (i.e. quando viaggia lungo weak ties, l'innovazione ha impatto debole sui nodi che raggiunge).

Possiamo quindi concludere che, mentre l'esperimento di Granovetter mostrava come la forza dei weak ties come fonte di vantaggi informativi, lo studio dei processi di diffusione ne evidenzia la debolezza in quanto ostacolo alla diffusione di innovazioni.

### Marketing virale e capacità di cascata
Il teo. appena dimostrato mostra che i cluster ostacolano la diffusione virale di un'innovazione. Ci chiediamo quindi ora come superare lo stato di stallo nel quale, a causa della presenza di cluster troppo densi, uno stato $A$ smette di diffondersi. 

Possiamo scegliere gli iniziatori in posizioni tali che ciascun cluster contenga almeno un iniziatore, oppure come visto aumentare l'appetibilità di $A$ diminuendo magari il suo prezzo di vendita ($a$)..  
Tuttavia dal punto di vista del venditore, abbassare il prezzo non è una strategia che adotterebbe con piacere, ci domandiamo quindi **quanto alto può esser mantenuto il prezzo (i.e. quanto bassa può essere l'attrattiva) di $A$ perché la diffusione sia comunque virale?**

Ci stiamo cioè chiedendo, **quanto alta può esser tenuta la soglia di adozione perché si generi una cascata completa?** 

Questo dipende anzitutto dalla *struttura della rete*: alcune strutture ostacolano maggiormente di altre la generazione di cascate complete...

**Problema**: *dato un grafo $G = (V, E)$, qual è la soglia di adozione massima $q_{max}$ in $G$ affinché un piccolo insieme $V_0$ di iniziatori di un nuovo stato $A$ generi una cascata completa?*

Il valore $q_{max}$ è detto **capacità di cascata di G** (massimo equivale a minimo $a$, i.e. minimo prezzo speso dal rivenditore per incentivare l'adozione di $A$).  
Tuttavia nel descriverlo così siamo un po' imprecisi... che significa che $V_0$ deve essere piccolo? Quanto piccolo? Chiaramente scegliere $V_0 = V$ genera cascata completa per qualunque $q$ ma non ha molto senso...  
chiedendo $V_0$ piccolo indendiamo che vogliamo $|V_0| \ll |V|$.

Ipotizziamo in particolare di lavorare su un grafo **infinito** $G$ (i.e. $|V| = \infty$) dove $G$ è **grafo regolare** (i.e. tutti i nodi hanno la stessa struttura del vicinato). Questa seconda condizione è necessaria per poter lavorare con il grafo tramite calcolatore.  
Allora l'ipotesi "$V_0$ piccolo" è che $|V_0|$ deve essere insieme **finito**.

Cioè: dato un grafo regolare infinito, qual è la soglia di adozione *massima* $q_{max}$ affinché esista un insieme finito di iniziatori che generi una cascata completa? (NB. nel caso di grafo infinito, si parla più che di cascata completa di **processo di diffusione infinito**)

Vediamo alcuni esempi, partendo da una catena infinita e poi affrontando il caso di una griglia infinita:

<img src="imgs/6_4.png" alt="Catena infinita" width="400"/>

<img src="imgs/6_5.png" alt="Griglia infinita" width="400"/>

<img src="imgs/6_6.png" alt="Griglia infinita" width="400"/>

Dagli esempi della catena e della griglia possiamo concludere quanto segue: la soglia di adozione massima è più bassa nella griglia (che ha topologia più ricca rispetto alla catena); che in entrambi i casi essa non supera $\frac{1}{2}$ ossia che affinché si generi una cascata completa, occorre che $A$ sia almeno appetibile quanto $B$, se lo fosse molto meno infatti $q$ si avvicinerebbe a $1$ ($q = \frac{b}{a+b} \le \frac{1}{2}$).

Poiché **la soglia di adozione massima è caratteristica della topologia della rete** la domanda sorge spontanea: **esistono topologie nelle quali la soglia di adozione massima è maggiore di $\frac{1}{2}$?**, ossia nelle quali innovazioni di qualità mediocre soppiantino uno status quo di qualità migliore?  
Fortunatamente **no**, come dimostra il seguente teorema.

Poiché come detto la soglia di adozione max è caratteristica della rete, indichiamo come $q_G$ la soglia di adozione max di un grafo $G$.

> **Teorema**
>
> Per ogni grafo infinito $G = (V, E)$ in cui **i nodi hanno grado finito**, vale che $q_G \le \frac{1}{2}$
>
> **Dimostrazione**
>
> Supponiamo per assurdo che esista insieme finito di iniziatori $V_0$ che, con soglia di adozione $q \gt \frac{1}{2}$, generi una cascata completa nel grafo $G$.  
> Sia al solito $V_t$ l'insieme di nodi che adottano $A$ al passo $t$ e sia, per ogni $t \ge 0$, $S_t = \bigcup_{i=0}^{t} V_i$ l'insieme di nodi che hanno adottato $A$ entro il passo $t$.
>
> Definiamo ora **l'interfaccia al passo $t$** come l'insieme $I_t$ degli archi che, al passo $t$, hanno un estremo in $S_t$ e l'altro in $V - S_t$:
> $$I_t = \{(u, v) \in E : u \in S_t, v \in V - S_t\}$$
> e **dimostriamo che, per ogni $t \ge 0$, se $I_t \neq I_{t+1}$ allora |I_t| > |I_{t+1}|
>
> Se $I_t \neq I_{t+1}$ allora esiste almeno un nodo $v$ che adotta $A$ al passo $t+1$, ossia $V_{t+1} \neq \emptyset$, e perché $v$ appartenga a $V_{t+1}$ deve esistere almeno un suo vicino $u \in N(v)$ che sia già nello stato $A$ al passo $t$, ossia $u \in S_t$. 
>
> Ma allora, per ogni nodo $v \in V_{t+1}$:
> - gli archi incidenti su $v$ il cui altro estremo è in $S_t$ sono in $I_t$ ma non in $I_{t+1}$ (perché $v$ adotta $A$ al passo $t+1$, quindi l'arco non è più parte dell'interfaccia, si trova completamente "immerso" in $S_{t+1}$)
> - gli archi incidenti su $v$ il cui altro estremo non è in $S_{t+1}$ sono in $I_{t+1}$ ma non in $I_t$ (prima erano "immersi" in $V - S_t$, ora sono parte dell'interfaccia)
>
> <img src="imgs/6_7.png" alt="Interfaccia" width="400"/>
>
> ossia:
> $$
> I_{t+1} = I_t - \left[\bigcup_{v \in V_{t+1}} \{(u, v) \in E : u \in S_t\}\right] \cup \left[\bigcup_{v \in V_{t+1}} \{(z,v) \in E : z \in V - S_{t+1}\}\right]
> $$
> Ma poiché se $v \ne z$ con $v, z \notin S_t$ allora 
> $${(u,v) \in E : u \in S_t} \cap {(u,z) \in E : u \in S_t} = \emptyset$$
>  e poiché se $v \ne z$ con $v, z \in V_{t+1}$ allora:
> $${(u,v) \in E : u \in V - S_t} \cap {(u,z) \in E : u \in V - S_{t+1}} = \emptyset$$
> (ossia due nodi diversi sono ovviamente adiacenti tramite due archi diversi a uno stesso nodo $u$)  
> Ma allora:
> $$|\bigcup_{v \in V_{t+1}} \{(u, v) \in E : u \in S_t\}| = \sum_{v \in V_{t+1}} |\{(u, v) \in E : u \in S_t\}|$$
> e 
> $$|\bigcup_{v \in V_{t+1}} \{(u,v) \in E : u \in V - S_{t+1}\}| = \sum_{v \in V_{t+1}} |\{(u,v) \in E : u \in V - S_{t+1}\}|$$
> ossia
> $$|I_{t+1}| = |I_t| - \sum_{v \in V_{t+1}} |\{(u, v) \in E : u \in S_t\}| + \sum_{v \in V_{t+1}} |\{(u,v) \in E : u \in V - S_{t+1}\}|$$
> Ma:
> $$|\{(u, v) \in E : u \in S_t\}| = |N(v) \cap S_t|$$
> e 
> $$|\{(u,v) \in E : u \in V - S_{t+1}\}| = |N(v) - S_{t+1}|$$
> ossia:
> $$|I_{t+1}| = |I_t| - \sum_{v \in V_{t+1}} |N(v) \cap S_t| + \sum_{v \in V_{t+1}} |N(v) - S_{t+1}|$$
> a questo punto:
> 
> <img src="imgs/6_8.png" alt="Interfaccia" width="400"/>
>
> ma allora 
> $$|I_{t+1}| = |I_t| - \sum_{v \in V_{t+1}} |N(v) \cap S_t| + \sum_{v \in V_{t+1}} |N(v) - S_{t+1}| = |I_t| - \sum_{v \in V_{t+1}} |N(v) \cap S_t| - |N(v) - S_{t+1}| < |I_t|$$
> dove il $<$ deriva dal fatto che $V_{t+1} \neq \emptyset$ poiché per hp. il processo di diffusione è infinito
>
> Abbiamo dimostrato che **per ogni $t \ge 0$ se $I_t \neq I_{t+1}$ allora $|I_t| > |I_{t+1}|$**.  
> Ma poiché $V_0$ è insieme finito di nodi e i nodi di $G$ hanno grado finito --> l'interfaccia iniziale ha dimensione finita $I_0 \lt k$ per qualche $k \in \mathbb{N}$.  
> Ma allora l'eventualità $|I_t| \ne |I_{t+1}|$ non può verificarsi per più di $k$ passi (al più $k$ si verifica se ad ogni passo si perde esattamente un solo arco dell'interfaccia) (perché ogni volta che $I_t \ne I_{t+1}$ allora $|I_t| > |I_{t+1}|$ e la dimensione dell'interfaccia non può essere negativa né tanto meno 0 in un grafo infinito)  
> --> esiste $T$ tale che per ogni $t \ge T$ vale che $I_t = I_{t+1}$, ossia il processo si ferma. Ma in un grafo infinito una cascata completa può verificarsi solo in seguito ad un processo di diffusione infinito --> $V_0$ non genera cascata completa, assurdo!
>
> $\blacksquare$

#### Nodi eterogenei
Il modello di diffusione considerato finora è un modello uniforme, dove tutti i nodi associano lo stesso beneficio reciproco nell'adottare $A$ o $B$ (rispettivamente beneficio $a$ e $b$).  
Ma questo modello è poco realistico: ciascun individuo nella rete ha un proprio beneficio nell'adottare $A$ o $B$ che può dipendere, ad es., dalle sue capacità di adattarsi al nuovo stato.

Allora per ogni nodo $u$ nella rete definiamo $a_u$ e $b_u$ come i benefici reciproci che $u$ ottiene nell'adottare rispettivamente $A$ o $B$.à

<img src="imgs/6_9.png" alt="Nodi eterogenei" width="300"/>

Ma allora un nodo $v \in V$ nello stato $B$ passerà ad $A$ sulla base del valore $q_v = \frac{b_v}{a_v + b_v}$ per gli stessi ragionamenti fatti all'inizio della lezione.

Nell'esempio ogni nodo ha accanto la propria soglia di adozione. Si osserva che il nodo $1$ è in posizione centrale, ma non riuscirebbe a portare i nodi del pentagono ad $A$ se non fosse per la soglia bassissima di $3$ pari a $0.1$...  

<img src="imgs/6_10.png" alt="Nodi eterogenei" width="300"/>

Quindi per nodi eterogenei **non è sufficiente scegliere gli iniziatori in base alla loro centralità nella rete**, ma occorre anche considerare **la loro possibilità di avere accesso a nodi facilmente influenzabili**.

Nel caso di nodi eterogenei, la struttura che impedisce la generazione di una cascata completa è il **blocking cluster**: un sottinsieme di nodi $V' \subseteq V$ è detto **blocking cluster** se $\forall v \in V'$ vale che:
$$\frac{|N(v) \cap V'|}{|N(v)|} > 1 - q_v$$
Si può dimostrare il teorema per cui, dato $G = (V, E)$ un grafo, nel modello a nodi eterogenei l'insieme di iniziatori $V_0 \subseteq V$ **non genera cascata completa** $\iff$ esiste un blocking cluster in $G - V_0$.

#### Azione collettiva
Vogliamo mostrare ora come modellare tramite processi di diffusione situazioni nelle quali è richiesto che un'azione abbia luogo *collettivamente*.  

Supponiamo che si voglia organizzare una protesta contro un regime dittatoriale e sanguinario, e che ciascun individuo (ragionevolmente) decida di aderire alla protesta solo se sa con certezza che un certo numero (elevato) di persone parteciperà (altrimenti rischia la vita per nulla..)

Poiché l'ambientazione è dittatoriale, possiamo ben pensare che la libertà di stampa sia ostacolata e le comunicazioni più in generale rese difficili... **ma perché i regimi totalitari sono così interessati ad ostacolare le comunicazioni?**

Modelliamo la situazione come segue: ogni nodo $v$ sceglie una "soglia di confidenza" $k_v$ e aderirà alla protesta solo se almeno $k_v$ persone aderiranno, ossia oltre a lui devono aderire almeno $k_v - 1$ individui.

Poiché però le comunicazioni circolano con difficoltà nella rete, le uniche info che $v$ può ottenere sono circa l'adesione o meno alla protesta da parte dei suoi vicini e sapere da ogni vicino $u$ anche qual è *la sua soglia di adesione $k_u$*.  
Tuttavia $v$ non può sapere direttamente se $u$ aderirà alla protesta, in quanto sono informazioni pericolose... sa solo le soglie di adesione dei vicini e, in base a quelle e al suo vicinato, può provare a dedurre cosa faranno i vicini.

Vediamo qualche esempio:

<img src="imgs/6_11.png" alt="Esempio azione collettiva" width="200"/>

Qui $x$ non aderisce di certo perché non ha abbastanza vicini che possano anche solo potenzialmente aderire.  

$v$ ha bisogno di due vicini oltre lui che aderiscano... ma vedendo che $x$ vuole almeno 3 vicini perché aderisca a sua volta e poiché $v$ vede di $x$ solo i vicini che hanno in comune ($u$) --> non può sapere se $x$ aderirà o meno --> neanche $v$ aderisce.

Ad $u$ sarebbe sufficiente che uno solo dei vicini aderisse... per lo stesso ragionamento di $v$ però $u$ non può sapere se $x$ aderirà o meno, e inoltre anche di $v$ non ha certezza in merito ad altri suoi vicini complementari che possano portarlo ad aderire... quindi anche $u$ non aderisce.

<img src="imgs/6_12.png" alt="Esempio azione collettiva" width="200"/>

Qui $u$ vede che $k_v = k_x = 3$ e che quindi loro tre potrebbero aderire, ma $u$ non vede $y$ e quindi non sa se $v$ ha o meno altri vicini oltre se stesso, non sa se $v$ può dedurre che almeno due suoi vicini aderiranno --> poiché $u$ ha bisogno di certezze, non aderisce!

Quindi nessuno aderisce alla protesta, nonostante qualora avessero avuto accesso a info complete, la protesta avrebbe avuto luogo!

<img src="imgs/6_13.png" alt="Esempio azione collettiva" width="200"/>

Qui $u, v, x$ si vedono l'un l'altro. Così $u$ sa che $v$ ed $x$ per partecipare hanno bisogno che altri due partecipino, ma sa anche che $v$ ed $x sanno le stesse cose che sa lui --> partecipano tutti e tre senza dover conoscere altro dalla rete, i.e. indipendentemente da $y$!  
(un nodo sa se un altro nodo ha come vicino un proprio vicino)

Quindi, in assenza di comunicazioni adeguate che abbiano luogo nella rete, l'azione collettiva si verifica difficilmente... ecco perché i regimi totalitari tendono a favorire **l'ignoranza pluralistica** che permette di concludere erroneamente che pochi individui abbiano una certa opinione, come avveniva nell'esempio in blu.

L'esempio in rosso invece permette di osservare l'importanza di disporre di una *base di conoscenza comune*: i giornali, le pubblicità etc.. non solo diffondono informazioni, ma permettono anche che tutti siano coscienti del fatto che un certo messaggio è conosciuto da tutti, innescando il meccanismo del "io so che tu sai che io so"...

#### Diffusione in presenza di compatibilità
Finora abbiamo considerato i due stati $A$ e $B$ come mutualmente esclusivi: un nodo non può trovarsi simultaneamente in entrambi.  
Tuttavia non è infrequente che due stati possano coesistere, ad es. se installo un nuovo SO sul mio computer, posso continuare ad usare quello vecchio tramite dual boot...

Allora d'ora in poi studieremo i processi di diffusione in presenza di compatibilità, per cui **un nodo può trovarsi in $A$, in $B$ o in entrambi**, nel modello **nodi omogenei** per cui i benefici (e, vedremo, gli eventuali costi) di trovarsi in un certo stato sono gli stessi per tutti i nodi.

Adottare lo stato misto porta a beneficiare di entrambi gli stati singoli, ma perché allora non adottarlo sempre? *Perché questo comporta un costo*, ad es. per il dual boot mi tocca partizionare parte del disco...  
I benefici d'altra parte sono *proporzionali al numero di vicini con i quali possiamo comunicare*, mentre il costo lo paghiamo una sola volta.

Sia $(u,v)$ un arco della rete. Il beneficio reciproco di adottare $A$, $B$ o $AB$ è illustrato in tabella:

<img src="imgs/6_14.png" alt="Tabella benefici" width="300"/>

mentre per ogni nodo $u$ il costo di essere nello stato $AB$ è $c$.  

In definitiva, se $V_A$, $V_B$ e $V_{AB}$ sono rispettivamente gli insiemi di nodi nello stato $A$, $B$ e $AB$, allora il beneficio di un nodo $u$ nello stato $AB$ è:
$$\sum_{v \in N(u) \cap V_A} a + \sum_{v \in N(u) \cap V_B} b + \sum_{v \in N(u) \cap V_{AB}} \max(a,b) - c$$
mentre per un nodo nello stato $A$ o $B$ è:
$$\sum_{v \in N(u) \cap (V_A \cup V_{AB})} a \quad \text{e} \quad \sum_{v \in N(u) \cap (V_B \cup V_{AB})} b$$
Ci proponiamo ora di studiare la capacità di cascata di una rete in presenza di compatibilità, e lo faremo solo:
- analizzando un esempio nel caso particolare in cui $G$ è catena infinita
- descrivendo i risultati di uno studio qualitativo nel caso bidimensionale ($G$ griglia infinita)
- dimostrando che, nel caso unidimensionale, l'andamento del processo di diffusione conferma quanto osservato dall'analisi qualitativa

Indichiamo con $p_A(u)$, $p_B(u)$ e $p_{AB}(u)$ il beneficio del nodo $u$ nell'adottare rispettivamente $A$, $B$ o $AB$.

<img src="imgs/6_15.png" alt="Tabella benefici" width="400"/>

Ricordiamo che nel caso di esclusività tra $A$ e $B$ la soglia di adozione era $q = \frac{b}{a+b}$ e rappresentava la frazione minima di vicini di un nodo che doveva essere nello stato $A$ affinché il nodo adottasse $A$.  
A partire da questa definizione, dividendo numeratore e denominatore per $b$, otteniamo $q = \frac{1}{\frac{a}{b} + 1}$ che ben evidenzia il rapporto tra il valore di $A$ e quello di $B$ nella decisione di un nodo di adottare $A$ o meno (più $a$ è grande rispetto a $b$, più bassa è la soglia di adozione $q$).

Abbiamo dimostrato nel caso di esclusività che **non si può generare cascata completa in un grafo infinito con nodi di grado finito se $q \gt \frac{1}{2}$**, ossia se $a \lt b$ (il beneficio di $A$ non è almeno pari a quello di $B$).  
Ma **cosa possiamo dire nel caso di compatibilità tra $A$ e $B$?**

Osserviamo anzitutto che l'operazione di dividere per $b$ fatta appena adesso ha permesso di ridursi a studiare i processi di diffusione con solo parametro $\frac{a}{b}$ invece che sui due parametri $a$ e $b$, e ciò è equivalente a fissare $b = 1$ e studiare i processi di diffusione in funzione di $a$.

Analogamente, nel caso di compatibilità, possiamo fissare $b = 1$ e studiare i processi di diffusione in funzione di $a$ e $c$.  
Uno studio qualitativo del 2007 di Kleinberg ha evidenziato uno strano comportamento: *lo stato $A$ si impone quando $a$ è molto più grande rispetto a $c$ (ovvio..)*, e al contrario *$A$ fa fatica a imporsi quando $c$ è molto più grande rispetto ad $a$ (ragionevole..)*, ma anche che **$A$ fa fatica ad imporsi anche quando $c$ non è né troppo grande né troppo piccolo rispetto ad $a$!**
