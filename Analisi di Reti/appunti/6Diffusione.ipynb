La presenza di una rete influenza il comportamento degli individui che la popolano, talvolta modificandolo (come visto con Granovetter e l'effetto dei weak ties).  

Molte delle nostre interazioni avvengono a livello locale --> ci interessa il comportamento degli individui con i quali siamo in relazione, piuttosto che quello dell'intera popolazione.  
In particolare **sono proprio le persone che possono influenzare il nostro comportamento quelle con le quali siamo in relazione**, in questa lezione cercheremo di entrare nel merito di questa "influenza di rete".

Abbiamo incontrato il fenomeno della chiusura triadica (i.e. tendenza che si stabiliscano relazioni tra individui che hanno relazioni forti in comune), questo fenomeno è strettamente connesso al concetto di **omofilia** che si esplica in due modi:
- in un senso, *è la tendenza a connetterci con gli individui che ci assomigliano* --> ciò ha senso perché tendenzialmente si sta meglio con chi ci somiglia (*selezione*) e *opportunità* (le persone simili tendono a trovarsi negli stessi contesti, es. mare o montagna)
- nell'altro senso, è la tendenza a *diventare simili* agli individui con i quali siamo in relazione --> ciò è dovuto all'esigenza di ridurre la tensione sociale (non si può sempre litigare se vedere thriller o commedia.. prima o poi uno dei due amici si farà piacere il genere che piace all'altro) ma anche motivazione del tutto *razionale* (es. se tutti i miei amici hanno linux, probabilmente anche io lo userò perché avrò più supporto e risorse disponibili).

Prendiamo proprio l'esempio del SO. Se io cambio a linux, magari anche i miei amici lo faranno, e forse anche i loro amici etc.. si parla di **processo di diffusione**.  

Il *processo di diffusione delle innovazioni* è stato studiato in sociologia da quasi un secolo: Ryan e Gross (1943) hanno osservato il processo di adozione di nuovi sistemi ibridi di mais da parte di un gruppo di agricoltori in Iowa. Sebbene la maggior parte di agricoltori veniva a conoscenza di questi tramite i venditori, la maggior parte di essi li iniziava a usare solo dopo aver osservato che un certo numero di vicini li stava utilizzando.

Per modellare il processo di diffusione in una rete, dobbiamo stabilire le regole in base alle quali un nodo decide di cambiare (comportamento, opinione, prodotto...). Per farlo definiamo anzitutto un modello di **decisioni individuali** (i.e. non c'è coalizzazione tra nodi per prendere collettivamente la stessa decisione) nel quale le scelte dei nodi sono guidate da *motivazioni di puro interesse personale*.

Assumiamo che nella rete si sia stabilizzato un certo stato delle cose $B$ (es. tutti gli individui nella rete usano lo stesso sistema operativo) e che a un certo istante, alcuni individui cambino il loro stato in $A$ (nuovo SO) --> ci chiediamo in quali casi un individuo decide di passare ad $A$, **assumendo che chi è nello stato $A$ non torni mai più in $B$**.
