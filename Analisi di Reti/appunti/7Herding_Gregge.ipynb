Iniziamo una serie di lezioni, la parte finale del corso, dove studiamo la rete come fonte di informazione.  
Studieremo cioè il contenuto informativo di una rete:
- che può indurre gli individui che la compongono a modificare il proprio comportamento (in questa lezione).
- che deve esser sintetizzato per derivare un'informazione cumulativa che tenga conto, in qualche modo, di pezzi di informazione derivanti dai singoli individui (nei sistemi di voto).
- dal quale deve essere individuato ed estratto quello rilevante ad una data richiesta (nell'ultima serie di lezioni, web search).
  
Parlando del primo caso, abbiamo già visto con i processi di diffusione come l'informazione presente in una rete può indurre gli individui che la compongono a modificare il proprio comportamento, ma ora **il punto di vista è diverso**.  

Mentre nei processi di diffusione un individuo cambiava comportamento 
1. come conseguenza di **comunicazioni esplicite** di una serie di info da parte degli individui con i quali è in relazione (es. semi ibridi di mais)
2. per ottenere una serie di benefici dalla presenza della rete (es. scambiarsi file con lo stesso SO)

ora invece studiamo il fenomeno in virtù del quale un individuo modifica il proprio comportamento semplicemente osservando quello che **globalmente** hanno gli altri individui della rete, in assenza di **comunicazione diretta** --> consideriamo una visione globale e non puntuale della rete, non ci interessa più l'analisi delle singole relazioni personali.

Vediamo un esempio: stiamo visitando da turisti una città sconosciuta, seguendo tutti i consigli di una nostra guida privata. A cena andiamo al bel ristorante consigliato dalla guida, ma notiamo che il locale è vuoto. Davanti al ristorante c'è una bettola di locale, e prima di entrare nel nostro notiamo che tanta gente entra nella bettolaccia...  
tendenzialmente allora una persona deciderà di seguire la folla nella bettola, perché intuitivamente si assume che se tanta gente va da quella parte un motivo, **anche se a noi sconosciuto**, ci deve pur essere!

Questa informazione la presumiamo noi, non ci viene detta esplicitamente ma la deduciamo dal loro comportamento, è una **informazione che ricaviamo dalla rete**.

Un esempio ancora più estremo riguarda sempre Milgram, negli anni '60, che ha piazzato gente in giro per strada e li ha fatti guardare in alto. Negli incroci dove c'era una sola persona a guardare il cielo, quasi nessun passante imitava il comportamento. Agli incroci con 5 persone, solo qualche passante imitava, mentre in quelli con 15 persone allora oltre il 45% dei passanti imitava il comportamento.

Il motivo anche qui è abbastanza chiaro: i passanti inferivano l'esistenza di una motivazione razionale che inducesse le persone a guardare in alto --> doveva esserci qualcosa di interessante, gli sarebbe convenuto anche a loro provare a guardare su!  

Naturalmente perché ciò avvenisse era necessario che ci fosse un numero minimo di persone che guardavano in alto, altrimenti si penserebbe semplicemente che quelle poche persone avessero un disagio mentale...  
Più formalmente stiamo dicendo che **perché il fenomeno di imitazione avesse inizio era necessaria una soglia critica di iniziatori!**

Ma come capire qual è la soglia critica?

Per rispondere consideriamo un terzo esempio: il gioco delle due urne.

### Il gioco delle due urne
C'è uno scommettitore in una stanza chiusa, e un insieme di giocatori che attendono fuori dalla stanza. Lo scommettitore mette 2 palline rosse e 1 blu in un'urna (**MR**) e 1 pallina rossa e 2 blu in un'altra urna (**MB**), poi sceglie una delle due urne u.a.r.

Le regole del gioco sono le seguenti:
- un giocatore alla volta entra nella stanza, estrae una palla dall'urna (che può essere rossa o blu) e la reinserisce nell'urna
- il giocatore comunica poi allo scommettitore e a tutti gli altri giocatori se, sulla base delle informazioni in suo possesso, ritiene che si tratti dell'urna MR o MB. Tuttavia il giocatore **non può comunicare l'esito dell'estrazione**, pena sconfitta di tutti i giocatori.
- al termine del gioco, solo i giocatori che hanno indovinato l'urna vincono 1kk euro (quindi l'informazione che il giocatore comunica non è a caso... è ragionata al fine di massimizzare le probabilità di vincita).

Vediamo ora di capire come rispondono i giocatori, e in base a quali motivazioni prenderanno le loro decisioni:
1. **Il Giocatore 1** prima di estrarre la pallina sa solo che le urne sono equiprobabili. Dopo aver estratto una palla, la probabilità che l'urna sia di un certo tipo viene modificata: se estrae r allora il giocatore penserà che è più probabile che l'urna sia MR, mentre se estrae b allora penserà che è più probabile che l'urna sia MB --> ipotizzando che abbia estratto b, comunica agli altri giocatori che l'urna è MB.
2. **Il Giocatore 2** prima di estrarre la pallina deduce deterministicamente ciò che ha estratto il giocatore 1 (b) --> si trova nella stessa situazione in cui si trovava il giocatore 1 prima dell'estrazione.  
Se pesca una pallina r allora dirà MR e sarà come se il gioco stesse iniziando dal giocatore 2, non consideriamo questo caso.  
Assumiamo cioè che abbia estratto una pallina blu: allora sa che sono estratte 2 b rafforzando l'ipotesi che l'urna sia MB, e quindi comunica MB
3. **Il Giocatore 3** sa deterministicamente che i primi due hanno estratto b, perché sa bene il ragionamento egoistico che seguono.  
A questo punto quindi, indipendentemente da ciò che estrae, al giocatore 3 converrà comunicare MB perché in ogni caso la sua conoscenza implica maggiore probabilità che l'urna sia MB: se pesca b a maggior ragione, ma anche se pesca r ha due su tre b... sembra che sia più probabile MB... quindi risponde con MB indipendentemente

A questo punto il giocatore 4 seguendo lo stesso ragionamento di 3 sa che le prime due estrazioni erano blu, ma non può dedurre niente circa l'estrazione di 3 perché sa che la sua risposta conveniva deterministicamente a MB indipendentemente da ciò che avesse estratto!  
Quindi il giocatore 4 ha le stesse info del giocatore 3 --> *non può che comportarsi come il giocatore 3* rispondendo MB indipendentemente da ciò che estrae...

Il ragionamento si ripete all'infinito, per ogni giocatore $x$ con $x \ge 3$... si è innestato il fenomeno di **cascata imitativa**.

Per capire i fondamenti razionali dell'intuizione dei giocatori ci sporchiamo le mani con un po' di calcolo bayesiano. Il teorema di Bayes afferma che, dati due eventi $A$ e $B$:
$$\Pr(A|B) = \frac{\Pr(B|A)\Pr(A)}{\Pr(B)}$$
e poiché
$$\Pr(B) = \Pr((B \cap A) \cup (B \cap A^c)) = \Pr(B \cap A) + \Pr(B \cap A^c) = \Pr(B|A)\Pr(A) + \Pr(B|A^c)\Pr(A^c)$$
si ha 
$$\Pr(A|B) = \frac{\Pr(B|A)\Pr(A)}{\Pr(B|A)\Pr(A) + \Pr(B|A^c)\Pr(A^c)}$$
Sia ora MR l'evento "l'urna è a maggioranza rossa", MB l'evento "l'urna è a maggioranza blu", $r$ l'evento "viene estratta una pallina rossa" e $b$ l'evento "viene estratta una pallina blu".

In generale il gioco delle urne, il ristorante e il guardare il cielo hanno caratteristiche comuni:
1. ogni individuo deve prendere una decisione
2. ogni individuo ha un'**informazione privata a disposizione** (nel caso delle urne l'estrazione della pallina, nel caso del ristorante la propria guida...)
3. ogni individuo riceve dalla rete un'**informazione incompleta** (cosa hanno deciso gli altri individui, *non il perché lo abbiano deciso*)
4. le decisioni vengono prese **sequenzialmente**: ogni individuo prende la decisione solo *dopo* aver osservato il comportamento degli altri individui
5. ogni individuo prende la decisione su base puramente razionale ed egoistica (non agisce sulla base di una pressione sociale a uniformarsi, come avviene nell'omofilia...)
6. la cascata imitativa si innesca solo quando una **massa critica** di individui ha preso la medesima decisione

### Un modello generale di decision making sequenziale
Definiamo ora un modello generale di decision making sequenziale generalizzando le caratteristiche appena evidenziate:

**1-** ogni individuo deve prendere una decisione: o **accettare (Y)** oppure **rifiutare (N)** una certa proposta.  

Solo una delle due scelte è giusta, e la probabilità che sia corretto accettare la proposta è $\Pr(Y) = p$ e quindi la probabilità che sia invece corretto non accettarla è $\Pr(N) = 1 - p$.

Se un individuo **accetta** la proposta può avere un profitto oppure una perdita: se accettare è giusto --> ha un profitto $v_g \gt 0$, se è sbagliato profitto negativo $v_b \le 0$.  
Se un individuo **non accetta la proposta** --> non ha alcun profitto né alcuna perdita

Affinché sia equivalente per un individuo accettare o rifiutare la proposta in assenza di info che permettano di guadagnare evidenza in favore di una delle alternative, il valore atteso del profitto in caso di accettazione $\mathbb{E}[B_Y]$ deve essere pari a quello del profitto in caso di non accettazione $\mathbb{E}[B_N]$ (che è pari a 0):
$$\mathbb{E}[B_Y] = v_g p + v_b (1 - p) = 0 = \mathbb{E}[B_N]$$
(i.e. in assenza di info che permettano di guadagnare evidenza a favore di una delle due alternative, per un individuo è indifferente accettare o rifiutare la proposta).

**2-** **ogni individuo possiede un'informazione privata**, che riceve nella forma di un segnale privato e può avere due valori: **A (accept)** o **R (reject)**.  

Se la scelta giusta è accettare --> **la prob. di ricevere $A$ è $q \gt \frac{1}{2}$** e quindi la probabilità di ricevere $R$ è $1 - q \lt \frac{1}{2}$.  
Simmetricamente se la scelta giusta è rifiutare --> **la prob. di ricevere $R$ è $q \gt \frac{1}{2}$** e quindi la probabilità di ricevere $A$ è $1 - q \lt \frac{1}{2}$. Formalmente:
$$\Pr(A|Y) = \Pr(R|N) = q > \frac{1}{2} \quad \text{e} \quad \Pr(R|Y) = \Pr(A|N) = 1 - q < \frac{1}{2}$$
**3-** ogni individuo riceve dalla rete solo un'informazione incompleta: sa *cosa* hanno deciso gli altri, ma non *perché* lo hanno deciso...

**4-** le decisioni vengono prese **sequenzialmente**: ogni individuo prende la decisione solo *dopo* aver osservato il comportamento degli altri individui

**5-** ogni individuo prende la decisione su base puramente razionale ed egoistica, ossia inferisce quale sia la decisione che, sulla base delle osservazioni dell'ambiente e del comportamento degli altri sembra esser quella che gli porterà benefici maggiori.

Se dopo aver ricevuto il proprio segnale privato e aver osservato le scelte degli altri la probabilità che la scelta giusta sia accettare è diventata $p'$ --> **l'individuo accetta la proposta se e solo se**:
$$v_g p' + v_b (1 - p') \ge 0$$
e poiché $v_g p + v_b (1 - p) = 0$, ossia $v_b = - \frac{p}{1 - p} v_g \le 0$ --> questo accade se e solo se $p' \ge p$ ($v_b$ è negativo, quindi il termine $v_b (1 - p')$ è negativo e quindi per avere un profitto atteso positivo bisogna che $p'$ sia maggiore o uguale a $p$..)

**6-** la cascata imitativa si innesca solo quando una **massa critica** di individui ha preso la medesima decisione. Andiamo adesso a studiare questa massa critica

---

Cominciamo analizzando come varia la probabilità che $Y$ sia la scelta migliore nel caso in cui venga ricevuta una sequenza $S$ di segnali.

Se $|S| = 1$ (solo un segnale) allora:
- se il segnale era $A$:
    $$\Pr(Y|S) = \frac{\Pr(A|Y) \Pr(Y)}{\Pr(A|Y) \Pr(Y) + \Pr(A|N) \Pr(N)} = \frac{qp}{qp + (1-q)(1-p)} \gt \frac{qp}{qp + q(1-p)} = p$$
    e $\Pr(N|A) = 1 - \Pr(Y|A) \lt 1 - p$
    (dove l'ultima disuguaglianza è vera perché $q \gt 1 - q$ in quanto $q \gt \frac{1}{2}$ e quindi il denominatore della frazione di sinistra è minore di quello della frazione di destra.)
- simmetricamente $R$:
    $$\Pr(Y|R) = \frac{\Pr(R|Y) \Pr(Y)}{\Pr(R|Y) \Pr(Y) + \Pr(R|N) \Pr(N)} = \frac{(1-q)p}{(1-q)p + q(1-p)} \lt \frac{(1-q)p}{(1-q)p + (1-q)(1-p)} = p$$
    e $\Pr(N|R) = 1 - \Pr(Y|R) \gt 1 - p$
    (diventa $p$ perché al denominatore si raccoglie $(1-q)$ e quindi il denominatore diventa $p + (1-p) = 1$)

Come era logico pensare quindi, quando un individuo riceve un segnale e non ha altre informazioni derivanti dal comportamento di altri individui, segue semplicemente il segnale privato in questione: sceglie $Y$ se riceve $A$ e sceglie $N$ se riceve $R$.

Ora vediamo che succede con una sequenza di segnali $|S| \gt 1$, in particolare $a$ segnali $A$ e $r$ segnali $R$:
$$\Pr(Y|S) = \frac{\Pr(S|Y) \Pr(Y)}{\Pr(S|Y) \Pr(Y) + \Pr(S|N) \Pr(N)} = \frac{q^a (1-q)^r p}{q^a (1-q)^r p + (1-q)^a q^r (1-p)}$$
dove $\Pr(S|Y) = q^a (1-q)^r$ come è logico pensare, dato che si tratta di segnali indipendenti. 

Consideriamo il secondo addendo al denominatore $(1-q)^a q^r (1-p)$: 
- se $a \gt r$ allora
  $$(1-q)^a q^r = (1-q)^{a-r} (1-q)^r q^r \lt q^{a-r} (1-q)^r q^r = q^{a} (1-q)^r$$
  perché sempre $q \gt \frac{1}{2}$ e quindi $q \gt 1 - q$ e quindi $(1-q)^{a-r} \lt q^{a-r}$, dato che $a \gt r$.  
  Ma quindi:
  $$\Pr(Y|S) = \frac{q^a (1-q)^r p}{q^a (1-q)^r p + (1-q)^a q^r (1-p)} \gt \frac{q^a (1-q)^r p}{q^a (1-q)^r p + q^{a} (1-q)^r (1-p)} = \frac{p}{p + 1 - p} = p$$
- se $a \lt r$ allora
  $$(1-q)^a q^r = (1-q)^a q^a q^{r-a} \gt (1-q)^a q^a (1-q)^{r-a} = (1-q)^r q^a$$
  e quindi 
  $$\Pr(Y|S) = \frac{q^a (1-q)^r p}{q^a (1-q)^r p + (1-q)^a q^r (1-p)} \lt \frac{q^a (1-q)^r p}{q^a (1-q)^r p + q^{a} (1-q)^r (1-p)} = \frac{p}{p + 1 - p} = p$$
- se $a = r$ --> $(1-q)^a q^r = q^a (1-q)^r$ e quindi è facile vedere che $\Pr(Y|S) = p$.

Ma quindi di nuovo, come era facile intuire, se viene ricevuta una sequenza $S$ di segnali che contiene $a$ segnali $A$ e $r$ segnali $R$, allora:
- se $a \gt r$ allora $\Pr(Y|S) \gt p$ e quindi l'individuo accetta la proposta (perché $p$ era tale che l'individuo era indifferente tra accettare o rifiutare la proposta)
- se $a \lt r$ allora $\Pr(Y|S) \lt p$ e quindi l'individuo rifiuta la proposta
- se $a = r$ allora $\Pr(Y|S) = p$ e quindi l'individuo è indifferente tra accettare o rifiutare la proposta.

Ma quindi è evidente che se un individuo conoscesse l'intera sequenza di segnali privati per ognuno, allora saprebbe prendere la decisione che massimizza la sua probabilità di successo.

Torniamo però alle regole del gioco: ogni individuo prende la decisione solo sequenzialmente (uno dopo l'altro) e sulla base del proprio segnale privato e delle decisioni prese dai loro predecessori in sequenza (senza conoscerne i segnali privati!)

Sulla base di quello che abbiamo visto a proposito di $\Pr(Y|S)$ (vedi esempio urna) sappiamo per certo che:
- il primo individuo della sequenza segue solo il proprio segnale privato, perché non ha altre informazioni
- anche il secondo individuo segue il proprio segnale privato
- il terzo segue il proprio segnale privato **solo se i primi due individui hanno preso decisioni differenti** (in tal caso come avevamo visto è come se "ricominciasse" il gioco, le probabilità tornano quelle iniziali).  
Se invece i due individui hanno preso la stessa decisione anche il terzo prende quella decisione, **indipendentemente dal proprio segnale privato**
- etc...

Abbiamo quindi capito come il segnale privato possa essere inferito solo se i primi individui hanno preso decisioni grossomodo alternate, altrimenti si innesca la cascata imitativa.

Consideriamo quindi l'individuo $n$ nella sequenza e **supponiamo che egli riesca ad inferire i segnali privati dei suoi predecessori**: come detto questo si verifica solo quando *la cascata imitativa non si è ancora innescata*, ossia le decisioni fra $Y$ e $N$ fino all'individuo $n$ si sono "grosso modo" alternate.  
Cerchiamo di chiarire matematicamente questo "grosso modo"...

Supponiamo che l'individuo $n$ riesca quindi a inferire i segnali privati dei predecessori, ossia sa che è stata ricevuta una sequenza $S$ di segnali che contiene $a$ segnali $A$ e $r$ segnali $R$, e sia $\sigma_n$ il segnale privato dell'individuo $n$: 
- Se $a = r$ allora $\Pr(Y|S) = p$ --> sceglie in base al segnale privato: $\Pr(Y|S \cup \sigma_n) \gt p$ se $\sigma_n = A$ e $\Pr(Y|S \cup \sigma_n) \lt p$ se $\sigma_n = R$.
- Se $a = r + 1$ allora $\Pr(Y|S \cup {\sigma_n}) \gt p$ se $\sigma_n = A$ e $\Pr(Y|S \cup {\sigma_n}) = p$ se $\sigma_n = R$: anche in questo caso l'individuo segue il suo segnale privato qualunque esso sia (equivalente al caso del giocatore 2). Simmetrico il caso $r = a + 1$.
- Se $a \ge r + 2$ allora $\Pr(Y|S \cup {\sigma_n}) \gt p$ qualunque sia $\sigma_n$ e quindi l'individuo accetta la proposta indipendentemente dal proprio segnale privato, si è innescata la cascata imitativa!  
Il caso $r \ge a + 2$ è simmetrico (rifiuta indipendentemente dal proprio segnale privato).

Ma che significa $a \ge r + 2$? In quali casi si verifica?  
Fino a quando i segnali si alternano, (ARARARA...) vale $a \le r + 1$ e $r \le a + 1$ i.e. $|a - r| \le 1$ --> per superare la soglia "+1" è necessario che il segnale $A$ venga ricevuto almento due volte di seguito --> ARARARAA --> solo a questo punto l'individuo successivo alla doppia A sceglie $Y$ indipendentemente dal proprio segnale privato...

Quindi **la cascata imitativa non si innesca fino a quando non vengono ricevuti due segnali consecutivi uguali**, ma attenzione... è una condizione **necessaria**... in generale ricevere due segnali consecutivi uguali non è sufficiente affinché si inneschi la cascata: ad esempio nella sequenza RARARARAA vale $a-r = 1$ --> non è sufficiente perché l'individuo successivo scelga $Y$ indipendentemente dal suo segnale privato.

Invece, **certamente la cascata imitativa si innesca non appena vengono ricevuti tre segnali consecutivi uguali**: questa è una condizione **sufficiente**! (ma non necessaria, perché ad esempio nella sequenza ARARARAA la cascata si scatena pure senza tre lettere consecutive...)

Usiamo ciò per calcolare la probabilità di occorrenza di una cascata imitativa.

Sia $H_n$ l'evento "la cascata imitativa si innesca entro il giocatore $n$":
$$\Pr(H_n) \ge \Pr(\bigcup_{i=1}^{n-2} (\sigma_i = \sigma_{i+1} = \sigma_{i+2})) \ge \Pr(\bigcup_{i=1}^{\frac{n}{3}} (\sigma_{3i-2} = \sigma_{3i-1} = \sigma_{3i}))$$
dove l'ultima disuguaglianza deriva dal fatto che calcolare ad es. $\Pr((\sigma_1 = \sigma_2 = \sigma_3) \cup (\sigma_2 = \sigma_3 = \sigma_4) \cup (\sigma_3 = \sigma_4 = \sigma_5) \cup (\sigma_4 = \sigma_5 = \sigma_6))$ è difficile perché gli eventi non sono disgiunti e noi volendo calcolare la probabilità di cascata vogliamo fare una stima inferiore (quindi la classica maggiorazione con union bound non va bene..).  
Pertanto si fa minorazione considerando solo le triple di segnali consecutivi disgiunte: 
$$\Pr((\sigma_1 = \sigma_2 = \sigma_3) \cup (\sigma_2 = \sigma_3 = \sigma_4) \cup (\sigma_3 = \sigma_4 = \sigma_5) \cup (\sigma_4 = \sigma_5 = \sigma_6)) \ge \Pr((\sigma_1 = \sigma_2 = \sigma_3) \cup (\sigma_4 = \sigma_5 = \sigma_6))$$

Da ciò otteniamo che:
$$\Pr(\lnot H_n) \le \Pr(\bigcap_{i=1}^{\frac{n}{3}} \lnot (\sigma_{3i-2} = \sigma_{3i-1} = \sigma_{3i})) = \prod_{i=1}^{\frac{n}{3}} \Pr(\lnot (\sigma_{3i-2} = \sigma_{3i-1} = \sigma_{3i}))$$
(perché eventi indipendenti)

Ma quindi 
$$\Pr(\lnot H_n) \le (3q - 3q^2)^{\frac{n}{3}}$$
e pertanto
$$\Pr(H_n) \ge 1 - (3q - 3q^2)^{\frac{n}{3}} \quad \text{e quindi} \quad \lim_{n \to \infty} \Pr(H_n) = 1$$
dove l'ultimo limite è valido perché $0 \lt 3q - 3q^2 \lt 1$ per $0 \lt q \lt 1$, e quindi abbiamo dimostrato che **la cascata imitativa si innesca quasi sicuramente.**

### Considerazioni finali
Le cascate imitative sono fenomeni facili da innescare, in quanto richiedono la condivisione di poche informazioni.  
Infatti, le sole info che precedono la sua insorgenza sono quelle che hanno rilevanza in questo fenomeno: tutte quelle che vengono ricevute dopo che la cascata si è innescata sono inutili --> se una cascata si innesca presto, la maggior parte delle informazioni che una popolazione acquisisce restano inutilizzate.

Questo fenomeno può indurre decisioni sbagliate, proprio in virtù del fatto che non viene usata tutta l'informazione disponibile della rete.  

Una cascata imitativa però è per fortuna facile da rompere: purché ciò avvenga è sufficiente che qualcuno diffonda il proprio segnale privato, e questa eventualità è tanto più probabile quanto è più grande la popolazione interessata al fenomeno.

Dal momento che una cascata imitativa può indurre decisioni sbagliate, è chiaro che non sia sempre una buona idea imitare il comportamento della massa...  

Nel libro *The wisdom of crowd* del 2004, James Surowiecki sostiene la tesi per cui "il comportamento aggregato di un numero elevato di persone in possesso di informazione molto limitata possa produrre risultati molto accurati".  
L'assunzione alla base di questa affermazione è che gli individui, ciascuno in possesso di un'info privata, operano le loro scelte **indipendentemente** gli uni dagli altri, senza che nessuno conosca le scelte operate dagli altri.  

Ciò quindi non contrasta la nostra discussione dove, invece, ogni individuo *osserva i comportamenti degli altri e li imita*.
