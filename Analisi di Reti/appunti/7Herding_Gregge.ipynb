Iniziamo una serie di lezioni, la parte finale del corso, dove studiamo la rete come fonte di informazione.  
Studieremo cioè il contenuto informativo di una rete:
- che può indurre gli individui che la compongono a modificare il proprio comportamento (in questa lezione).
- che deve esser sintetizzato per derivare un'informazione cumulativa che tenga conto, in qualche modo, di pezzi di informazione derivanti dai singoli individui (nei sistemi di voto).
- dal quale deve essere individuato ed estratto quello rilevante ad una data richiesta (nell'ultima serie di lezioni, web search).
  
Parlando del primo caso, abbiamo già visto con i processi di diffusione come l'informazione presente in una rete può indurre gli individui che la compongono a modificare il proprio comportamento, ma ora **il punto di vista è diverso**.  

Mentre nei processi di diffusione un individuo cambiava comportamento 
1. come conseguenza di **comunicazioni esplicite** di una serie di info da parte degli individui con i quali è in relazione (es. semi ibridi di mais)
2. per ottenere una serie di benefici dalla presenza della rete (es. scambiarsi file con lo stesso SO)

ora invece studiamo il fenomeno in virtù del quale un individuo modifica il proprio comportamento semplicemente osservando quello che **globalmente** hanno gli altri individui della rete, in assenza di **comunicazione diretta** --> consideriamo una visione globale e non puntuale della rete, non ci interessa più l'analisi delle singole relazioni personali.

Vediamo un esempio: stiamo visitando da turisti una città sconosciuta, seguendo tutti i consigli di una nostra guida privata. A cena andiamo al bel ristorante consigliato dalla guida, ma notiamo che il locale è vuoto. Davanti al ristorante c'è una bettola di locale, e prima di entrare nel nostro notiamo che tanta gente entra nella bettolaccia...  
tendenzialmente allora una persona deciderà di seguire la folla nella bettola, perché intuitivamente si assume che se tanta gente va da quella parte un motivo, **anche se a noi sconosciuto**, ci deve pur essere!

Questa informazione la presumiamo noi, non ci viene detta esplicitamente ma la deduciamo dal loro comportamento, è una **informazione che ricaviamo dalla rete**.

Un esempio ancora più estremo riguarda sempre Milgram, negli anni '60, che ha piazzato gente in giro per strada e li ha fatti guardare in alto. Negli incroci dove c'era una sola persona a guardare il cielo, quasi nessun passante imitava il comportamento. Agli incroci con 5 persone, solo qualche passante imitava, mentre in quelli con 15 persone allora oltre il 45% dei passanti imitava il comportamento.

Il motivo anche qui è abbastanza chiaro: i passanti inferivano l'esistenza di una motivazione razionale che inducesse le persone a guardare in alto --> doveva esserci qualcosa di interessante, gli sarebbe convenuto anche a loro provare a guardare su!  

Naturalmente perché ciò avvenisse era necessario che ci fosse un numero minimo di persone che guardavano in alto, altrimenti si penserebbe semplicemente che quelle poche persone avessero un disagio mentale...  
Più formalmente stiamo dicendo che **perché il fenomeno di imitazione avesse inizio era necessaria una soglia critica di iniziatori!**

Ma come capire qual è la soglia critica?

Per rispondere consideriamo un terzo esempio: il gioco delle due urne.

### Il gioco delle due urne
C'è uno scommettitore in una stanza chiusa, e un insieme di giocatori che attendono fuori dalla stanza. Lo scommettitore mette 2 palline rosse e 1 blu in un'urna (**MR**) e 1 pallina rossa e 2 blu in un'altra urna (**MB**), poi sceglie una delle due urne u.a.r.

Le regole del gioco sono le seguenti:
- un giocatore alla volta entra nella stanza, estrae una palla dall'urna (che può essere rossa o blu) e la reinserisce nell'urna
- il giocatore comunica poi allo scommettitore e a tutti gli altri giocatori se, sulla base delle informazioni in suo possesso, ritiene che si tratti dell'urna MR o MB. Tuttavia il giocatore **non può comunicare l'esito dell'estrazione**, pena sconfitta di tutti i giocatori.
- al termine del gioco, solo i giocatori che hanno indovinato l'urna vincono 1kk euro (quindi l'informazione che il giocatore comunica non è a caso... è ragionata al fine di massimizzare le probabilità di vincita).

Vediamo ora di capire come rispondono i giocatori, e in base a quali motivazioni prenderanno le loro decisioni:
1. **Il Giocatore 1** prima di estrarre la pallina sa solo che le urne sono equiprobabili. Dopo aver estratto una palla, la probabilità che l'urna sia di un certo tipo viene modificata: se estrae r allora il giocatore penserà che è più probabile che l'urna sia MR, mentre se estrae b allora penserà che è più probabile che l'urna sia MB --> ipotizzando che abbia estratto b, comunica agli altri giocatori che l'urna è MB.
2. **Il Giocatore 2** prima di estrarre la pallina deduce deterministicamente ciò che ha estratto il giocatore 1 (b) --> si trova nella stessa situazione in cui si trovava il giocatore 1 prima dell'estrazione.  
Se pesca una pallina r allora dirà MR e sarà come se il gioco stesse iniziando dal giocatore 2, non consideriamo questo caso.  
Assumiamo cioè che abbia estratto una pallina blu: allora sa che sono estratte 2 b rafforzando l'ipotesi che l'urna sia MB, e quindi comunica MB
3. **Il Giocatore 3** sa deterministicamente che i primi due hanno estratto b, perché sa bene il ragionamento egoistico che seguono.  
A questo punto quindi, indipendentemente da ciò che estrae, al giocatore 3 converrà comunicare MB perché in ogni caso la sua conoscenza implica maggiore probabilità che l'urna sia MB: se pesca b a maggior ragione, ma anche se pesca r ha due su tre b... sembra che sia più probabile MB... quindi risponde con MB indipendentemente

A questo punto il giocatore 4 seguendo lo stesso ragionamento di 3 sa che le prime due estrazioni erano blu, ma non può dedurre niente circa l'estrazione di 3 perché sa che la sua risposta conveniva deterministicamente a MB indipendentemente da ciò che avesse estratto!  
Quindi il giocatore 4 ha le stesse info del giocatore 3 --> *non può che comportarsi come il giocatore 3* rispondendo MB indipendentemente da ciò che estrae...

Il ragionamento si ripete all'infinito, per ogni giocatore $x$ con $x \ge 3$... si è innestato il fenomeno di **cascata imitativa**.

Per capire i fondamenti razionali dell'intuizione dei giocatori ci sporchiamo le mani con un po' di calcolo bayesiano. Il teorema di Bayes afferma che, dati due eventi $A$ e $B$:
$$\Pr(A|B) = \frac{\Pr(B|A)\Pr(A)}{\Pr(B)}$$
e poiché
$$\Pr(B) = \Pr((B \cap A) \cup (B \cap A^c)) = \Pr(B \cap A) + \Pr(B \cap A^c) = \Pr(B|A)\Pr(A) + \Pr(B|A^c)\Pr(A^c)$$
si ha 
$$\Pr(A|B) = \frac{\Pr(B|A)\Pr(A)}{\Pr(B|A)\Pr(A) + \Pr(B|A^c)\Pr(A^c)}$$
Sia ora MR l'evento "l'urna è a maggioranza rossa", MB l'evento "l'urna è a maggioranza blu", $r$ l'evento "viene estratta una pallina rossa" e $b$ l'evento "viene estratta una pallina blu".

In generale il gioco delle urne, il ristorante e il guardare il cielo hanno caratteristiche comuni:
1. ogni individuo deve prendere una decisione
2. ogni individuo ha un'**informazione privata a disposizione** (nel caso delle urne l'estrazione della pallina, nel caso del ristorante la propria guida...)
3. ogni individuo riceve dalla rete un'**informazione incompleta** (cosa hanno deciso gli altri individui, *non il perché lo abbiano deciso*)
4. le decisioni vengono prese **sequenzialmente**: ogni individuo prende la decisione solo *dopo* aver osservato il comportamento degli altri individui
5. ogni individuo prende la decisione su base puramente razionale ed egoistica (non agisce sulla base di una pressione sociale a uniformarsi, come avviene nell'omofilia...)
6. la cascata imitativa si innesca solo quando una **massa critica** di individui ha preso la medesima decisione
